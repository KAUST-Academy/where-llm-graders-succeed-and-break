```python
# Write your code here
import torch.nn as nn
import torch.nn.functional as F

class PotatoCNN(nn.Module):
    def __init__(self):
        super(PotatoCNN, self).__init__()

        #Layer 1
        self.conv1 = nn.Conv2d(3, 16, kernel_size=3, padding=1)
        self.bn1 = nn.BatchNorm2d(16)

        #Layer2
        self.conv2 = nn.Conv2d(16, 32, kernel_size=3, padding=1)
        self.bn2 = nn.BatchNorm2d(32)
        self.pool1 = nn.MaxPool2d(2, 2)

        #Layer 3
        self.conv3 = nn.Conv2d(32, 64, kernel_size=3, padding=1)
        self.bn3 = nn.BatchNorm2d(64)

        #Layer 4
        self.conv4 = nn.Conv2d(64, 128, kernel_size=3, padding=1)
        self.bn4 = nn.BatchNorm2d(128)
        self.pool2 = nn.MaxPool2d(2, 2)

        #Layer 5
        self.conv5 = nn.Conv2d(128, 256, kernel_size=3, padding=1)
        self.bn5 = nn.BatchNorm2d(256)

        #Fully Connected Layers
        self.fc1 = nn.Linear(256 * 8 * 8, 512)
        self.dropout = nn.Dropout(0.5)
        self.fc2 = nn.Linear(512, 3) #3 categories

    def forward(self, x):
        #Apply Layers 1 & 2
        x = F.relu(self.bn1(self.conv1(x)))
        x = self.pool1(F.relu(self.bn2(self.conv2(x))))

        #Apply Layers 3 & 4
        x = F.relu(self.bn3(self.conv3(x)))
        x = self.pool2(F.relu(self.bn4(self.conv4(x))))

        #Apply Layer 5
        x = F.relu(self.bn5(self.conv5(x)))

        #Flatten and Classify
        x = x.view(-1, 256 * 8 * 8)
        x = F.relu(self.fc1(x))
        x = self.dropout(x)
        x = self.fc2(x)

        return x

model = PotatoCNN()
print(model)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import matplotlib.pyplot as plt

#1
class ResidualPotatoCNN(nn.Module):
    def __init__(self):
        super(ResidualPotatoCNN, self).__init__()

        #layer 1 & 2
        self.conv1 = nn.Conv2d(3, 16, kernel_size=3, padding=1)
        self.bn1 = nn.BatchNorm2d(16)
        self.conv2 = nn.Conv2d(16, 32, kernel_size=3, padding=1)
        self.bn2 = nn.BatchNorm2d(32)
        self.pool1 = nn.MaxPool2d(2, 2) # Size becomes 16x16

        #layer 3 & 4
        self.conv3 = nn.Conv2d(32, 64, kernel_size=3, padding=1)
        self.bn3 = nn.BatchNorm2d(64)
        self.conv4 = nn.Conv2d(64, 128, kernel_size=3, padding=1)
        self.bn4 = nn.BatchNorm2d(128)
        self.pool2 = nn.MaxPool2d(2, 2) # Size becomes 8x8

        self.skip_conv = nn.Sequential(
            nn.Conv2d(32, 128, kernel_size=1, stride=2),
            nn.BatchNorm2d(128)
        )

        #layer 5
        self.conv5 = nn.Conv2d(128, 256, kernel_size=3, padding=1)
        self.bn5 = nn.BatchNorm2d(256)

        self.fc1 = nn.Linear(256 * 8 * 8, 512)
        self.dropout = nn.Dropout(0.5)
        self.fc2 = nn.Linear(512, 3)

    def forward(self, x):
        #layer 1 & 2
        x = F.relu(self.bn1(self.conv1(x)))
        x = self.pool1(F.relu(self.bn2(self.conv2(x))))

        identity = x

        #layer 3 & 4
        x = F.relu(self.bn3(self.conv3(x)))
        x = self.pool2(F.relu(self.bn4(self.conv4(x))))

        x += self.skip_conv(identity)
        x = F.relu(x)

        #layer 5
        x = F.relu(self.bn5(self.conv5(x)))

        x = x.view(-1, 256 * 8 * 8)
        x = F.relu(self.fc1(x))
        x = self.dropout(x)
        x = self.fc2(x)
        return x

#2
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model_res = ResidualPotatoCNN().to(device)
criterion = nn.CrossEntropyLoss()
optimizer_res = optim.Adam(model_res.parameters(), lr=0.001)

res_train_losses, res_val_losses = [], []
res_train_accs, res_val_accs = [], []
num_epochs = 20

#3
for epoch in range(num_epochs):
    #training
    model_res.train()
    t_loss, t_correct, t_total = 0, 0, 0
    for imgs, lbls in train_loader:
        imgs, lbls = imgs.to(device), lbls.to(device)
        optimizer_res.zero_grad()
        out = model_res(imgs)
        loss = criterion(out, lbls)
        loss.backward()
        optimizer_res.step()

        t_loss += loss.item()
        _, pred = torch.max(out, 1)
        t_total += lbls.size(0)
        t_correct += (pred == lbls).sum().item()

    #validation
    model_res.eval()
    v_loss, v_correct, v_total = 0, 0, 0
    with torch.no_grad():
        for imgs, lbls in test_loader:
            imgs, lbls = imgs.to(device), lbls.to(device)
            out = model_res(imgs)
            loss = criterion(out, lbls)
            v_loss += loss.item()
            _, pred = torch.max(out, 1)
            v_total += lbls.size(0)
            v_correct += (pred == lbls).sum().item()

    res_train_losses.append(t_loss/len(train_loader))
    res_val_losses.append(v_loss/len(test_loader))
    res_train_accs.append(100 * t_correct/t_total)
    res_val_accs.append(100 * v_correct/v_total)

    print(f"Epoch {epoch+1}/{num_epochs} | Train Acc: {res_train_accs[-1]:.2f}% | Val Acc: {res_val_accs[-1]:.2f}%")

#4
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(res_train_losses, label='Train Loss')
plt.plot(res_val_losses, label='Val Loss')
plt.title('Residual Model: Loss History')
plt.xlabel('Epochs')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(res_train_accs, label='Train Acc')
plt.plot(res_val_accs, label='Val Acc')
plt.title('Residual Model: Accuracy History')
plt.xlabel('Epochs')
plt.legend()

plt.tight_layout()
plt.show()
```
