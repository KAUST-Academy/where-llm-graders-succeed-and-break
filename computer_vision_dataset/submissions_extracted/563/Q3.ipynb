```python
# TO DO
import os
import pandas as pd
from PIL import Image
import torch
import torchvision.transforms as transforms
from torch.utils.data import Dataset
import numpy as np
import glob
# Custom Dataset Class

class UnderwaterImagery(Dataset):
    def __init__(self, image_dir, mask_dir, transform=None, target_transform=None):
        self.image_dir = image_dir
        self.mask_dir = mask_dir
        self.transform = transform
        self.target_transform = target_transform

        # Support multiple extensions
        self.image_paths = sorted(
            glob.glob(os.path.join(image_dir, "*.png")) +
            glob.glob(os.path.join(image_dir, "*.jpg")) +
            glob.glob(os.path.join(image_dir, "*.jpeg"))
        )

        self.mask_paths = sorted(
            glob.glob(os.path.join(mask_dir, "*.png")) +
            glob.glob(os.path.join(mask_dir, "*.jpg")) +
            glob.glob(os.path.join(mask_dir, "*.jpeg"))
        )

        print(f"Found {len(self.image_paths)} images")
        print(f"Found {len(self.mask_paths)} masks")

        if len(self.image_paths) == 0:
            raise ValueError("No images found. Check image_dir path.")

        if len(self.image_paths) != len(self.mask_paths):
            raise ValueError("Number of images and masks do not match.")

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        image = Image.open(self.image_paths[idx]).convert("RGB")
        mask = Image.open(self.mask_paths[idx])

        if self.transform:
            image = self.transform(image)

        if self.target_transform:
            mask = self.target_transform(mask)

        mask = torch.squeeze(mask).long()

        return image, mask
```
