```python
# Task 4: Write your code here: Apply feature scaling to numerical features (Use StandardScaler)

# Answer: Based on the describe table above most values looks to be already transformed!!
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

dataset = scaler.fit_transform(q3_dataset.drop(["Target"], axis=1))
```

```python
# Task 1: Write your code here:
X_features = dataset
Y_target = q3_dataset["Target"]

# Task 1: Write your code here:
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(dataset, Y_target, test_size=0.2)
```

```python
# Task 2,3,4,5: Write your code here:
#%pip install catboost
from catboost import CatBoostClassifier

# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import KFold
from sklearn.metrics import accuracy_score, f1_score
import numpy as np

cat = CatBoostClassifier(iterations= 100, learning_rate= 0.001)
kf = KFold(n_splits=5, shuffle=True, random_state=30)

acc_scores = []
f1_scores = []

for train_idx, val_idx in kf.split(X_train):
    X_fold_train, X_fold_val = X_train[train_idx], X_train[val_idx]
    y_fold_train, y_fold_val = Y_train.iloc[train_idx], Y_train.iloc[val_idx]

    # Train and predict
    cat.fit(X_fold_train, y_fold_train)
    y_fold_pred = cat.predict(X_fold_val)

    # Calculate metrics
    acc_scores.append(accuracy_score(y_fold_val, y_fold_pred))
    f1_scores.append(np.sqrt(accuracy_score(y_fold_val, y_fold_pred)))

acc_scores = np.array(acc_scores)
f1_scores = np.array(f1_scores)

print(f"5-Fold CV Results:")
print(f"Mean F1 across folds (Primary Metric Due to unbalance):  %{(f1_scores.mean()*100):,.2f}")
print(f"Mean Accuracy Score: %{(acc_scores.mean()*100):,.2f}")
```
