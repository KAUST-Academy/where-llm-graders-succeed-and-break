```python
# Task 1: Write your code here:
df.columns = df.columns.str.strip().str.lower()
for col in df.columns:
    if df[col].dtype == "object":
        df[col] = df[col].fillna(df[col].mode()[0])
    else:
        df[col] = df[col].fillna(df[col].median())
```

```python
# Task 2: Write your code here:
df = df.drop_duplicates()
```

```python
# Task 3: Write your code here:
df = pd.get_dummies(df, drop_first=True)
```

```python
# Task 4: Write your code here:
# حدد عمود الهدف الصحيح
target_col = "Target" if "Target" in df.columns else "target"

X = df.drop(columns=[target_col])
y = df[target_col]

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Task 1: Write your code here:
target_col = "Target" if "Target" in df.columns else "target"

X = df.drop(columns=[target_col])
y = df[target_col]
```

```python
# Task 2,3,4,5: Write your code here:
import numpy as np
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.ensemble import RandomForestClassifier

skf = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)

model = RandomForestClassifier(
    n_estimators=100,
    max_depth=12,
    random_state=42,
    n_jobs=1
)

score_name = "f1" if y.value_counts(normalize=True).max() > 0.7 else "accuracy"
scores = cross_val_score(model, X, y, cv=skf, scoring=score_name)

print("Metric:", score_name)
print("Scores:", scores)
print("Average:", scores.mean())

```

```python
# Task 1: Write your code here:
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=1)
model.fit(X, y)

importances = model.feature_importances_
idx = np.argsort(importances)[::-1]

top = 15
plt.barh(X.columns[idx[:top]][::-1], importances[idx[:top]][::-1])
plt.xlabel("Importance")
plt.title("Top Feature Importances")
plt.show()
```

```python
# Task 2: Write your code here:
golden_feature = X.columns[idx[0]]
print("Golden Feature:", golden_feature)
```

```python
# Task Bonus: Write your code here:
import numpy as np
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.ensemble import RandomForestClassifier

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

model_full = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=1)
acc_full = cross_val_score(model_full, X, y, cv=skf, scoring="accuracy").mean()

X_golden = X[[golden_feature]]

model_golden = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=1)
acc_golden = cross_val_score(model_golden, X_golden, y, cv=skf, scoring="accuracy").mean()

print("Full model accuracy:", acc_full)
print("Golden-only accuracy:", acc_golden)
```
