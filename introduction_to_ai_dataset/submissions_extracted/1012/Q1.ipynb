```python
# Task 1: Write your code here:
# Drop the 'Order_ID' column from the data
df_clean = df_foodDeliv.drop(columns=['Order_ID']).copy()
df_clean.info()
```

```python
# Task 2: Write your code here:
# Handle missing values appropriately (Hint: I guess you want to have a closer look at the columns with missing values :) )
missing_columns = ["Weather", "Traffic_Level", "Delivery_Time", "Courier_Experience_yrs", "Time_of_Day"]
categorize_Data = df_clean[missing_columns].select_dtypes(include=["object"]).columns
numbers_Data = df_clean[missing_columns].select_dtypes(include=["float64"]).columns
for column in categorize_Data:
  df_clean[column] = df_clean[column].fillna('unknown')


for column in numbers_Data:
  df_clean[column] = df_clean[column].fillna(df_clean[column].mean())

print("Missing values remaining:", df_clean.isnull().sum().sum())

```

```python
# Task 3: Write your code here:
# Check and remove duplicates if any exist
def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

print("Duplicated:", df_clean.duplicated().sum())
check_duplicates(df_clean)
print("Duplicated:", df_clean.duplicated().sum())
```

```python
# Task 4: Write your code here:
# Encode categorical variables if needed (Bonus if used One Hot Encoding)
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()
categorize_Data = df_clean.select_dtypes(include=["object"]).columns
for column in categorize_Data:
  df_clean[column] = label_encoder.fit_transform(df_clean[column])

df_clean.head()
```
