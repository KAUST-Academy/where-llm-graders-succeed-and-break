```python
# Task 1: Write your code here:
from sklearn.preprocessing import StandardScaler
import pandas as pd

print("Missing values:")
print(df.isnull().sum())

for col in df.columns:
    if df[col].isnull().sum() > 0:
        if df[col].dtype == 'object':
            df[col] = df[col].fillna(df[col].mode()[0])
        else:
            df[col] = df[col].fillna(df[col].median())

print("\nMissing values handled")

if len(categorical_cols) > 0:

    df = pd.get_dummies(df, columns=categorical_cols, drop_first=True)
    print("Categorical variables encoded.")
```

```python
# Task 2: Write your code here:
duplicates = df.duplicated().sum()
print(f"\nNumber of duplicate rows: {duplicates}")

if duplicates > 0:
    df = df.drop_duplicates()
    print("Duplicates removed successfully.")
```

```python
# Task 3: Write your code here:
categorical_cols = df.select_dtypes(include=['object']).columns
print(f"\nCategorical columns founded are: {list(categorical_cols)}")
```

```python
# Task 4: Write your code here:
scaler = StandardScaler()
X = df.drop(columns=['Target'])
y = df['Target']

X_scaled = scaler.fit_transform(X)

df_cleaned = pd.DataFrame(X_scaled, columns=X.columns)
df_cleaned['Target'] = y.values

print("\nFeature scaling done.")
```

```python
# Task 5: Write your code here:
target_counts = df['Target'].value_counts()
print("Target Class Distribution:")
print(target_counts)

imbalance_ratio = target_counts.min() / target_counts.max()
if imbalance_ratio < 0.5:
    print("State imbalanced.")
elif imbalance_ratio < 0.75:
    print("State slightly imbalanced.")
else:
    print("State is balanced.")



print(f"\nClean data shape: {df_cleaned.shape}")
```

```python
# Task 1: Write your code here:

from catboost import CatBoostClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import f1_score
import numpy as np

X = df_cleaned.drop(columns=['target'])
y = df_cleaned['target']


```

```python
# Task 2,3,4,5: Write your code here:
 # task2:

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

f1_scores = []

print("Starting training with CatBoost")

# task-3:
for fold, (train_index, test_index) in enumerate(skf.split(X, y)):

    X_train_fold, X_test_fold = X.iloc[train_index], X.iloc[test_index]
    y_train_fold, y_test_fold = y.iloc[train_index], y.iloc[test_index]

    model = CatBoostClassifier(iterations=200, learning_rate=0.1, depth=6, silent=True)

    model.fit(X_train_fold, y_train_fold)


    y_pred = model.predict(X_test_fold)

    # Task-4:
    score = f1_score(y_test_fold, y_pred)
    f1_scores.append(score)

    print(f"Fold {fold + 1} - F1 Score: {score:.4f}")

# Task-5:
average_f1 = np.mean(f1_scores)
print("\n--- Model Evaluation Summary ---")
print(f"Average F1 Score: {average_f1:.4f}")
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import f1_score

!pip install catboost
from catboost import CatBoostClassifier

X = df_cleaned.drop(columns=['Target'])
y = df_cleaned['Target']
```
