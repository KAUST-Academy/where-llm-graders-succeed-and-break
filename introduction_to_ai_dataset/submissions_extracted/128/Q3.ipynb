```python
# Task 1: Write your code here:
# Select relevant columns, we do not select 'model' (too many unique values, too sparse and will hurt the model performance)
cols = ['P_2','D_39',	'B_1',	'B_2','R_1'	,'S_3','D_41','B_3','D_42','D_43','D_137','D_138','D_139','D_140'	,'D_141','D_142','D_143','D_144','D_145','Target']
df_clean = df_gold[cols].copy()

# Drop rows where target (price) or key features are missing - can't predict without them
print(f"Before: {df_clean.shape}")
df_clean = df_clean.dropna(subset=['D_42', 'D_137','D_138'])
print(f"After dropping missing : {df_clean.shape}")
```

```python
# Task 2: Write your code here:
# Fill categorical columns with 'unknown' - missing likely means "not specified"
for col in ['D_42','D_43','D_137','D_138','D_142']:
    df_clean[col] = df_clean[col].fillna('unknown')

# Fill S_3 with mode - discrete feature, mode is most representative
df_clean['S_3'] = df_clean['S_3'].fillna(df_clean['S_3'].mode()[0])

print("Missing values remaining:", df_clean.isnull().sum().sum())
```
