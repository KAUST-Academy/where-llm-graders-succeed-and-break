```python
# Task 1: Write your code here:
#Read the dataset
data_path = os.path.join(path, 'Q1_data.csv')
df_data = pd.read_csv(data_path)
```

```python
# Task 1: Write your code here:
#Drop the 'Order_ID'
df_clean = df_data.copy()
df_clean=df_data.drop(columns=['Order_ID'])
print("After dropping:\n" ,df_clean)
```

```python
# Task 2: Write your code here:
# missing values
missing_values = df_clean.isnull().sum()
print("Missing values:\n" ,missing_values)
```

```python
# Task 3: Write your code here:
#Encode categorical variables
duplicates=df_clean.drop_duplicates(inplace=True)
duplicates
```

```python
# Task 4: Write your code here:
# scaling
from sklearn.preprocessing import LabelEncoder
categorical_cols = df_clean.select_dtypes(include=["object"]).columns
for col in categorical_cols:
    print(f"Encoding column: {col}")
    le = LabelEncoder()
    df_clean[col] = le.fit_transform(df_clean[col])
df_clean
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler
data_for_scale=df_clean.copy()
print('data before scaling:\n', data_for_scale) #show before scaling
standard_scaler = StandardScaler() # Instantiate StandardScaler
data_standard_scaled = standard_scaler.fit_transform(data_for_scale) # Apply fit_transform

print('\nData after scaling:\n', data_standard_scaled) #show after scaling
```

```python
# Task 6: Write your code here:
#Check for target imbalance
import seaborn as sns
def check_target_imbalance(df_data, target_column):
  print("Target Distribution:")
  print(df_data[target_column].value_counts(normalize=True))
  sns.countplot(x=df_data[target_column])
  plt.title("Target Distribution")
  plt.show()

check_target_imbalance(df_data, "Delivery_Time")
```
