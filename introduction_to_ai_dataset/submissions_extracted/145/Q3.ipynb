```python
# Task 1: Write your code here:
#Read the dataset
data_path = os.path.join(path, 'Q3_data.csv')
df_data = pd.read_csv(data_path)
```

```python
# Task 1: Write your code here:
#Handle missing values
missing_values = df_data.isnull().sum()
print("Missing values:\n" ,missing_values)
```

```python
# Task 2: Write your code here:
#Check and remove duplicates
df_data.duplicated()
duplicates = df_data.drop_duplicates(inplace=True)
print(duplicates)
```

```python
# Task 3: Write your code here:
#Encode categorical variables
from sklearn.preprocessing import LabelEncoder
categorical_cols = df_data.select_dtypes(include=["object"]).columns
for col in categorical_cols:
    print(f"Encoding column: {col}")
    le = LabelEncoder()
    df_data[col] = le.fit_transform(df_data[col])
df_data
```

```python
# Task 4: Write your code here:
#scaling
from sklearn.preprocessing import StandardScaler
data_for_scale=df_data.copy()
print('data before scaling:\n', data_for_scale) #show before scaling
standard_scaler = StandardScaler() # Instantiate StandardScaler
data_standard_scaled = standard_scaler.fit_transform(data_for_scale) # Apply fit_transform

print('\nData after scaling:\n', data_standard_scaled) #show after scaling
```

```python
# Task 5: Write your code here:
#Check for target imbalance
import seaborn as sns
def check_target_imbalance(df_data, target_column):
  print("Target Distribution:")
  print(df_data[target_column].value_counts(normalize=True))
  sns.countplot(x=df_data[target_column])
  plt.title("Target Distribution")
  plt.show()

check_target_imbalance(df_data, "Target")
```

```python
# Task 1: Write your code here:
from sklearn.model_selection import train_test_split
X_s = df_data.drop("target", axis=1).astype(float)
y_s = df_data["target"].astype(float)
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import KFold


X, y = X_s.copy(), y_s.copy()

# K-Fold Cross Validation
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# Iterate through folds
for fold, (train_idx, test_idx) in enumerate(kf.split(X), start=1):
    # indexing for each fold
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
    # print shapes
    print(f"Fold {fold}")
    print("  X_train shape:", X_train.shape)
    print("  X_test shape :", X_test.shape)
    print("  y_train shape:", y_train.shape)
    print("  y_test shape :", y_test.shape)
    print("-" * 30)
```
