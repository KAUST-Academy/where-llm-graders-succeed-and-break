```python
clean_data = data.drop(columns=['Order_ID'])
clean_data.head()
```

```python
# 2. Do we have missing values?
def check_missing_values(df):

  # Get missing values using pandas
  missing_values = df.isnull().sum()

  print("Missing Values per Column:")
  print(missing_values[missing_values > 0])

  if missing_values.any():
    df['Weather'].fillna(df['Weather'].mode()[0])
    df['Traffic_Level'].fillna(df['Traffic_Level'].mode()[0])
    df['Courier_Experience_yrs'].fillna(df['Courier_Experience_yrs'].mode()[0])
    df['Delivery_Time'].fillna(df['Delivery_Time'].mean())
    print("handled")

  else:
    print("\nNo Missing Values Found.")

check_missing_values(clean_data)

```

```python
def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(clean_data)
clean_data.info()
```

```python

from sklearn.preprocessing import LabelEncoder
categorical_cols = ['Weather', 'Traffic_Level', 'Time_of_Day', 'Vehicle_Type']

for col in categorical_cols:
    le = LabelEncoder()
    clean_data[col] = le.fit_transform(clean_data[col].astype(str))
```
