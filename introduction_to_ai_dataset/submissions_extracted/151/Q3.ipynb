```python
import kagglehub


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
import kagglehub

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

import warnings
warnings.filterwarnings('ignore')


# Download latest version

path = kagglehub.dataset_download("mohammad2012191/q3-ka-ai-2026")



print("Path to dataset files:", path)
```

```python
# Task 1: Write your code here:
Q3_path = os.path.join(path, 'Q3_data.csv')
df_Q3 = pd.read_csv(Q3_path)
```

```python
# Task 1: Write your code here:
print("Missing values:")
print(df_Q3.isnull().sum())

df_Q3['D_39'] = df_Q3['D_39'].fillna('none')
df_Q3['B_1'] = df_Q3['B_1'].fillna('none')
df_Q3['R_1'] = df_Q3['R_1'].fillna('none')
df_Q3['Target'] = df_Q3['Target'].fillna('none')
```

```python
# Task 2: Write your code here:
print("Checking for duplicates...")
dups = df_Q3.duplicated().sum()
if dups > 0:
    print(f"Found {dups} duplicates. Removing them...")
    df_Q3.drop_duplicates(inplace=True)
    print("Duplicates removed.")
else:
    print("No duplicates found.")
```

```python
# Task 3: Write your code here:
categorical_cols = ['', '', '', '']
for col in categorical_cols:
    le = LabelEncoder()
    df_Q3[col] = le.fit_transform(df_Q3[col].astype(str))
```
