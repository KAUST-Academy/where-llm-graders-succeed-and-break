```python
# Task 1: Write your code here:
import os
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

path = os.path.join(path, "Q3_data.csv")
df = pd.read_csv(path)
```

```python
# Task 1: Write your code here:
def check_missing_values(df):
  missing_values = df.isnull().sum()
  if missing_values.sum() == 0:
    print("No Missing Values Found.")
  else:
    print(missing_values[missing_values>0])
    print(f"{missing_values.sum()} missing values found.")
check_missing_values(df)

# Task continued below
```

```python
#Since all the data is numerical and a large portion is missing, the best choice is imputing missing values with the mean

missing_values = df.isnull().sum()
missing_cols = missing_values[missing_values>0].index

for col in missing_cols:
  df[col] = df[col].fillna(df[col].mean())

df.head()
```

```python
# Task 2: Write your code here:
duplicates = df.duplicated().sum()
if duplicates>0:
  df.drop_duplicates(inplace=True)
  print("Duplicated removed")
else:
  print("no duplicates found")
```

```python
# Task 3: Write your code here:
categorical_cols = df.select_dtypes(include=['object']).columns
print(categorical_cols)
# No categorical columns
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
features = df.drop(columns='Target').columns
df[features] = scaler.fit_transform(df[features])
df.describe()
```

```python
# Task 5: Write your code here:

target = df['Target']
plt.bar(target.value_counts().index, target.value_counts().values, color='blue')
plt.xlabel("Target")
plt.ylabel("Frequency")
plt.title("Target Distribution")
# Task continued below
```

```python
# Target is clearly imbalanced, I must use Stratified KFold & F1 score
```

```python
# Task 1: Write your code here:

X = df[features]
y = df['Target']
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import StratifiedKFold
from catboost import CatBoostClassifier
from sklearn.metrics import accuracy_score, f1_score
skf = StratifiedKFold(n_splits=5, random_state=42, shuffle=True)

model = CatBoostClassifier(verbose=0, n_estimators=300, max_depth = 5, learning_rate=.05)

accuracy = []
f1 = []

for fold, (train_idx, test_idx) in enumerate(skf.split(X, y), start=1):
  X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
  y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

  model.fit(X_train, y_train)

  y_pred = model.predict(X_test)

  accuracy.append(accuracy_score(y_test, y_pred))
  f1.append(f1_score(y_test, y_pred, zero_division=0))

acc_mean = np.mean(accuracy)
f1_mean = np.mean(f1)

print("Results\n")
print(f"Mean Accuracy: {acc_mean:.4f}")
print(f"F1 Score Mean: {f1_mean:.4f}")
```

```python
# Task 1: Write your code here:

features = X.columns
importances = model.feature_importances_
sorted_idx = np.argsort(importances)

plt.figure(figsize=(6, 10))
plt.barh(features[sorted_idx], importances[sorted_idx], color='purple')
plt.xlabel("Importances")
plt.ylabel("Features", fontsize=.01)
plt.title("Feature Importances")
```

```python
# Task 2: Write your code here:
highest_import_idx = np.argsort(importances)[-1]
golden_feature=features[highest_import_idx]
print(f"The Golden Feature is : {golden_feature}")
```
