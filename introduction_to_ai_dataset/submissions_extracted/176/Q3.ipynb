```python
# Task 1: Write your code here:
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import warnings
import seaborn as sns
warnings.filterwarnings('ignore')

file_path = os.path.join(path, 'Q3_data.csv')
df = pd.read_csv(file_path)
```

```python
# Task 1: Write your code here:
for col in df.columns:
  if df[col].isnull().sum != 0 and (df[col].dtypes == 'float64' or df[col].dtypes == 'int64'):
    df[col].fillna(df[col].mean(),inplace=True)

df.dropna(inplace=True)
```

```python
# Task 2: Write your code here:
def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    df.drop_duplicates(inplace=True)

check_duplicates(df)
```

```python
# Task 3: Write your code here:
from sklearn.preprocessing import LabelEncoder


for col in df.columns:
  if df[col].dtypes == 'object': # Checking the category columns
    label_encoder = LabelEncoder()
    df[col] = label_encoder.fit_transform(df[col])
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df_new = scaler.fit_transform(df)
```

```python
# Task 1: Write your code here:
X = df.drop(columns=['Target'],inplace=False)
y = df['Target']
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import KFold
from catboost import CatBoostRegressor
from sklearn.metrics import accuracy_score, f1_score

#accuracy for each Fold
accuracies = 0
f1_scores = 0

# Define K-Fold Cross Validation
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# Iterate through folds
for fold, (train_idx, test_idx) in enumerate(kf.split(X), start=1):
    # indexing for each fold
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    # Train the Model
    model_cat = CatBoostRegressor(verbose=0)
    model_cat.fit(X_train, y_train)

    # Predict and evaluate
    y_pred = model_cat.predict(X_test)
    accuracy = accuracy_score(y_test, bool(y_pred).sum())
    f1_score = f1_score(y_test, bool(y_pred).sum())

    accuracies += accuracy
    f1_scores +=f1_score

print(f"Averaged score across all folds: {accuracies/5}")
```

```python
# Task 1: Write your code here:
# Feature importance
feature_importance = pd.DataFrame({
    'feature': X.columns,
    'importance': model_cat.feature_importances_
}).sort_values('importance', ascending=False)

plt.figure(figsize=(10, 6))
plt.barh(feature_importance['feature'], feature_importance['importance'])
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show()
```
