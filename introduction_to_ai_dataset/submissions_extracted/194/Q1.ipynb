```python
# Task Bonus: Write your code here:
 import numpy as np
import pandas as pd

from sklearn.model_selection import KFold
from sklearn.metrics import mean_absolute_error
from sklearn.ensemble import RandomForestRegressor
from catboost import CatBoostRegressor

# Load dataset (you already have this part)

df = pd.read_csv("food_delivery.csv")

X = df.drop("delivery_time", axis=1)
y = df["delivery_time"]

# KFold setup
kf = KFold(n_splits=5, shuffle=True, random_state=42)

mae_scores = []

# KFold loop with ensemble

for fold, (train_idx, val_idx) in enumerate(kf.split(X)):
    print(f"Fold {fold + 1}")

    X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
    y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

    rf_model = RandomForestRegressor(
        n_estimators=200,
        max_depth=None,
        random_state=42,
        n_jobs=-1
    )


    cb_model = CatBoostRegressor(
        iterations=500,
        learning_rate=0.05,
        depth=8,
        loss_function="MAE",
        verbose=False,
        random_seed=42
    )

    # Train both models
    rf_model.fit(X_train, y_train)
    cb_model.fit(X_train, y_train)

    # Predictions
    rf_preds = rf_model.predict(X_val)
    cb_preds = cb_model.predict(X_val)

    ensemble_preds = (rf_preds + cb_preds) / 2

    # MAE
    mae = mean_absolute_error(y_val, ensemble_preds)
    mae_scores.append(mae)

    print(f"MAE: {mae:.4f}")

print("\nAverage MAE across folds:", np.mean(mae_scores))
```
