```python
# Task 1: Write your code here:
# handle missing values
def check_missing_values(df_anonymized):
  missing_values = df_anonymized.isnull().sum()
  print("Missing Values per Column:")
  print(missing_values[missing_values > 0])
  if missing_values.any():
    print("\nHandle Missing Values as needed.")
  else:
    print("\nNo Missing Values Found.")

check_missing_values(df_anonymized)
```

```python
# fill empty values
for col in ['P_2',	'D_39',	'B_1',	'B_2',	'R_1',	'S_3',	'D_41',	'B_3',	'D_42',	'D_43', 'D_137', 	'D_138',	'D_139',	'D_140',	'D_141',	'D_142',	'D_143',	'D_144',	'D_145'	]:
   df_anonymized[col]=df_anonymized[col].fillna(df_anonymized[col].mean())
```

```python
# Task 2: Write your code here:
def check_duplicates(df_anonymized):
  duplicates = df_anonymized.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df_anonymized.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df_anonymized)
```
