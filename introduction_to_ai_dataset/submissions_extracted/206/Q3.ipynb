```python
# Task 1: Write your code here:
# Task 2: Write your code here:
def check_missing_values(df): # to check if we have missing value
  missing_values = df.isnull().sum()
  print("Missing Values per Column:")
  print(missing_values[missing_values > 0])
  if missing_values.any():
    print("\nHandle Missing Values as needed.")
  else:
    print("\nNo Missing Values Found.")

check_missing_values(df_Golden_Feature)

for col in df_Golden_Feature.columns: #here we fill the int col with mode
    df_Golden_Feature[col] = df_Golden_Feature[col].fillna(df_Golden_Feature[col].mode()[0])

print("Missing values remaining:", df_Golden_Feature.isnull().sum().sum()) #here to check if we still have missing value

```

```python
# Task 2: Write your code here:
def check_duplicates(df): #check if we have duplicates if we have drop it
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df_Golden_Feature)
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import StandardScaler #import StandardScaler

features = df_Golden_Feature.columns.drop("Target")  #we won't SCALE THE TARGET
print(features)
scaler = StandardScaler()
df_Golden_Feature[features] = scaler.fit_transform(df_Golden_Feature[features])

df_Golden_Feature.head()
```

```python
# Task 5: Write your code here:
import matplotlib.pyplot as plt # import the pylot to plot the target
import seaborn as sns
def check_target_imbalance(df, target_column):
  print("Target Distribution:")
  print(df[target_column].value_counts(normalize=True))
  sns.countplot(x=df[target_column])
  plt.title("Target Distribution")
  plt.show()

check_target_imbalance(df_Golden_Feature, "Target") # the target is imbalnce
```

```python
# Task 1: Write your code here:
features = df_Golden_Feature.columns.drop("Target")
X= df_Golden_Feature[features]
y= df_Golden_Feature["Target"]


X
```

```python
# Task 2,3,4,5: Write your code here:
%pip install kagglehub catboost xgboost tqdm -q

from catboost import CatBoostClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, f1_score




strfold = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

lr_accuracy = []
lr_f1 = []

for train_idx, val_idx in strfold.split(X,y):

    X_fold_train, X_fold_val = X.iloc[train_idx], X.iloc[val_idx]
    y_fold_train, y_fold_val = y.iloc[train_idx], y.iloc[val_idx]

    model = CatBoostClassifier(
      verbose=0,
      n_estimators=320,
      max_depth=4
    )

    model.fit(X_fold_train, y_fold_train)
    y_fold_pred = model.predict(X_fold_val)

    accuracy = accuracy_score(y_fold_val, y_fold_pred)
    f1 = f1_score(y_fold_val, y_fold_pred, zero_division=0)

    lr_accuracy.append(accuracy)
    lr_f1.append(f1)


print(f"5-Fold df_Golden_Feature Results:")
for x in lr_accuracy:
  sum1=0
  sum1 += x
average_accuracy = sum1 / 5

for x in lr_f1:
  sum1=0
  sum1 += x
average_f1 = sum1 / 5

print(f"accuracy:  {average_accuracy:,.2f}")
print(f"f1:  {average_f1:,.2f}")


```

```python
# Task 1: Write your code here:
features = df_Golden_Feature.columns.drop("Target")
feature_importance = pd.DataFrame({
    'feature': features,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

plt.figure(figsize=(30, 20))
plt.barh(feature_importance['feature'], feature_importance['importance'])
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show() # the golden_feature is P_2
```
