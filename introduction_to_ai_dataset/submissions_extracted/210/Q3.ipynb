```python
# Analyze missing values
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_data = pd.DataFrame({
    'Column': missing_percentage.index,
    'Missing_Percentage': missing_percentage.values
})
missing_data = missing_data[missing_data['Missing_Percentage'] > 0].sort_values('Missing_Percentage', ascending=False)

print("Missing Data Analysis:")
missing_data
```

```python
# Task 1: Write your code here:
df.dropna(axis=1,inplace=True)

```

```python
# Task 2: Write your code here:
# 4. Do we have duplicate samples?
def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df)
```

```python
# Task 3: Write your code here:
# 3. Do we have categorical columns?
categorical_cols = df.select_dtypes(include=["object"]).columns

print("Categorical Columns:", list(categorical_cols))
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import StandardScaler #import StandardScaler

print('data before scaling:\n', df) #show before scaling
feature = df.drop(columns=["Target"])
standard_scaler = StandardScaler() # Instantiate StandardScaler
feature[feature.columns] = standard_scaler.fit_transform(feature) # Apply fit_transform
df[feature.columns] = feature[feature.columns]
print('\nData after scaling:\n', df) #show after scaling
```

```python
# Import models
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from catboost import CatBoostClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, f1_score
import numpy as np
n_splits = 2

skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)
# Task 1: Write your code here:
X = df.drop(columns=["Target"]).copy()
y = df["Target"].copy()

f1_ = []
model = CatBoostClassifier(random_state=42,verbose=0)
for fold_idx, (train_index, test_index) in enumerate(skf.split(X, y)):

  print(f"\nFold {fold_idx + 1}/{n_splits}")

  # Get the train & test split for this fold
  X_train, X_test = X.iloc[train_index], X.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]

  # Train using gradient descent with learning rate = 0.5
  model.fit(X_train,y_train)

  # Pick the predicted classes with the highest probability
  y_pred = model.predict(X_test)

  # Calculate evaluation metrics
  f1 = f1_score(y_test, y_pred)  # for multiclass f1 score, you should set the average hyperparameter ("macro", "micro", "weighted")
  print("f1-scroe: ",f1)

  # Store results
  f1_.append(f1)


# Print the average of each evaluation metric across folds
print(f"  F1-Score:  {np.mean(f1_):.4f}")
```
