```python
# Task 1: Write your code here:
import os
import pandas as pd

path = os.path.join(path, 'Q3_data.csv')
df = pd.read_csv(path)
```

```python
# Task 1: Write your code here:
print("Missing values per column:")
df_missing = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df)) * 100
mpct_col = missing_pct[missing_pct > 0].sort_values(ascending=False)
print(mpct_col.head(50))

# remove columns that have waaaaaay too many missing values (over 50 percent)
for col, missing_percentage in mpct_col.items():
  # fill the rest of the missing values with the mode
  df[col] = df[col].fillna(df[col].mode()[0])

  #drop the cols with most values missing
  if missing_percentage > 50:
    df = df.drop(col,axis=1)




print(df.shape)
df.head()
```

```python
# Task 2: Write your code here:
print(f"Duplicates: {df.duplicated().sum()}")
df.drop_duplicates(inplace=True)
```

```python
# Task 3: Write your code here:
# no need for encoding
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import StandardScaler

features = df.columns.drop("Target")
scaler = StandardScaler()
df[features] = scaler.fit_transform(df[features])
df.head()
```

```python
# Task 5: Write your code here:
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
df['Target'].value_counts().plot(kind='bar')
plt.title('Target Distribution')
plt.xlabel('Target')
plt.ylabel('Count')
plt.show()
```

```python
# Task 1: Write your code here:
X = df.drop('Target',axis=1)
y = df['Target']
```

```python
# Task 2,3,4,5: Write your code here:
!pip install catboost
from sklearn.model_selection import StratifiedKFold
from catboost import CatBoostClassifier
from sklearn.metrics import f1_score
from sklearn.metrics import accuracy_score

model = CatBoostClassifier(n_estimators=100, max_depth=5, verbose=0)


skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
avg_f1 = []
avg_acc = []

for fold, (train_idx, test_idx) in enumerate(skf.split(X, y)):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    f1= f1_score(y_test, y_pred)
    acc = accuracy_score(y_test, y_pred)
    avg_f1.append(f1)
    avg_acc.append(acc)

print("\nThe average F1-score across all 5 folds: \n",avg_f1)
print("\nThe average Accuracy across all 5 folds: \n",avg_acc)

# i would personally judge off of f1-score instead of accuracy
```

```python
# Task 1: Write your code here:
feature_importance = pd.DataFrame({
    'feature': X.columns,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=True)

plt.figure(figsize=(10, 8))
plt.barh(feature_importance['feature'][-10:], feature_importance['importance'][-10:])
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.show()

# to validate
print(max(feature_importance['importance']))
```
