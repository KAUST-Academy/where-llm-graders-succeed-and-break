```python
Missing_data = (df.isnull().sum() / len(df))
df_Missing_data = pd.DataFrame(
    {
        'column':Missing_data.index,
        'Percent':Missing_data.values
    }
)
df_Missing_data = df_Missing_data[df_Missing_data['Percent'] > 0].sort_values('Percent',ascending=False)
missing_columns = []
for cols in df_Missing_data['column']:
  missing_columns += [cols]
for col in missing_columns:
  df_clean = df.dropna(subset= col).copy()
df_clean
```
