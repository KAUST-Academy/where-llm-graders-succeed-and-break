```python
# Task 1: Write your code here:
X = df[cols]
y = df['Target']
```

```python
# Task 2,3,4,5: Write your code here:
from catboost import CatBoostClassifier

n_splits=5

# To store scores
accuracy_all = []
f1_all = []

skf = StratifiedKFold(n_splits, shuffle=True, random_state=42 )

for fold_idx, (train_index, test_index) in enumerate(skf.split(X, y)):
    print(f"\nFold {fold_idx + 1}/{n_splits}")

    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

    model = CatBoostClassifier(n_estimators=320, max_depth=4)
    model.fit(X_train, y_train) # train
    y_pred = model.predict(X_test) # validate

    # Calculate evaluation metrics
    accuracy = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred, zero_division=0)

    # Store results
    accuracy_all.append(accuracy)
    f1_all.append(f1)
```

```python
accuracy_avg = sum(accuracy_all) / len(accuracy_all)
f1_avg = sum(f1_all) / len(f1_all)

print(f"Average accuracy is: {accuracy_avg}")
print(f"Average f1 is: {f1_avg}")
```

```python
# Task 1: Write your code here:

logistic_importance = list(zip(X.columns, model.feature_importances_))
sorted_logistic_importance = sorted(logistic_importance, key=lambda x: abs(x[1]), reverse=True)
sorted_logistic_importance = sorted_logistic_importance[:10] # Choosing the highest 10 features


# Extract sorted features and their coefficients
features, coefficients = zip(*sorted_logistic_importance)



# Plot feature importances
plt.figure(figsize=(10, 6))
plt.barh(features, np.abs(coefficients), color='darkblue')
plt.xlabel('Coefficient Value')
plt.ylabel('Features')
plt.title('Logistic Regression Feature Importance')
plt.gca().invert_yaxis()  # Invert y-axis to show the most important features at the top
plt.show()
```
