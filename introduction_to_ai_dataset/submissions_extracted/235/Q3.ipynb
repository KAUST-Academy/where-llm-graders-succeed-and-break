```python
# Task 4: Write your code here:
scaler = StandardScaler()
numerical_cols = df.select_dtypes(include=['int64','float64']).columns.drop('Target')
df[numerical_cols] = scaler.fit_transform(df[numerical_cols])
```

```python
# Task 1: Write your code here:
X = df.drop(columns=['Target'])
y = df['Target']
```

```python
# Task 2,3,4,5: Write your code here:
# Task 2,3,4,5: Write your code here:
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42) # classifiers

f1_mean = []
acc_mean = []
for fold_idx, (train_index, test_index) in enumerate(skf.split(X, y)):
  X_train, X_test = X.iloc[train_index], X.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]
  model = CatBoostClassifier(verbose=0,n_estimators=300,max_depth=4)
  model.fit(X_train,y_train)
  y_pred = model.predict(X_test)
  f1 = f1_score(y_test,y_pred)
  acc = accuracy_score(y_test,y_pred)
  f1_mean.append(f1)
  acc_mean.append(acc)

print(f'Mean MAE: {np.mean(f1_mean)}')
print(f'Mean accuracy: {np.mean(acc_mean)}')
```

```python
# Task 1: Write your code here:
pd.Series(model.feature_importances_,index=X.columns)[:14].sort_values().plot(kind='barh') #limited to only 10 because we have alot and all
#are 0 importance
```

```python
# Task 2: Write your code here:
print(f'most important feature is: {pd.Series(model.feature_importances_,index=X.columns)[:1].sort_values(ascending=False)}')
```
