```python
# Task 4: Write your code here:
#Encode categorical variables if needed (Bonus if used One Hot Encoding)


df_clean = df_food.dropna().copy()
le = LabelEncoder()
# Fit on ALL types found in both columns so the numbers match
all_types = pd.concat([df_clean['Time_of_Day'], df_clean['Vehicle_Type']])
le.fit(all_types)
df_clean['Time_of_Day'] = le.transform(df_clean['Time_of_Day'])
df_clean['Vehicle_Type'] = le.transform(df_clean['Vehicle_Type'])
```

```python
# Task 5: Write your code here:
# Apply feature scaling for all features (Use StandardScaler)

#1- first doing is Feature Engineering to better
df_clean['late_Delivery_Time'] = 10 - df_clean['Delivery_Time'].head()
feature_cols = ['Order_ID' , 'Distance_km' , 'Preparation_Time_min' ,
                'Courier_Experience_yrs' , 'Delivery_Time']
#2-Scale features
# Train-test split (80% train, 20% test)
x = df_clean[feature_cols]
y = df_clean['Courier_Experience_yrs']
# TODO: Split data
X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42, stratify=y
)


scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```
