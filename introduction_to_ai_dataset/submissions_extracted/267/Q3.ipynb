```python
# Task 2: Write your code here:
df = df.drop_duplicates()
```

```python
# Task 3: Write your code here:

categorical_cols = df.select_dtypes(include=["object"]).columns
label_encoders = {}
for col in categorical_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col].astype(str))
    label_encoders[col] = le
```

```python
# Task 4: Write your code here:

target_col = "Target"
num_cols = df.drop(columns=[target_col]).select_dtypes(include=["number"]).columns

scaler = StandardScaler()
df[num_cols] = scaler.fit_transform(df[num_cols])
```

```python
# Task 5: Write your code here:

df["Target"].value_counts(normalize=True)
```

```python
# Task 1: Write your code here:

X = df.drop("Target", axis=1)
y = df["Target"]

```

```python
# Task 2,3,4,5: Write your code here:

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

model = CatBoostClassifier(verbose=0, random_state=42)

if y.value_counts(normalize=True).max() > 0.7:
    scorer = make_scorer(f1_score, average="weighted")
else:
    scorer = make_scorer(accuracy_score)

scores = cross_val_score(model, X, y, cv=skf, scoring=scorer)

print(np.mean(scores))
```

```python
# Task 1: Write your code here:

model.fit(X, y)

importances = model.get_feature_importance()
features = X.columns

import matplotlib.pyplot as plt
import numpy as np

indices = np.argsort(importances)

plt.figure()
plt.title("Feature Importance")
plt.barh(range(len(indices)), importances[indices])
plt.yticks(range(len(indices)), features[indices])
plt.show()
```

```python
# Task 2: Write your code here:

golden_feature = features[np.argmax(importances)]
print(golden_feature)
```
