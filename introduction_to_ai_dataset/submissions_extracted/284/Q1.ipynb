```python
# Task 1: Write your code here:
df_clean = df.drop(columns=["Order_ID"])
df_clean.info()
```

```python
# Task 2: Write your code here:
df_clean = df_clean.dropna(subset= ['Weather', "Traffic_Level", "Time_of_Day", "Courier_Experience_yrs", "Delivery_Time"])
df_clean.info()
```

```python
# Task 3: Write your code here:
df_clean.drop_duplicates(inplace=True)
```

```python
# Task 4: Write your code here:
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, KFold
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error

import warnings
warnings.filterwarnings('ignore')



categorical_cols = ['Vehicle_Type', 'Time_of_Day', 'Traffic_Level', 'Weather']
for col in categorical_cols:
    le = LabelEncoder()
    df_clean[col] = le.fit_transform(df_clean[col].astype(str))

df_clean.head()
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler

numerical_cols = df_clean.select_dtypes(include=["int64", "float64"]).columns.drop("Delivery_Time")  # DON'T SCALE THE TARGET

scaler = StandardScaler()
df_clean[numerical_cols] = scaler.fit_transform(df_clean[numerical_cols])

df_clean.head()
```

```python
# Task 6: Write your code here:
# our target is a cont value, so no such thing
```

```python
# Task 1: Write your code here:
X = df_clean.drop("Delivery_Time", axis=1).astype(float)
y = df_clean['Delivery_Time'].astype(float)
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import KFold

n_splits = 5 # K=5 Folds

# 5-Fold, shuffled
kf = KFold(n_splits=n_splits, shuffle=True, random_state=42)

mse_scores = []
model = RandomForestRegressor()
for fold, (train_idx, test_idx) in enumerate(kf.split(X, y), start=1):
    # indexing for each fold
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
              # instantiate
    model.fit(X_train, y_train)         # fit
    y_pred = model.predict(X_test)      # predict

    mse_scores.append(mean_squared_error(y_test, y_pred))

print(np.mean(mse_scores))
```
