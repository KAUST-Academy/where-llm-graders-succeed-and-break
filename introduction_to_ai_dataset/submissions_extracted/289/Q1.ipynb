```python
# Task 2: Write your code here:
cols = ['Order_ID',	'Distance_km',	'Weather',	'Traffic_Level',	'Time_of_Day',
        'Vehicle_Type',	'Preparation_Time_min',	'Courier_Experience_yrs',	'Delivery_Time']
df_clean = df_delivery[cols].copy()
df_clean = df.drop(columns=(cols))
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import LabelEncoder
for col in df_clean:
    le = LabelEncoder()
    df_clean[col] = le.fit_transform(df_clean[col].astype(str))
print(df_clean)
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler

data_for_scale = pd.DataFrame(df_clean)

standard_scaler = StandardScaler()
data_standard_scaled = standard_scaler.fit_transform(data_for_scale)

print('\nData after scaling:\n', data_standard_scaled)
```

```python
# Task 1: Write your code here:
feature_cols = pd.DataFrame(['Order_ID',	'Distance_km',	'Weather',	'Traffic_Level', 'Time_of_Day',
        'Vehicle_Type',	'Preparation_Time_min',	'Courier_Experience_yrs'])
feature = df_clean[feature_cols]
target = df_clean['Delivery_Time']
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import mean_squared_error as mean_absolute_error

X, y = feature.copy(), target.copy()
kf = KFold(n_splits=5, shuffle=True)

for fold, (train_indx, test_indx) in enumerate(kf.split(X), start=1):
    X_train, X_test = X.iloc[train_indx], X.iloc[test_indx]
    y_train, y_test = y.iloc[train_indx], y.iloc[test_indx]

model= RandomForestClassifier()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

mae= []
lr_mae = mean_absolute_error(y_test, y_pred)
lr_mae.append(mae)

```
