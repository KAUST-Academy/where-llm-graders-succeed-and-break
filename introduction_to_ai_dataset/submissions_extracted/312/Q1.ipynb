```python
# Task Bonus: Write your code here:

## ==================================================
## TRAIN MULTIPLE MODELS + K-FOLD CV + F1-SCORE
## ==================================================
## Goal:
## - Train several classifiers (LogReg, RandomForest, CatBoost)
## - Build a soft VotingClassifier (ensemble)
## - Evaluate each model using 5-Fold Cross-Validation
## - Use weighted F1-score as the main metric

import numpy as np

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, VotingClassifier
from sklearn.model_selection import KFold
from sklearn.metrics import f1_score
from catboost import CatBoostClassifier


## -----------------------------
## 1) Define classification models
## -----------------------------

## Create a dictionary of models so we can evaluate them in a loop
models = {
    "Logistic Regression": LogisticRegression(max_iter=10000),
    "Random Forest Classifier": RandomForestClassifier(n_estimators=100, random_state=42),
    "CatBoost Classifier": CatBoostClassifier(verbose=0, random_state=42)
}

## -----------------------------
## 2) Build an Ensemble (Soft Voting)
## -----------------------------

## Soft voting = combines predicted probabilities (better when models are strong)
models["Ensemble"] = VotingClassifier(
    estimators=[
        ("cb", models["CatBoost Classifier"]),
        ("lr", models["Logistic Regression"]),
        ("rf", models["Random Forest Classifier"])
    ],
    voting="soft"
)

## -----------------------------
## 3) K-Fold Cross Validation (K=5)
## -----------------------------

## KFold splits the dataset into 5 parts and evaluates 5 times
## shuffle=False matches the test requirement (no shuffling)
kf = KFold(n_splits=5, shuffle=False)

## -----------------------------
## 4) Evaluate each model using F1-score
## -----------------------------

for model_name, model in models.items():

    ## Store F1 score from each fold
    scores_f1 = []

    ## Split indices into train and test folds
    for train_idx, test_idx in kf.split(X, y):

        ## Build fold train/test data
        X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
        y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

        ## Train the model on training fold
        model.fit(X_train, y_train)

        ## Predict on the test fold
        y_pred = model.predict(X_test)

        ## Compute weighted F1-score (handles class imbalance better than accuracy)
        f1 = f1_score(y_test, y_pred, average="weighted")
        scores_f1.append(f1)

    ## Print mean F1-score across the 5 folds
    print(f"{model_name} F1-Score: {np.mean(scores_f1):.4f}")
    print()
```
