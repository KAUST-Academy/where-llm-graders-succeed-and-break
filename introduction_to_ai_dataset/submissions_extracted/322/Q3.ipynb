```python

df = pd.read_csv(path + "/Q3_data.csv")
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

target_col = df.columns[-1]

missing = df.isnull().sum()
print("Columns with missing values:\n", missing[missing > 0])

num_cols = df.drop(columns=[target_col]).select_dtypes(include=["int64", "float64"]).columns
cat_cols = df.drop(columns=[target_col]).select_dtypes(include=["object"]).columns

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df = df.drop_duplicates()
```

```python
if len(cat_cols) > 0:
    df = pd.get_dummies(df, columns=cat_cols, drop_first=True)
```

```python
X = df.drop(columns=[target_col])
y = df[target_col]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
counts = y.value_counts()
ratio = counts.min() / counts.max()
print("Target counts:\n", counts)
print("Minority/Majority ratio:", round(ratio, 3))
print("Imbalanced:", "Yes" if ratio < 0.5 else "No")
```

```python
import numpy as np
from sklearn.model_selection import StratifiedKFold
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score, f1_score
```

```python
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
# Note: Keeps class distribution in each fold.

scores = []
use_f1 = (y.value_counts().min() / y.value_counts().max()) < 0.5

for train_idx, test_idx in skf.split(X, y):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    model = GradientBoostingClassifier(random_state=42)
    model.fit(X_train, y_train)
    # Note: Model is trained on each fold.

    y_pred = model.predict(X_test)

    if use_f1:
        score = f1_score(y_test, y_pred)
        metric_name = "F1"
        # Note: F1 is used when data is imbalanced.
    else:
        score = accuracy_score(y_test, y_pred)
        metric_name = "Accuracy"
        # Note: Accuracy is used when classes are balanced.

    scores.append(score)

# Task 5: Print average score
print(f"{metric_name} per fold:", np.round(scores, 3))
print(f"Average {metric_name}:", round(np.mean(scores), 3))
# Note: Final score is averaged across folds.
```
