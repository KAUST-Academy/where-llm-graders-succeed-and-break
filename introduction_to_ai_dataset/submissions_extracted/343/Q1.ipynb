```python
# Task 4: Write your code here:

df = pd.get_dummies(df)


df.head()
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

features = df.drop('delivery_time', axis=1)
df_scaled = pd.DataFrame(scaler.fit_transform(features), columns=features.columns)


df_scaled['delivery_time'] = df['delivery_time'].values
```

```python
# Task 6: Write your code here:

plt.hist(df['delivery_time'])
print("Target is checked for distribution.")
```

```python
# Task 1: Write your code here:

X = df_scaled.drop('delivery_time', axis=1)
y = df_scaled['delivery_time']
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
import numpy as np

kf = KFold(n_splits=5, shuffle=True, random_state=42)
model = RandomForestRegressor(n_estimators=100, random_state=42)
mae_scores = []

for train_index, test_index in kf.split(X):
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    mae = mean_absolute_error(y_test, predictions)
    mae_scores.append(mae)

print(f"Average MAE: {np.mean(mae_scores)}")
```

```python
# Task 1: Write your code here:

import pandas as pd
import matplotlib.pyplot as plt

importances = model.feature_importances_
feature_names = X.columns
feature_importance_df = pd.Series(importances, index=feature_names).sort_values(ascending=True)

feature_importance_df.plot(kind='barh', color='teal')
plt.title('Feature Importance')
plt.show()
```

```python
# Task Bonus: Write your code here:

from catboost import CatBoostRegressor

mae_ensemble_scores = []

for train_index, test_index in kf.split(X):
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]


    rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
    cb_model = CatBoostRegressor(silent=True, random_state=42)

    rf_model.fit(X_train, y_train)
    cb_model.fit(X_train, y_train)


    rf_preds = rf_model.predict(X_test)
    cb_preds = cb_model.predict(X_test)
    avg_preds = (rf_preds + cb_preds) / 2

    mae_ensemble_scores.append(mean_absolute_error(y_test, avg_preds))

print(f"Average Ensemble MAE: {np.mean(mae_ensemble_scores)}")
```
