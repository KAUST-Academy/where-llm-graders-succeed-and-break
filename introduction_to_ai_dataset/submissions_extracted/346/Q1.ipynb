```python
# Task 2: Write your code here:
for col in [ 'Distance_km', 'Weather', 'Traffic_Level', 'Time_of_Day', 'Vehicle_Type', 'Preparation_Time_min', 'Courier_Experience_yrs', 'Delivery_Time']:
    df_clean[col] = df_clean[col].fillna('unknown')

# Fill cylinders with mode - discrete feature, mode is most representative
df_clean['Order_ID'] = df_clean['Order_ID'].fillna(df_clean['Order_ID'].mode()[0])

print("Missing values remaining:", df_clean.isnull().sum().sum())
```
