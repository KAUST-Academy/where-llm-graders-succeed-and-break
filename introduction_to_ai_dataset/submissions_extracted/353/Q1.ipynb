```python
import kagglehub
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
import matplotlib.pyplot as plt
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
import numpy as np


# Download latest version
path = kagglehub.dataset_download("mohammad2012191/q1-ka-ai-2026")

print("Path to dataset files:", path)
```

```python
# Task 1: Write your code here:
dil_path = os.path.join(path, 'Q1_data.csv')
df_dil = pd.read_csv(dil_path)
```

```python
# Task 1: Write your code here:
df_dil = df_dil.drop(columns=["Order_ID"])
```

```python
# Task 2: Write your code here:
df_dil = df_dil.dropna(subset=["Delivery_Time"])

df_dil["Courier_Experience_yrs"] = df_dil["Courier_Experience_yrs"].fillna(df_dil["Courier_Experience_yrs"].median())

for col in ["Weather", "Traffic_Level", "Time_of_Day"]:
    df_dil[col] = df_dil[col].fillna(df_dil[col].mode()[0])
```

```python
# Task 3: Write your code here:
#### نشوف كم صف مكرر
dup=df_dil.duplicated().sum()
### نتحذف الصفوف المكرره اذا في
df_dil = df_dil.drop_duplicates()
print(dup)
```

```python
# Task 4: Write your code here:

#### نحدد الأعمدة
cat_cols = ["Weather", "Traffic_Level", "Time_of_Day", "Vehicle_Type"]
#### نستخدم الonHotEncoding
preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(drop="first", sparse_output=False), cat_cols)
    ],
    remainder="passthrough"
)
df_dil_encoded = preprocessor.fit_transform(df_dil)
```

```python
# Task 5: Write your code here:

### نسوي سكيلنق
scaler = StandardScaler()
### نسوي لها فيت
df_dil_scaled = scaler.fit_transform(df_dil_encoded)
```

```python
# Task 6: Write your code here:
plt.hist(df_dil["Delivery_Time"])
plt.xlabel("Delivery Time")
plt.ylabel("Frequency")
plt.title("Target Distribution Check")
plt.show()
###بناءً على البلوت فا لا يوجد امبلس داتا
```

```python
# Task 1: Write your code here:
X = df_dil_scaled
y = df_dil["Delivery_Time"]
```

```python
# Task 2,3,4,5: Write your code here:

###K-fold
kf = KFold(n_splits=5, shuffle=True, random_state=42)

mae_scores = []
## لووب عشان ك فولد
for train_idx, test_idx in kf.split(X):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

   ### الموديل
    model = RandomForestRegressor(random_state=42)
    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)
    mae = mean_absolute_error(y_test, y_pred)
    mae_scores.append(mae)

#  متوسطMAE
print("Average MAE across folds:", np.mean(mae_scores))
```

```python
# Task 1: Write your code here:

impor = model.feature_importances_
plt.bar(range(len(impor)), impor)
plt.xlabel("Features")
plt.title("Plot feature importance from trained model")
plt.show() ### نلاحظ بان الفيتشر رقم 14 هي اهم شي بناء على الموديل
```

```python
# Task 2: Write your code here:

y_pred = model.predict(X)
plt.hist(y_pred)
plt.xlabel("^yDelivery Time")
plt.title("Plot predicted delivery time histogram")
plt.show()
```
