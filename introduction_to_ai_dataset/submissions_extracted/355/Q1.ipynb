```python
# Task 1: Write your code here:
# Why droping the Order_ID because this column is not relative to or data or the target value
df_q1 = df_q1.drop(columns=['Order_ID'])
df_q1.head()
```

```python
# Task 2: Write your code here:
# Analyze missing values
missing_percentage = (df_q1.isnull().sum() / len(df_q1)) * 100
missing_data = pd.DataFrame({
    'Column': missing_percentage.index,
    'Missing_Percentage': missing_percentage.values
})
missing_data = missing_data[missing_data['Missing_Percentage'] > 0].sort_values('Missing_Percentage', ascending=False)

print("Missing Data Analysis:")
missing_data.head(10)
```

```python
# fill the numaric values
df_q1['Delivery_Time'] = df_q1['Delivery_Time'].fillna(df_q1['Delivery_Time'].mean())
df_q1['Courier_Experience_yrs'] = df_q1['Courier_Experience_yrs'].fillna(df_q1['Courier_Experience_yrs'].mean())

```

```python
# Cheking the Delivery_Time and Courier_Experience_yrs

if  df_q1['Delivery_Time'].isnull or  df_q1['Courier_Experience_yrs'].isnull:
  print("There is no missing data in Delivery_Time or Courier_Experience_yrs")
  print(df_q1['Delivery_Time'].isnull().sum())
  print(df_q1['Courier_Experience_yrs'].isnull().sum())
else:
    print("There is missing data in Delivery_Time or Courier_Experience_yrs")
    print(df_q1['Delivery_Time'].isnull().sum())
    print(df_q1['Courier_Experience_yrs'].isnull().sum())
```

```python
# fill the object  values
df_q1['Time_of_Day'] = df_q1['Time_of_Day'].fillna(df_q1['Time_of_Day'].mode()[0])
df_q1['Traffic_Level'] = df_q1['Traffic_Level'].fillna(df_q1['Traffic_Level'].mode()[0])
df_q1['Weather'] = df_q1['Weather'].fillna(df_q1['Weather'].mode()[0])
```

```python
if   df_q1['Time_of_Day'].isnull or df_q1['Traffic_Level'].isnull or df_q1['Weather'].isnull :
  print("There is missing data in Traffic_Level, Weather or Time_of_Day")
  print(df_q1['Time_of_Day'].isnull().sum())
  print(df_q1['Traffic_Level'].isnull().sum())
  print(df_q1['Weather'].isnull().sum())

else:
    print("There is missing data in Traffic_Level, Weather or Time_of_Day")
    print(df_q1['Time_of_Day'].isnull().sum())
    print(df_q1['Traffic_Level'].isnull().sum())
    print(df_q1['Weather'].isnull().sum())
```

```python
# Task 3: Write your code here:
# Check if there is a duplicates
duplicates = df_q1.duplicated().sum()
if duplicates > 0:
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df_q1.drop_duplicates(inplace=True) #Important
    duplicates = df_q1.duplicated().sum()
    print(f"Duplicates Dropped and the duplicates now is {duplicates}")
  else:
    print("No Duplicate Samples Found.")
```
