```python
#Handle missing values
def check_missing_values(df):
  missing_values = df.isnull().sum()
  print("Missing Values per Column:")
  print(missing_values[missing_values > 0])
  if missing_values.any():
    print("\nHandle Missing Values as needed.")
  else:
    print("\nNo Missing Values Found.")
check_missing_values(data)
```

```python
#Handle missing values
#for Categorical col
Categorical= data.select_dtypes(include=["object"]).columns
for col in list(Categorical):
  data[col]= data[col].fillna(data[col].mode()[0])


#for col Courier_Experience_yrs fil it with the mean becaues there is not outliers
data['Courier_Experience_yrs']=data['Courier_Experience_yrs'].fillna(data['Courier_Experience_yrs'].mean())

#for col Delivery_Time fil it with the median becaues there is outliers
data['Delivery_Time']=data['Delivery_Time'].fillna(data['Delivery_Time'].median())

#last cheack
def check_missing_values(df):
  missing_values = df.isnull().sum()
  print("Missing Values per Column:")
  print(missing_values[missing_values > 0])
  if missing_values.any():
    print("\nHandle Missing Values as needed.")
  else:
    print("\nNo Missing Values Found.")
check_missing_values(data)
```

```python
# Task 3: Write your code here:
#Check and remove duplicates
def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(data)
```

```python
# Task 1: Write your code here:
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score


X = data.drop("Delivery_Time", axis=1).astype(float)
y = data["Delivery_Time"].astype(float)
y
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import StratifiedKFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
n_splits = 5 # K

skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)

for fold_idx, (train_index, test_index) in enumerate(skf.split(X, y)):
  print(f"\nFold {fold_idx + 1}/{n_splits}")

  # 1. Split data
  X_train, X_test = X.iloc[train_index], X.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]

  # 2. Train & Validate the model
  model = RandomForestRegressor(n_estimators=100, max_depth=20, random_state=42, n_jobs=-1)
  model.fit(X_train, y_train) # train
  y_pred = model.predict(X_test) # validate
  mae = mean_absolute_error(y_test, y_pred)
print("MAE :", mae)
```

```python
# Task 1: Write your code here:
# Feature importance
feature_importance = pd.DataFrame({
    'feature': X,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

plt.figure(figsize=(10, 6))
plt.barh(feature_importance['feature'], feature_importance['importance'])
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show()
```
