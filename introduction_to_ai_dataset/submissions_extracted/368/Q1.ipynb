```python
# Task 1: Write your code here:
df_noid=df.drop('Order_ID',axis=1).copy()
df_noid.head()
```

```python
# Task 2: Write your code here:
# Analyze missing values
missing_percentage = (df_noid.isnull().sum() / len(df)) * 100
missing_data = pd.DataFrame({
    'Column': missing_percentage.index,
    'Missing_Percentage': missing_percentage.values
})
missing_data = missing_data[missing_data['Missing_Percentage'] > 0].sort_values('Missing_Percentage', ascending=False)

print("Missing Data Analysis:")
missing_data.head(10)
cols=["Weather","Traffic_Level","Time_of_Day"]
for miss in cols:
  df_noid[miss] = df_noid[miss].fillna("unknown")
for miss in ['Delivery_Time',"Courier_Experience_yrs"]:
  df_noid[miss] = df_noid[miss].fillna(df_noid[miss].mean())
```

```python
# Task 3: Write your code here:
df_noid.drop_duplicates(inplace=True)
```

```python
# Task 4: Write your code here:
df_noid.info()
cat_col = ["Weather","Traffic_Level","Time_of_Day","Vehicle_Type"]
for col in cat_col:
  le = LabelEncoder()
  df_noid[col] = le.fit_transform(df_noid[col])
```

```python
# Task 1: Write your code here:
#i dont know how to split after scaleing but i will do it like this
# Define features (X) and target (y)
print(df_noid)
X = df_noid.drop(columns=["Delivery_Time"]).copy()
y = df_noid["Delivery_Time"]
kf = KFold(n_splits=5, shuffle=True, random_state=42)
# StratifiedKFold can be used in both regrssion and classfcation but i will use kfold
for fold, (train_idx, test_idx) in enumerate(kf.split(X), start=1): # "start" for fold try changeing it ( not important) "enuyumrate " for indexing a vlue
    # indexing for each fold
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
```

```python
# Task 2,3,4,5: Write your code here:
model = RandomForestRegressor(n_estimators=100, max_depth=20, random_state=42, n_jobs=-1)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
mae = mean_absolute_error(y_test, y_pred)
print(f"MAE:  ${mae:,.2f}")
```

```python
# Task 1: Write your code here:
# Plot for Decision Tree Regression Predictions vs. Ground Truth
plt.figure(figsize=(6, 4))
plt.scatter(y_test, y_pred, alpha=0.7)
plt.plot([min(y_test), max(y_test)],
         [min(y_test), max(y_test)],
         'r--', linewidth=2)

plt.xlabel("Actual y_test (Ground Truth)")
plt.ylabel("Predicted y_pred (Decision Tree)")
plt.title("Decision Tree Regression: Predictions vs. Ground Truth")
plt.grid(True)

plt.tight_layout()
plt.show()
```
