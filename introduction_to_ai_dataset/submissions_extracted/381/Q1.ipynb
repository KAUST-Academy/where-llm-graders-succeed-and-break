```python
# Task 1: Write your code here:

df_data = df_data.drop(columns="Order_ID", axis=1)
```

```python
# Task 2: Write your code here:
# Analyze missing values
missing_percentage = (df_data.isnull().sum() / len(df_data)) * 100
missing_data = pd.DataFrame({
    'Column': missing_percentage.index,
    'Missing_Percentage': missing_percentage.values
})
missing_data = missing_data[missing_data['Missing_Percentage'] > 0].sort_values('Missing_Percentage', ascending=False)

print("Missing Data Analysis:")
missing_data.head(10)
```

```python
# Task 3: Write your code here:
cols = ['Distance_km','Weather','Traffic_Level','Time_of_Day','Vehicle_Type','Preparation_Time_min','Courier_Experience_yrs','Delivery_Time']
df_clean = df_data[cols].copy()
print(f"Before: {df_clean.shape}")
df_clean = df_clean.dropna(subset=['Delivery_Time','Weather','Traffic_Level','Time_of_Day','Courier_Experience_yrs'])
print(f"After dropping missing values: {df_clean.shape}")

df_clean.info()
```
