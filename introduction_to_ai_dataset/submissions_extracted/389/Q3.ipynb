```python
# Task 1: Write your code here:
def check_missing_values(df):
  missing_values = df.isnull().sum()
  print("Missing Values per Column:")
  print(missing_values[missing_values > 0])
  if missing_values.any():
    print("\nHandle Missing Values as needed.")
  else:
    print("\nNo Missing Values Found.")

check_missing_values(df)
df = df.drop('D_142', axis=1)
df = df.drop('S_3', axis=1)
for col in df.columns:
    if df[col].dtype in ['int64', 'float64']:
        # Impute numerical features with Mean
        df[col] = df[col].fillna(df[col].mean())
    else:
        # Impute categorical features with Mode
        # [0] is used because .mode() returns a Series
        if not df[col].mode().empty:
            df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# Task 2: Write your code here:
def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df)
df
```

```python
# Task 2,3,4,5: Write your code here:

import pandas as pd
import numpy as np
from sklearn.model_selection import KFold, cross_val_score
from sklearn.preprocessing import StandardScaler
from catboost import CatBoostClassifier

# 1. Impute missing values
# Loop through columns: Mean for numerical, Mode for categorical
for col in df.columns:
    if df[col].dtype in ['int64', 'float64']:
        df[col] = df[col].fillna(df[col].mean())
    else:
        if not df[col].mode().empty:
            df[col] = df[col].fillna(df[col].mode()[0])

# 2. Feature Scaling (StandardScaler)
num_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
if 'target' in num_cols: num_cols.remove('Target') # Exclude target from scaling

scaler = StandardScaler()
df[num_cols] = scaler.fit_transform(df[num_cols])

# 3. Check for target imbalance
target_counts = df['Target'].value_counts(normalize=True)
minority_class_pct = target_counts.min()

# Define status and metric
# If minority class is < 20%, we consider it imbalanced and use F1-Score
if minority_class_pct < 0.20:
    imbalance_status = "IMBALANCED"
    metric = 'f1' # Or 'f1_weighted' if multiclass
    print(f"Status: {imbalance_status} (Minority class: {minority_class_pct:.2%})")
    print(f"Appropriate Metric: F1-Score")
else:
    imbalance_status = "BALANCED"
    metric = 'accuracy'
    print(f"Status: {imbalance_status} (Minority class: {minority_class_pct:.2%})")
    print(f"Appropriate Metric: Accuracy")

# 4. Split into features (X) and target (y)
X = df.drop('Target', axis=1)
y = df['Target']

# Identify categorical feature indices for CatBoost
cat_features = X.select_dtypes(exclude=[np.number]).columns.tolist()

# 5. Setup Normal KFold
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# 6. Train CatBoostClassifier
# iterations=100 for speed; increase for better performance
model = CatBoostClassifier(verbose=0, iterations=100, cat_features=cat_features)

# 7. Evaluate and Print averaged score
scores = cross_val_score(model, X, y, cv=kf, scoring=metric)

print("-" * 30)
print(f"Average {metric.capitalize()} across 5 folds: {scores.mean():.4f}")
```
