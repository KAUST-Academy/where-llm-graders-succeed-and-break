```python
# Task 1: Write your code here:
df_dropped_id = df.drop('Order_ID',axis=1)
```

```python
# Task 2: Write your code here:
n =df_dropped_id.isnull().sum()
print(n)
#i will fill every thing with the 0
df['Traffic_Level'] = df['Traffic_Level'].fillna(df['Traffic_Level'].mode()[0])
df['Time_of_Day'] = df['Time_of_Day'].fillna(df['Time_of_Day'].mode()[0])
df['Courier_Experience_yrs'] = df['Courier_Experience_yrs'].fillna(df['Courier_Experience_yrs'].mode()[0])
df['Delivery_Time'] = df['Delivery_Time'].fillna(df['Delivery_Time'].mode()[0])


df.head()

df.info()
```

```python
# Task 3: Write your code here:
def check_duplicates(df):

  # Checks the duplicates in the dataframe
  duplicates = df.duplicated().sum()

  if duplicates > 0:
        print("There are duplicates: " + str(duplicates))
        df=df.drop_duplicates()
        duplicates = df.duplicated().sum()

        print("duplicates after handling: " + str(duplicates))
  else:
        print("There are no duplicates")

check_duplicates(df)

```

```python
from sklearn.preprocessing import LabelEncoder

# To find Categorical columns:
categorical_cols = df.select_dtypes(include=["object"]).columns

# There are three, we will use LabelEncoder() for each of them:
for categorical in categorical_cols:
  le = LabelEncoder()
  df[categorical] = le.fit_transform(df[categorical])


df.head()
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler
import numpy as np

# Defining the scaler
scaler = StandardScaler()


# Choosing the numerical columns
numerical_cols = df.select_dtypes(include=["Float64","int64"]).columns.drop("Delivery_Time")
df[numerical_cols] = scaler.fit_transform(df[numerical_cols])

df.head()
```

```python
# Task 6: Write your code here:
import seaborn as sns
# Takes in the dataframe + the target column which we will check
def imbalance_func(df, target_col):

    # checking the number of entries of each class in the target
    print(df[target_col].value_counts(normalize=True))

    #Visualizing the results
    sns.countplot(x=df[target_col])


imbalance_func(df, "Delivery_Time")
```

```python
# Task 1: Write your code here:
import numpy as np
X = df.drop("Delivery_Time", axis=1)
y = df["Delivery_Time"]
print(np.shape(X))
print(np.shape(y))
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import KFold, StratifiedKFold, cross_val_score
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
import numpy as np

scores_time = []
scores_avg =0

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

for train_index, test_index in skf.split(X, y):
    # Split data into training and testing sets using the indices defined by the KFold
    X_Train, X_Test = X.loc[train_index, :], X.loc[test_index, :]
    y_Train, y_Test = y.iloc[train_index], y.iloc[test_index]

    # Train the model on the training data
    model = RandomForestRegressor(n_estimators=150, max_depth=20, random_state=42).fit(X_Train, y_Train)

    # Predict on the test set
    y_pred = model.predict(X_Test)

    # Metric from the sklearn.metrics package
    scores_time.append(mean_absolute_error(y_Test, y_pred))
scores_avg= np.mean(scores_time)
print(f"MAE: {(scores_avg):.2f}")
```

```python
import matplotlib.pyplot as plt
feature_importance = pd.DataFrame({
    'feature': X.columns,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=True)

plt.figure(figsize=(10, 8))
plt.barh(feature_importance['feature'], feature_importance['importance'])
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.show()
```
