```python
# Task 1: Write your code here:
df = df.drop(columns="Order_ID", axis=1)
```

```python
# Task 2: Write your code here:
def check_missing_values(df):
  missing_values = df.isnull().sum()
  print("Missing Values per Column:")
  print(missing_values[missing_values > 0])
  if missing_values.any():
    print("\nHandle Missing Values as needed.")
  else:
    print("\nNo Missing Values Found.")

check_missing_values(df)
```

```python
#Task 2 countinue
#we will select all the cat data and fill it with the "unknown"
#if the acc bad we will change it mode or etc
for col in df.select_dtypes(include=["object"]).columns:
      df[col] = df[col].fillna('unknown')


df.head()
```

```python
import numpy as np

num_col=df.select_dtypes(include=["number"]).columns
df[num_col]=df[num_col].fillna(np.mean(df[num_col]))
check_missing_values(df)
```

```python
df.head()
```

```python
# Task 3: Write your code here:
def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df)
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()

for col in df.select_dtypes(include=["object"]).columns:
    df[col] = le.fit_transform(df[col].astype(str))

df.head()
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler
Scaler=StandardScaler()
features = df.columns.drop("Delivery_Time")

df[features]=Scaler.fit_transform(df[features])
df.head()
```

```python
feature=df.columns.drop("Delivery_Time")
X=df[feature]
y=df["Delivery_Time"]
```

```python
# Task 2,3,4,5: Write your code here:
# split ratio (80% train, 20% test)
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,        # 20% for test, remaining 80% for train
    random_state=42,      # reproducible output
    shuffle=True,         # representative splits
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error


kf = KFold(n_splits=5, shuffle=True, random_state=42)
model=RandomForestRegressor()
mae_scores = []
# Iterate through folds
for fold, (train_idx, test_idx) in enumerate(kf.split(X), start=1):
    # indexing for each fold
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    # Train model
    model.fit(X_train, y_train)

    # Predict
    y_pred = model.predict(X_test)

    # Evaluation metrics
    mae_scores.append(mean_absolute_error(y_test, y_pred))


```

```python
# Print Evaluation Metrics
print("\nModel Evaluation Metrics (K-Fold)\n" + "-"*40)
print(f"MAE : {np.mean(mae_scores):.2f}")
print("-"*40)
```
