```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import kagglehub
import os
from tqdm import tqdm
from sklearn.preprocessing import OneHotEncoder

%matplotlib inline
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("mohammad2012191/q1-ka-ai-2026")

print("Path to dataset files:", path)
```

```python
csv_path = os.path.join(path, "Q1_data.csv") # joins path with Q1_data.csv  taht we found downloaded in our folder

df = pd.read_csv(csv_path)
```

```python

def check_target_distribution(df, target_column): # function is reusable .. use when hit is needed
  df[target_column].hist(bins=30, edgecolor='black')

  plt.title(f"Target Distribution ({target_column})")
  plt.xlabel(target_column)
  plt.ylabel("Frequency")
  plt.grid(False)

  plt.show()

check_target_distribution(df, "Delivery_Time")
```

```python
df.drop(columns=['Order_ID'])
```

```python
def check_missing_values(df):
  missing_values = df.isnull().sum()
  print("Missing Values per Column:")
  print(missing_values[missing_values > 0])
  if missing_values.any():
    print("\nHandle Missing Values as needed.")
  else:
    print("\nNo Missing Values Found.")

check_missing_values(df)

#handling, filling target column and dropping rows where target (delivery time) or key features are missing

cols = ['Distance_km', 'Weather', 'Traffic_Level', 'Time_of_Day', 'Vehicle_Type', 'Preparation_Time_min', 'Courier_Experience_yrs', 'Delivery_Time']
df_clean = df[cols].copy()
print(f"Before: {df[cols].shape}")
cols = df_clean.dropna(subset=['Weather', 'Traffic_Level', 'Time_of_Day', 'Courier_Experience_yrs'])
print(f"After dropping missing Weather, Traffic_Level, Time_of_Day, Courier_Experience_yrs: {df[cols].shape}")

cols['Delivery_Time'] = cols['Delivery_Time'].fillna(cols['Delivery_Time'].mode()[0])


print("Missing values remaining:", cols.isnull().sum())
```

```python
def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True) # changes in original data
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df)

```

```python
categorical_cols = df.select_dtypes(include=["object"]).columns

print("Categorical Columns:", list(categorical_cols))

ohe = OneHotEncoder(sparse_output=False)
encoded = ohe.fit_transform(df[categorical_cols])

print('\nData after encoding:\n', encoded)

print(df['Weather'].dtype)
```

```python
from sklearn.preprocessing import StandardScaler

features = df.columns.drop("Delivery_Time")
scaler = StandardScaler()
df[features] = scaler.fit_transform(df[features])
df.head()
```
