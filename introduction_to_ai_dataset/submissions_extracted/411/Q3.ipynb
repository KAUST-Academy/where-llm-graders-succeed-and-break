```python
import numpy as np
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import f1_score, accuracy_score
from catboost import CatBoostClassifier
```

```python
# Task 2,3,4,5: Write your code here:
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Lists to store scores
f1_scores = []

print(f"Starting Cross-Validation with {skf.get_n_splits()} folds...\n")

fold = 1
for train_index, val_index in skf.split(X, y):

    X_train_fold, X_val_fold = X[train_index], X[val_index]
    y_train_fold, y_val_fold = y[train_index], y[val_index]



    model = CatBoostClassifier(iterations=500, learning_rate=0.1, depth=6, verbose=0, random_state=42)


    model.fit(X_train_fold, y_train_fold)

    # predict
    y_pred = model.predict(X_val_fold)

    #  f1
    score = f1_score(y_val_fold, y_pred)
    f1_scores.append(score)

    print(f"Fold {fold} F1 Score: {score:.4f}")
    fold += 1

# scors
avg_f1 = np.mean(f1_scores)
print("-" * 30)
print(f"Average F1 Score: {avg_f1:.4f}")
print("-" * 30)
```

```python
# Task 1: Write your code here:
feature_importances = model.get_feature_importance()
feature_names = X_scaled.columns


feat_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': feature_importances
})

feat_df = feat_df.sort_values(by='Importance', ascending=False)

plt.figure(figsize=(10, 6))
# top 10 featurs
top = feat_df.head(10)
plt.barh(top['Feature'], top['Importance'], color='gold')
plt.xlabel('Importance Score')
plt.title('Top 10 Most Important Features')
plt.gca().invert_yaxis() # Put the #1 feature at the top
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```
