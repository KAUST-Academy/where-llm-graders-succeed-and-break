```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import KFold

kf = KFold(n_splits=5, shuffle=True, random_state=42)

rf_model = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)

mae_scores = []

for fold, (train_idx, val_idx) in enumerate(kf.split(X_scaled), 1):
    X_train, X_val = X_scaled.iloc[train_idx], X_scaled.iloc[val_idx]
    y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

    rf_model.fit(X_train, y_train)

    y_pred = rf_model.predict(X_val)

    mae = mean_absolute_error(y_val, y_pred)
    mae_scores.append(mae)

    print(f"Fold {fold} - MAE: {mae:.4f}")

print(f"\n{'='*50}")
print(f"Average MAE across all folds: {np.mean(mae_scores):.4f}")
print(f"Standard deviation: {np.std(mae_scores):.4f}")
print(f"{'='*50}")
```

```python
# Task Bonus: Write your code here:
from catboost import CatBoostRegressor

kf = KFold(n_splits=5, shuffle=True, random_state=42)

rf_model = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
cb_model = CatBoostRegressor(iterations=100, random_state=42, verbose=0)

ensemble_mae_scores = []
rf_mae_scores = []
cb_mae_scores = []

for fold, (train_idx, val_idx) in enumerate(kf.split(X_scaled), 1):

    X_train, X_val = X_scaled.iloc[train_idx], X_scaled.iloc[val_idx]
    y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

    rf_model.fit(X_train, y_train)
    rf_pred = rf_model.predict(X_val)
    rf_mae = mean_absolute_error(y_val, rf_pred)
    rf_mae_scores.append(rf_mae)

    cb_model.fit(X_train, y_train)
    cb_pred = cb_model.predict(X_val)
    cb_mae = mean_absolute_error(y_val, cb_pred)
    cb_mae_scores.append(cb_mae)

    ensemble_pred = (rf_pred + cb_pred) / 2

    ensemble_mae = mean_absolute_error(y_val, ensemble_pred)
    ensemble_mae_scores.append(ensemble_mae)

    print(f"Fold {fold}:")
    print(f"  RandomForest MAE: {rf_mae:.4f}")
    print(f"  CatBoost MAE: {cb_mae:.4f}")
    print(f"  Ensemble MAE: {ensemble_mae:.4f}")
    print()

print(f"{'='*60}")
print(f"Average RandomForest MAE: {np.mean(rf_mae_scores):.4f} (+/- {np.std(rf_mae_scores):.4f})")
print(f"Average CatBoost MAE: {np.mean(cb_mae_scores):.4f} (+/- {np.std(cb_mae_scores):.4f})")
print(f"Average Ensemble MAE: {np.mean(ensemble_mae_scores):.4f} (+/- {np.std(ensemble_mae_scores):.4f})")
print(f"{'='*60}")

plt.figure(figsize=(10, 6))
models = ['RandomForest', 'CatBoost', 'Ensemble']
avg_scores = [np.mean(rf_mae_scores), np.mean(cb_mae_scores), np.mean(ensemble_mae_scores)]
colors = ['blue', 'orange', 'green']

plt.bar(models, avg_scores, color=colors, alpha=0.7, edgecolor='black')
plt.ylabel('Mean Absolute Error')
plt.title('Model Performance Comparison')
plt.grid(True, alpha=0.3, axis='y')
for i, v in enumerate(avg_scores):
    plt.text(i, v + 0.01, f'{v:.4f}', ha='center', va='bottom', fontweight='bold')
plt.tight_layout()
plt.show()
```
