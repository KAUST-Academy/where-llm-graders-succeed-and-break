```python
# Task 1: Write your code here:
df_copy = df.copy()
df_copy.drop(columns=['Order_ID'], inplace= True)
df_copy.head(10)
```

```python
# Task 2: Write your code here:
df_copy.dropna(subset=['Delivery_Time'], inplace = True)  #Dropping rows from Delivery_Time column
df_copy['Weather'].fillna(df['Weather'].mode()[0], inplace = True)
df_copy['Traffic_Level'].fillna(df['Traffic_Level'].mode()[0], inplace = True)
df_copy['Time_of_Day'].fillna(df['Time_of_Day'].mode()[0], inplace = True)
df_copy['Courier_Experience_yrs'].fillna(df['Courier_Experience_yrs'].mode()[0], inplace = True)

df_copy.isnull().sum()
```

```python
# Task 3: Write your code here:
duplicates = df_copy.duplicated().sum()
print(f"Number of duplicate rows: {duplicates}")

if duplicates > 0:
    df_copy.drop_duplicates(inplace=True)
    print(f"Removed {duplicates} duplicate rows")
else:
    print("No duplicates found!")
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import LabelEncoder
df_copy[['Weather','Traffic_Level','Time_of_Day','Vehicle_Type']]

for col in ['Weather','Traffic_Level','Time_of_Day','Vehicle_Type']:
    if col in df_copy.columns:
        unique_count = df_copy[col].nunique()
        print(f"  {col}: {unique_count} unique values")


le = LabelEncoder()
df_copy['Weather'] = le.fit_transform(df_copy['Weather'])
df_copy['Traffic_Level'] = le.fit_transform(df_copy['Traffic_Level'])
df_copy['Time_of_Day'] = le.fit_transform(df_copy['Time_of_Day'])
df_copy['Vehicle_Type'] = le.fit_transform(df_copy['Vehicle_Type'])

df_copy.head(10)
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler
ss = StandardScaler()

features = ['Distance_km','Weather','Traffic_Level','Time_of_Day','Vehicle_Type','Preparation_Time_min','Courier_Experience_yrs']


print('data before scaling:\n', df_copy[features]) # before scaling
df_copy[features] = ss.fit_transform(df_copy[features]) # Applying fit_transform to the copied
print('\nData after scaling:\n', df_copy[features]) # after scaling


```

```python
# Task 1: Write your code here:
from sklearn.model_selection import KFold
feature_cols = ['Distance_km','Weather','Traffic_Level','Time_of_Day','Vehicle_Type','Preparation_Time_min','Courier_Experience_yrs']

X = df_copy[feature_cols]
y = df_copy['Delivery_Time']
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

kf = KFold(n_splits=5, shuffle=True, random_state=42)

for fold, (train_idx, test_idx) in enumerate(kf.split(X), start=1):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]


rf = RandomForestRegressor(n_estimators=100, max_depth=5, random_state=42)
rf.fit(X_train, y_train)
print("Random Forest R²:", rf.score(X_test, y_test))


y_pred = rf.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)










```
