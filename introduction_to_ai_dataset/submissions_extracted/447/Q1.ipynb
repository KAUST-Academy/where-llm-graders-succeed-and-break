```python
# Task 1: Write your code here:
df = df.drop(['Order_ID'], axis=1)
```

```python
# Task 2: Write your code here:

#first we check for missing values per column
def check_missing_values(df):
  missing_values = df.isnull().sum()
  print("Missing Values per Column:")
  print(missing_values[missing_values > 0])
  if missing_values.any():
    print("\nthere are missing values")
  else:
    print("\nNo Missing Values Found.")

check_missing_values(df)



```

```python
#subtask 2
# they are : [Weather, Traffic_Level, Time_of_Day, Courier_Experience_yrs]
df['Courier_Experience_yrs'] = df['Courier_Experience_yrs'].fillna(df['Courier_Experience_yrs'].median())
df['Delivery_Time'] = df['Delivery_Time'].fillna(df['Delivery_Time'].median())
df['Weather'] = df['Weather'].fillna(df['Weather'].median())
df['Traffic_Level'] = df['Traffic_Level'].fillna(df['Traffic_Level'].median())
df['Time_of_Day'] = df['Time_of_Day'].fillna(df['Time_of_Day'].median())
check_missing_values(df)
```

```python
# Task 3: Write your code here:

#check dublicates
duplicates = df.duplicated().sum()
print(f"Duplicates: {duplicates}")

#remove them
df.drop_duplicates(inplace=True)
df.info()



```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder() # Instantiate LabelEncoder

df['Weather'] = label_encoder.fit_transform(df['Weather'])
df['Traffic_Level'] = label_encoder.fit_transform(df['Traffic_Level'])
df['Time_of_Day'] = label_encoder.fit_transform(df['Time_of_Day'])
df['Vehicle_Type'] = label_encoder.fit_transform(df['Vehicle_Type'])
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import MinMaxScaler

features = df.columns.drop("Delivery_Time")  # we don't scale the target

scaler = MinMaxScaler()
df[features] = scaler.fit_transform(df[features])
df.head()
```

```python
# Task 6: Write your code here:
df['Delivery_Time'].hist()
#data appears to be imbalancced
```

```python
# Task 1: Write your code here:
from sklearn.model_selection import StratifiedKFold, train_test_split
X = df.drop(['Delivery_Time'], axis=1)
y = df['Delivery_Time']
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error as  mean_absolute_error

#initialize the model
model = RandomForestRegressor(n_estimators=100, max_depth=5, random_state=42)

#initialize k-fold
kf = KFold(n_splits=5, shuffle=True, random_state=42)


# Iterate through folds
for fold, (train_idx, test_idx) in enumerate(kf.split(X), start=1):
    # indexing for each fold
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    model.fit(X_train, y_train) #fit
    y_pred = model.predict(X_test) #prediction
    mae = mean_absolute_error(y_test, y_pred) # absolute erro
    print("Random Forest MAE:", mae)
#1 hour left... guess those errors are going to stay

```
