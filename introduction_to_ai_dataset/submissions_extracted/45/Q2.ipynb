```python
# Task 1: Write your model class here:
class HAR_Net(nn.Module):
    def __init__(self, num_classes):
        super(HAR_Net, self).__init__()

        # Layer 1: Convolution
        self.conv1 = nn.Conv2d(
            in_channels=X.shape[1],   # عدد القنوات (WiFi channels)
            out_channels=16,
            kernel_size=3,
            padding=1
        )

        # Layer 2: Activation
        self.relu = nn.ReLU()

        # Layer 3: Pooling
        self.pool = nn.MaxPool2d(2)

        # Layer 4: Fully Connected
        self.fc1 = nn.Linear(16 * (X.shape[2]//2) * (X.shape[3]//2), 64)

        # Layer 5: Output Layer
        self.fc2 = nn.Linear(64, num_classes)

    def forward(self, x):
        x = self.conv1(x)
        x = self.relu(x)
        x = self.pool(x)

        x = x.view(x.size(0), -1)
        x = self.relu(self.fc1(x))
        x = self.fc2(x)

        return x
```
