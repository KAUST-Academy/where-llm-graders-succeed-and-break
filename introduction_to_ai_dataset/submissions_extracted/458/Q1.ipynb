```python
# Task 1: Write your code here:
# Define features (X) and target (y)
feature_cols = ['Order_ID',	'Distance_km',	'Weather',	'Traffic_Level',	'Time_of_Day',	'Vehicle_Type',
                'Preparation_Time_min	', 'Courier_Experience_yrs' ]
X = df_food_delivery[feature_cols]
y = df_food_delivery['Delivery_Time']
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import KFold
# Define K-Fold Cross Validation
kf = KFold(n_splits=5, shuffle=True, random_state=42) #shuffle=true is to make sure the data is not in the same order they were and now it is random

# Iterate through folds
for fold, (train_idx, test_idx) in enumerate(kf.split(X), start=1):
    # indexing for each fold
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
    # print shapes
    print(f"Fold {fold}")
    print("  X_train shape:", X_train.shape)
    print("  X_test shape :", X_test.shape)
    print("  y_train shape:", y_train.shape)
    print("  y_test shape :", y_test.shape)
    print("-" * 30)

from sklearn.ensemble import RandomForestClassifier
# Train RandomForestClassifier
model = RandomForestClassifier(n_estimators=100, max_depth=15,
                               class_weight='balanced', random_state=42)
model.fit(X_train, y_train)
print("Model trained!")


from sklearn.metrics import mean_absolute_error
y_pred = model.predict(X_test) #here er put x test inside the model to make a prediction and the result of the prediction will be store in y_pred

  mae = mean_absolute_error(y_test, y_pred)
  print("MAE :", mae)


print(avg=kf/n_splits=5)
```
