```python
# Task 1: Write your code here:
# Show missing values before
print(df.isnull().sum())

# Fill missing values
df.fillna(df.mean(), inplace=True)

# Show after
print(df.isnull().sum())
```

```python
# Task 2: Write your code here:

# Show number of duplicates before
print("Duplicates before:", df.duplicated().sum())

# Remove duplicates
df.drop_duplicates(inplace=True)

# Show after
print("Duplicates after:", df.duplicated().sum())
```

```python
# Task 3: Write your code here:
# Show data types before encoding
print(df.dtypes)

# Encode categorical variables
df = pd.get_dummies(df, drop_first=True)

# Show data types after encoding
print(df.dtypes)
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import StandardScaler

# Show data before scaling
print(df.head())

# Apply scaling
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)

# Show data after scaling
print(df_scaled[:5])
```

```python
# Task 1: Write your code here:
X = df.drop(df.columns[-1], axis=1)
y = df.iloc[:, -1]
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score
from catboost import CatBoostClassifier
import numpy as np

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

scores = []

for train_idx, test_idx in skf.split(X, y):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    model = CatBoostClassifier(verbose=0)
    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    scores.append(acc)

print("Average Accuracy:", np.mean(scores) * 100, "%")
```

```python
# Task 1: Write your code here:
import matplotlib.pyplot as plt
import numpy as np

importances = model.get_feature_importance()

# Get top 10 features
idx = np.argsort(importances)[-10:]

plt.barh(X.columns[idx], importances[idx])
plt.xlabel("Importance")
plt.ylabel("Features")
plt.show()

```

```python
# Task 2: Write your code here:
golden_feature = X.columns[importances.argmax()]
print("Golden Feature:", golden_feature)
```

```python
# Task Bonus: Write your code here:
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score
from catboost import CatBoostClassifier
import numpy as np

# Use only the golden feature
X_golden = X[[golden_feature]]

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

scores_golden = []

for train_idx, test_idx in skf.split(X_golden, y):
    X_train, X_test = X_golden.iloc[train_idx], X_golden.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    model = CatBoostClassifier(verbose=0)
    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    scores_golden.append(acc)

print("Golden Feature Accuracy:", np.mean(scores_golden) * 100, "%")
```
