```python
# Task 1: Write your code here:
from sklearn.preprocessing import StandardScaler


# 1) Drop Order_ID column if exists
if "Order_ID" in df.columns:
    df = df.drop(columns=["Order_ID"])
    print("Order_ID column dropped")
else:
    print("Order_ID column not found")

```

```python
# Task 2: Write your code here:
missing_cols = df.columns[df.isnull().any()]

if len(missing_cols) > 0:
    print("Handling missing values in columns:", list(missing_cols))
    for col in missing_cols:
        if df[col].dtype == "object":
            df[col] = df[col].fillna(df[col].mode()[0])
        else:
            df[col] = df[col].fillna(df[col].median())
else:
    print("No missing values found")
```

```python
# Task 3: Write your code here:

# 3) Check & remove duplicates

dup_count = df.duplicated().sum()
if dup_count > 0:
    df = df.drop_duplicates()
    print(f"Removed {dup_count} duplicate rows")
else:
    print("No duplicate rows found")
```

```python
# Task 4: Write your code here:

# 4) Encode categorical variables (One-Hot Encoding)

cat_cols = df.select_dtypes(include=["object"]).columns
if len(cat_cols) > 0:
    df = pd.get_dummies(df, columns=cat_cols, drop_first=True)
    print("Categorical variables encoded using One-Hot Encoding")
else:
    print("No categorical variables to encode")
```

```python
# Task 5: Write your code here:
# 5) Feature scaling (StandardScaler)

X = df.drop(columns=['Delivery_Time'])
y = df['Delivery_Time']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Feature scaling applied using StandardScaler")
print("Final feature shape:", X_scaled.shape)
```

```python
# Task 6: Write your code here:
```

```python
# Task 1: Write your code here:
from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
import numpy as np

# 1) Split features and target
X = X_scaled
y = y.values

print("Features and target prepared")
print("X shape:", X.shape)
print("y shape:", y.shape)
print("-" * 40)
```

```python
# Task 2,3,4,5: Write your code here:

# 2) KFold (Regression → NOT Stratified)
kf = KFold(n_splits=5, shuffle=True, random_state=42)

mae_scores = []

# 3) Train RandomForest + 4) Evaluate using MAE only
for fold, (train_idx, test_idx) in enumerate(kf.split(X), 1):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    model = RandomForestRegressor(random_state=42)
    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)
    mae = mean_absolute_error(y_test, y_pred)
    mae_scores.append(mae)

    print(f"Fold {fold} MAE:", round(mae, 2))

print("-" * 40)


# 5) Average MAE across folds
avg_mae = np.mean(mae_scores)
print("Average MAE across all folds:", round(avg_mae, 2))
```

```python
# Task 1: Write your code here:
import matplotlib.pyplot as plt
import numpy as np


# Use the last trained RandomForest model from Part 3
final_model = model

# Task 1: Feature Importance Plot

feature_names = X.columns if hasattr(X, "columns") else df.drop(columns=['Delivery_Time']).columns
importances = final_model.feature_importances_

plt.figure()
plt.barh(feature_names, importances)
plt.title("Feature Importance (RandomForest)")
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.show()



```

```python
# Task 2: Write your code here:
y_pred_all = final_model.predict(X)

plt.figure()
plt.hist(y_pred_all, bins=20)
plt.title("Predicted Delivery Time Distribution")
plt.xlabel("Predicted Delivery Time")
plt.ylabel("Frequency")
plt.show()
```
