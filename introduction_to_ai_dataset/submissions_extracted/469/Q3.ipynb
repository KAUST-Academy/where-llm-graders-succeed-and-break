```python
# Task 1: Write your code here:
# Task 2: Write your code here:
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0]) # not so many missing values so we drop them
df.dropna(inplace=True,axis=0)
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0])
```

```python
# Task 2: Write your code here:
duplicates = df.duplicated().sum()
print(f"number of Duplicate Samples: {duplicates}")
df.drop_duplicates(inplace=True)
duplicates = df.duplicated().sum()
print(f"number of Duplicate Samples: {duplicates}")
```

```python
# Task 3: Write your code here:
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
categorical_cols = df.select_dtypes(include=["object"]).columns
print("Categorical Columns:", list(categorical_cols))

if len(categorical_cols >= 1):
  le = LabelEncoder()
  df[categorical_cols] = le.fit_transform(df[categorical_cols])
  categorical_cols = df.select_dtypes(include=["object"]).columns
  print("Categorical Columns:", list(categorical_cols))
else:
  print("no categorical cols")
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import StandardScaler
standard_scaler = StandardScaler()
num_cols = df.drop('Target').select_dtypes(include=["float64","int64"]).columns
df[num_cols] = standard_scaler.fit_transform(df[num_cols])
print(num_cols)
```

```python
# Task 5: Write your code here:
df['Target'].hist()
```

```python
# Task 1: Write your code here:
from sklearn.model_selection import train_test_split
y = df['Target']
x = df.drop(['Target'],axis=1)
```

```python
# Task 2,3,4,5: Write your code here:
# Task 2,3,4,5: Write your code here:
# 2.kfold
from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
import numpy as np

model = RandomForestRegressor()

kfold = KFold(n_splits=5, shuffle=True, random_state=42)
mae_scores = []
for i, (train_idx, val_idx) in enumerate(kfold.split(x),start=1):
 X_fold_train, X_fold_test = x.iloc[train_idx], x.iloc[val_idx]
 y_fold_train, y_fold_val = y.iloc[train_idx], y.iloc[val_idx]
 # Train and test
 model.fit(X_fold_train, y_fold_train)
 y_fold_pred = model.predict(X_fold_test)
 # Calculate metrics
 mae = mean_absolute_error(y_fold_val, y_fold_pred)
 print(f"fold: {i}, mae: {mae}")
 mae_scores.append(mae)
mae_scores = np.array(mae_scores)
print(f"5-Fold CV Results:")
print(f"avg MAE: ${mae_scores.mean():,.2f}")
```
