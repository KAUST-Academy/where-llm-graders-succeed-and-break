```python
# Task 4: Write your code here:

catagorical_cols = ['Weather', 'Traffic_Level', 'Time_of_Day', 'Vehicle_Type']

encoder = LabelEncoder()

for col in catagorical_cols:
  df_missing_droped[col] = encoder.fit_transform(df_missing_droped[col])

df_missing_droped.head()
df_missing_droped.info()
```

```python
# Task 5: Write your code here:
features_except_target = catagorical_cols + ['Distance_km', 'Preparation_Time_min', 'Courier_Experience_yrs']

scaler = StandardScaler()
df_features_scaled = scaler.fit_transform(df_missing_droped[features_except_target])
```

```python
print(type(df_missing_droped))
```

```python
# Task 6: Write your code here:
#imblanced is not possible here, the target feture is continuous not classes
```

```python
# Task 1: Write your code here:



X = df_features_scaled
y = df_missing_droped['Delivery_Time']
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.metrics import mean_absolute_error, mean_squared_error

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size =0.2)

# model train and inatialize
model = RandomForestClassifier(n_estimators=100, max_depth=15, class_weight='balanced', random_state=42)
model.fit(X_train, y_train)

# model evaluate
outputs = model.predict(X_test)
mae = mean_absolute_error(y_test, outputs)
```

```python
# Task 1: Write your code here:

importance = pd.DataFrame({
    'feature': ['Distance_km', 'Preparation_Time_min', 'Courier_Experience_yrs'],
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

plt.figure(figsize=(10, 6))
plt.barh(importance['feature'], importance['importance'], color='purple')
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show()
```
