```python
# Task 5: Write your code here:
import matplotlib.pyplot as plt
def check_target_distribution(df, target_column): #defines a function that takes a DataFrame (df) and the name of the target_column
  df[target_column].hist(bins=30, edgecolor='black') # plots a histogram of the target column

  plt.title(f"Target Distribution ({target_column})") # sets the title of the plot
  plt.xlabel(target_column) # labels the horizontal axis (X) with the name of the target variable.
  plt.ylabel("Frequency") #labels the vertical axis (Y) to show the frequency "how many times each value range appears"
  plt.grid(False) #disables the background grid lines to keep the visualization clean

  plt.show()

check_target_distribution(df, "Delivery_Time") #calling for the function
```

```python
# Task 2: Write your code here:
def check_missing_values(df): #difining a missing value function
  missing_values = df.isnull().sum() #df.isnull a built-in function that checkes for non values data, sum is also a bult in function that sums up how many values that are missing in this context
  print("Missing Values per Column:")
  print(missing_values[missing_values > 0])  #Filters 'missing_values' to show only columns with a count greater than zero and prints them
  if missing_values.any():# Checks if there is at least one True value (any missing data) in the 'missing_values' series.
    print("\nHandle Missing Values as needed.")
  else:
    print("\nNo Missing Values Found.")

check_missing_values(df) #calling on the function to perform missing value analysis on the DataFrame 'df'
print("Missing values remaining:", df.isnull().sum().sum()) #Calculates the total count of all missing entries in the entire DataFrame and prints the final summation "sum().sum()
```

```python
# Task 3: Write your code here:
def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")
check_duplicates(df)
```

```python
# Task 4: Write your code here:
categorical_features = df.select_dtypes(include=['object']).columns

# Apply one-hot encoding to these categorical features using pd.get_dummies
df = pd.get_dummies(df, columns=categorical_features, drop_first=False)

df = df.astype(float)
df.head()
```

```python
# Task 1: Write your code here:
X = df.drop("Delivery_Time", axis=1).astype(float)
y = df['Delivery_Time'].astype(float)
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error as sklearn_mae
import numpy as np

n_splits = 5  # K=5 Folds
lr_mae = []
y_pred_list=[]
# 5-Fold Cross-Validation, shuffled
kf = KFold(n_splits=n_splits, shuffle=True, random_state=42)
for fold_idx, (train_index, test_index) in enumerate(kf.split(X)):
  print(f"\nFold {fold_idx + 1}/{n_splits}")

  X_train, X_test = X.iloc[train_index], X.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]
  model= RandomForestRegressor(n_estimators=200)
  model.fit(X_train, y_train)
  y_pred = model.predict(X_test)
  mae = sklearn_mae(y_test, y_pred)
  lr_mae.append(mae)
  y_pred_list.append(y_pred)

print(np.mean(lr_mae))
```

```python
# Task 1: Write your code here:
# Feature importance
feature_cols=['Distance_km', 'Weather', 'Traffic_Level', 'Time_of_Day', 'Vehicle_Type', 'Preparation_Time_min','Courier_Experience_yrs','Delivery_Time']
feature_importance = pd.DataFrame({
    'feature': feature_cols,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

plt.figure(figsize=(10, 6))
plt.barh(feature_importance['feature'], feature_importance['importance'])
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show()
```
