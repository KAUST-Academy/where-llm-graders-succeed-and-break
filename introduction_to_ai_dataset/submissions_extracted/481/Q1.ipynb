```python
# Task 1: Write your code here:
df.drop("Order_ID",axis=1, inplace=True) #dropping the colume
df.head() #seeing the head after dropping
```

```python
# Task 2: Write your code here:
print("Missing values:")
print(df.isnull().sum()) #showing the sum of missing values in data

df_clean = df.dropna() #droping the missing values (we have more than 1k+ of valed sample, one hundred of NaN dropped is tolarable)
df_clean.describe() #showing the cleaned data
```

```python
# Task 3: Write your code here:
def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df_clean)

#cleaned the duplicate, and ran the method second time for confirmation
```

```python
# Task 4: Write your code here:
Categorical = ["Weather", "Traffic_Level","Time_of_Day","Vehicle_Type"] #all Categorical colms

from sklearn.preprocessing import LabelEncoder

label_encoders = {}
for col in Categorical:
  le = LabelEncoder()
  df_clean[col] = le.fit_transform(df_clean[col])
  label_encoders[col] = le

df_clean #looking at the data after encoding
```

```python
df_clean.head()
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler

no_target = df_clean.columns.drop("Delivery_Time").copy()  # so we do not scale the target

scaler = StandardScaler()
df_clean[no_target] = scaler.fit_transform(df_clean[no_target])
df_clean.head()
```

```python
# Task 5: Write your code here:
def check_target_imbalance(df, target_column):
  print("Target Distribution:")

  df[target_column].hist()  # Yeah you can just do this :)
  plt.show()

check_target_imbalance(df_clean, "Delivery_Time")

#Based on the distribution we have an imbalenced data, so will need strastified k-fold to handle it
```

```python
# Task 1: Write your code here:
X = df_clean.drop("Delivery_Time", axis=1).astype(float) #features
y = df_clean["Delivery_Time"].astype(float) # target
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score



n_splits = 5 # K=5 folds
# Train RandomForestClassifier
model = RandomForestClassifier(n_estimators=100, max_depth=15,
                               class_weight='balanced', random_state=42)
model.fit(X, y)
print("Model trained!")


y_pred = model.predict(X)

print(f"Accuracy: {accuracy_score(y, y_pred):.4f}\n")
```
