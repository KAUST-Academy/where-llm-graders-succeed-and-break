```python
# Task 2: Write your code here:
# Analyze missing values
df.isnull().sum()
```

```python
#Categorical columns

# Fill categorical columns
for col in ['Weather','Traffic_Level','Time_of_Day']:
 df[col] = df[col].fillna(df[col].mode()[0])

df['Courier_Experience_yrs']= df['Courier_Experience_yrs'].fillna(df['Courier_Experience_yrs'].mean())
df['Delivery_Time']= df['Delivery_Time'].fillna(df['Delivery_Time'].mean())

print("Missing values remaining:\n", df.isnull().sum())
```

```python
# Task 3: Write your code here:
# 4. Do we have duplicate samples?
def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df)
```

```python
# Task 4: Write your code here:

from sklearn.preprocessing import OneHotEncoder #import OneHotEncoder
cat_col=['Weather','Traffic_Level','Time_of_Day']
print('data before encoding:\n', df) #show before encoding

onehot_encoder = OneHotEncoder(sparse_output=False) # Instantiate OneHotEncoder نستخدم سبراس عشان يضغط الداتا
for col in ['Weather','Traffic_Level','Time_of_Day']:
 df[col] = onehot_encoder.fit_transform(df[col]) # Apply fit_transform to the copied

print('\nData after encoding:\n', df) #show after encoding
```

```python
from sklearn.preprocessing import LabelEncoder #import LabelEncoder

print('data before encoding:\n', df) #show before encoding

label_encoder = LabelEncoder() # Instantiate LabelEncoder
for col in ['Weather','Traffic_Level','Time_of_Day','Vehicle_Type']:
  df[col] = label_encoder.fit_transform(df[col]) # Apply fit_transform to the column

print('\nData after encoding:\n', df) #show after encoding
```

```python
# Task 1: Write your code here:

# Define features (X) and target (y)
#feature_cols = ['Distance_km' , 'Weather' ,'Traffic_Level','Time_of_Day' ,'Vehicle_Type' ,'Preparation_Time_min' ,'Courier_Experience_yrs']
#y = df['Delivery_Time']

X = df.drop("Delivery_Time",axis=1)
y = df['Delivery_Time']
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler #import StandardScaler
standard_scaler = StandardScaler() # Instantiate StandardScaler
df = standard_scaler.fit_transform(df) # Apply fit_transform
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import KFold

# Define K-Fold Cross Validation
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# Iterate through folds
for fold, (train_idx, test_idx) in enumerate(kf.split(X), start=1): # فين تعرفت train indx
    # indexing for each fold
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx] #
    # print shapes
    print(f"Fold {fold}")
    print("  X_train shape:", X_train.shape)
    print("  X_test shape :", X_test.shape)
    print("  y_train shape:", y_train.shape)
    print("  y_test shape :", y_test.shape)
    print("-" * 30)
```

```python
from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor(n_estimators=100, max_depth=20, random_state=42, n_jobs=-1)
model.fit(X_train, y_train)
print("Model trained!")
```

```python
# Predict and evaluate

from sklearn.metrics import mean_absolute_error
y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)


print(f"MAE: {mae:,.2f}")
```

```python
# Task 1: Write your code here:
# Feature importance

feature_cols = ['Distance_km' , 'Weather' ,'Traffic_Level','Time_of_Day' ,'Vehicle_Type' ,'Preparation_Time_min' ,'Courier_Experience_yrs']
importance = pd.DataFrame({
    'feature': feature_cols,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

plt.figure(figsize=(10, 6))
plt.barh(importance['feature'], importance['importance'], color='purple')
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show()
```
