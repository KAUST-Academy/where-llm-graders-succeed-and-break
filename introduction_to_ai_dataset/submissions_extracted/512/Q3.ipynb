```python
# Task 1: Write your code here:
print("Missing values:")
print(df.isnull().sum())

df_clean = df.copy()
for col in df_clean.columns:
  df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])

print("Missing values:")
print(df_clean.isnull().sum())
```

```python
# Task 2: Write your code here:
def check_duplicates(df):
  duplicates = df_clean.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df_clean)

df_clean = df_clean.drop_duplicates()
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
features = df_clean.columns.drop("Target")  # DON'T SCALE THE TARGET
df_clean[features] = scaler.fit_transform(df_clean[features])
df_clean.head()
```

```python
# Task 5: Write your code here:
# Target distribution
print(f"1: {(df_clean['Target'] == 1).sum()}")
print(f"0: {(df_clean['Target'] == 0).sum()}")

# It is imbalanced because the zero is way more than the 1
```

```python
# Task 1: Write your code here:
from sklearn.model_selection import StratifiedKFold, KFold
X = df_clean.drop('Target', axis=1).astype(float)
y = df_clean['Target'].astype(float)
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.ensemble import CatBoostClassifier
n_splits = 5  # K=5 Folds
kfold = KFold(n_splits=n_splits, shuffle=True, random_state=42)

model = CatBoostClassifier(n_estimators=100, max_depth=15,
                               class_weight='balanced', random_state=42)

mae_scores = []
for fold_idx, (train_index, test_index) in enumerate(kfold.split(X)):
  print(f"\nFold {fold_idx + 1}/{n_splits}")

  X_train, X_test = X.iloc[train_index], X.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]

  # Train and predict
  model.fit(X_train, y_train)
  y_fold_pred = model.predict(X_test)

  accuracy = accuracy_score(y_test, y_fold_pred)
  precision = precision_score(y_test, y_fold_pred)
  print("Accuracy :", accuracy)
  print("Precision:", precision)

print(f"5-Fold CV Results:")
```
