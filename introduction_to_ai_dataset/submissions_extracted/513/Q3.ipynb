```python
import pandas as pd
import numpy as np
import os

import matplotlib.pyplot as plt

from sklearn.preprocessing import OneHotEncoder,LabelEncoder,StandardScaler
from sklearn.model_selection import train_test_split,KFold,StratifiedKFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, f1_score,accuracy_score

from catboost import CatBoostClassifier
```

```python
# Task 1: Write your code here:
csv_path= os.path.join(path,'Q3_data.csv')
df= pd.read_csv(csv_path)
```

```python
# Task 1: Write your code here:
for col in df.columns:
  df[col]= df[col].fillna(df[col].mean())
```

```python
# Task 4: Write your code here:
cols= df.columns.drop('Target')
scaler= StandardScaler()
df[cols]= scaler.fit_transform(df[cols])
```

```python
# Task 5: Write your code here:
df['Target'].value_counts()
```

```python
# Task 1: Write your code here:
X= df.drop(columns='Target')
y= df['Target']
```

```python
# Task 2,3,4,5: Write your code here:

# Task 2,3,4,5: Write your code here:
kf= StratifiedKFold(n_splits=5 ,random_state=42,shuffle=True)
model = CatBoostClassifier(n_estimators=100,random_state=42, verbose=0)

acc=[]
for train_idx,test_idx in kf.split(X,y):
  X_train,y_train = X.iloc[train_idx], y.iloc[train_idx]
  X_test,y_test = X.iloc[test_idx], y.iloc[test_idx]

  model.fit(X_train,y_train)

  y_pred= model.predict(X_test)

  accuracy = accuracy_score(y_test,y_pred)

  acc.append(accuracy)

print(f'avg accuracy = {np.mean(acc)}')
```

```python
# Task 1: Write your code here:
# Task 1: Write your code here:
importance = pd.DataFrame({
    'feature': X.columns,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

plt.figure(figsize=(10, 10))
plt.barh(importance['feature'][:20], importance['importance'][:20], color='purple')
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show()
```

```python
# Task 2: Write your code here:
best_featue= importance['feature'].iloc[0]
print(f'the golden feature is {best_featue}')
```

```python
# Task Bonus: Write your code here:
X = df[[best_featue]]
# Task 2,3,4,5: Write your code here:

# Task 2,3,4,5: Write your code here:
kf= StratifiedKFold(n_splits=5 ,random_state=42,shuffle=True)
model = CatBoostClassifier(n_estimators=100,random_state=42, verbose=0)

acc_best_feature=[]
for train_idx,test_idx in kf.split(X,y):
  X_train,y_train = X.iloc[train_idx], y.iloc[train_idx]
  X_test,y_test = X.iloc[test_idx], y.iloc[test_idx]

  model.fit(X_train,y_train)

  y_pred= model.predict(X_test)

  accuracy = accuracy_score(y_test,y_pred)

  acc_best_feature.append(accuracy)

print(f'traning with all features = {np.mean(acc)}\ntraining with the golden Feature only {np.mean(acc_best_feature)}')
```
