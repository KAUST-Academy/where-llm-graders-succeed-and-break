```python
# Task 2: Write your code here:
# Do we have missing values?
def check_missing_values(df_Q1):
  missing_values = df_Q1.isnull().sum()
  print("Missing Values per Column:")
  print(missing_values[missing_values > 0])
  if missing_values.any():
    print("\nHandle Missing Values as needed.")
  else:
    print("\nNo Missing Values Found.")

check_missing_values(df_Q1)
```

```python


df_Q1['Weather'] = df_Q1['Weather'].fillna(df_Q1['Weather'].median())

df_Q1['Courier_Experience_yrs'] = df_Q1['Courier_Experience_yrs'].fillna(df_Q1['Courier_Experience_yrs'].median())

df_Q1['Time_of_Day'] = df_Q1['Time_of_Day'].fillna(df_Q1['Time_of_Day'].median())

df_Q1['Traffic_Level'] = df_Q1['Traffic_Level'].fillna(df_Q1['Traffic_Level'].median())

df_Q1['Delivery_Time'] = df_Q1['Delivery_Time'].fillna(df_Q1['Delivery_Time'].median())


```

```python
# Task 3: Write your code here:
#Do we have duplicate samples?
def check_duplicates(df_Q1):
  duplicates = df_Q1.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df_Q1.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df_Q1)

df_Q1.drop_duplicates(inplace=True)
```

```python
# Task 4: Write your code here:
 #Encode categorical columns - converts text to integers
categorical_cols = ['Weather', 'Traffic_Level', 'Time_of_Day', 'Courier_Experience_yrs']
for col in categorical_cols:
    le = LabelEncoder()
    df_Q1[col] = le.fit_transform(df_Q1[col].astype(str))

df_Q1.head()
```

```python
# Task 5: Write your code here: categorical_cols

scaler = StandardScaler()
xCateg =df_Q1[categorical_cols]
categorical_cols_scaled = scaler.fit_transform(xCateg)
```

```python
# Task 6: Write your code here:
def check_target_imbalance(df, target_column):
  print("Target Distribution:")
  print(df[target_column].value_counts(normalize=True))
  sns.countplot(x=df[target_column])
  plt.title("Target Distribution")
  plt.show()

check_target_imbalance(df_Q1, "Delivery_Time")
```

```python
# Task 1: Write your code here:
X = df_Q1[categorical_cols]
y = df_Q1['Delivery_Time']
```

```python
# Task 2,3,4,5: Write your code here:

from sklearn.model_selection import KFold


# Define K-Fold Cross Validation
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# Iterate through folds
for fold, (train_idx, test_idx) in enumerate(kf.split(X), start=1):
    # indexing for each fold
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
    # print shapes
    print(f"Fold {fold}")
    print("  X_train shape:", X_train.shape)
    print("  X_test shape :", X_test.shape)
    print("  y_train shape:", y_train.shape)
    print("  y_test shape :", y_test.shape)
    print("-" * 30)
```

```python
from sklearn.ensemble import RandomForestClassifier

# Scale features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Train RandomForestClassifier
model = RandomForestClassifier(n_estimators=100, max_depth=15,
                               class_weight='balanced', random_state=42)
model.fit(X_train_scaled, y_train)
print("Model trained!")
```
