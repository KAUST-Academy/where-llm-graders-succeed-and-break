```python
# Task 5: Write your code here:
def check_target_distribution(df, target_column):
  df[target_column].hist(bins=30, edgecolor='black')

  plt.title(f"Target Distribution ({target_column})")
  plt.xlabel(target_column)
  plt.ylabel("Frequency")
  plt.grid(False)

  plt.show()

check_target_distribution(df, "Delivery_Time")
```

```python
# Task 1: Write your code here:
df.drop(columns=['Order_ID'])
```

```python
# Task 2: Write your code here:
df.dropna(subset=['Delivery_Time'],inplace = True)
df['Courier_Experience_yrs'].fillna(df['Courier_Experience_yrs'].mean(),inplace = True)
df['Time_of_Day'].fillna(df['Time_of_Day'].mode()[0],inplace = True)
df['Traffic_Level'].fillna(df['Traffic_Level'].mode()[0],inplace = True)
df['Weather'].fillna(df['Weather'].mode()[0],inplace = True)
def check_missing_values(df):
  missing_values = df.isnull().sum()
  print("Missing Values per Column:")
  print(missing_values[missing_values > 0])
  if missing_values.any():
    print("\nHandle Missing Values as needed.")
  else:
    print("\nNo Missing Values Found.")

check_missing_values(df)
```

```python
# Task 3: Write your code here:
def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df)
```

```python
# Task 4: Write your code here:
categorical_cols = ['Weather', 'Traffic_Level', 'Time_of_Day', 'Vehicle_Type']
for col in categorical_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col].astype(str))

df.head()
```

```python
# Task 5: Write your code here:

feature_cols = ['Order_ID', 'Distance_km', 'Weather', 'Traffic_Level',
                'Time_of_Day', 'Vehicle_Type', 'Preparation_Time_min', 'Courier_Experience_yrs']
X = df[feature_cols]
y = df['Delivery_Time']

scaler = StandardScaler()
X = pd.DataFrame( scaler.fit_transform(X))
X
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import KFold
n_splits = 5
kf = KFold(n_splits=5, shuffle=True, random_state=42)
scors = []

for fold_idx, (train_index, test_index) in enumerate(kf.split(X)):
  print(f"\nFold {fold_idx + 1}/{n_splits}")

  X_train, X_test = X.iloc[train_index], X.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]

  model = RandomForestRegressor(n_estimators=100, max_depth=20, random_state=42, n_jobs=-1)
  model.fit(X_train, y_train)
  y_pred = model.predict(X_test)

  # Calculate metrics
  mae = mean_absolute_error(y_test, y_pred)
  scors.append(mae)
  print(mae)
print()
scors = np.array(scors)
print("the averaged score across all folds " , scors.mean())
```

```python
# Task 1: Write your code here:
# Feature importance
feature_importance = pd.DataFrame({
    'feature': feature_cols,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

plt.figure(figsize=(10, 6))
plt.barh(feature_importance['feature'], feature_importance['importance'])
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show()
```
