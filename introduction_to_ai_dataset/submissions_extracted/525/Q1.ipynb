```python
# Task 2: Write your code here:
print("\nMissing values:")
print(foodeliver.isnull().sum())

delivclean =foodeliver.copy()
delivclean = delivclean.fillna({'Traffic_Level': 'Low', 'Weather': 'Clear','Time_of_Day':'Afternoon'})


delivclean['Courier_Experience_yrs'] = delivclean['Courier_Experience_yrs'].fillna(delivclean['Courier_Experience_yrs'].mean())
delivclean['Delivery_Time'] = delivclean['Delivery_Time'].fillna(delivclean['Delivery_Time'].mean())
print("\n handeld missing values")
print(delivclean)
print("\nmissing values = " , delivclean.isnull().sum().sum())
```

```python
# Task 3: Write your code here:
def check_duplicates(delivclean):
  duplicates = delivclean.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    delivclean.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(delivclean)
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import OneHotEncoder

print('data before encoding:\n', delivclean)

onehot_encoder = OneHotEncoder(sparse_output=False)
delivclean['Weather'] = onehot_encoder.fit_transform(delivclean)
delivclean['Traffic_Level'] = onehot_encoder.fit_transform(delivclean['Weather'])
delivclean['Time_of_Day'] = onehot_encoder.fit_transform(delivclean['Time_of_Day'])
delivclean['Vehicle_Type'] = onehot_encoder.fit_transform(delivclean['Vehicle_Type'])


print('\nData after encoding:\n', delivclean)
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler

standard_scaler = StandardScaler()
delivclean = standard_scaler.fit_transform(delivclean)
print('\nData after scaling:\n', delivclean)
```
