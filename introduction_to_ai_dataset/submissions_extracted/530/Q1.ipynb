```python
# Task 1: Write your code here
df_clean = df.drop(columns=["Order_ID"]).copy()
df_clean.head()
```

```python
# Task 2: Write your code here:
null_values = df_clean.isnull().sum()
print(null_values)

# i will drop the missing target
df_clean = df_clean.dropna(subset=["Delivery_Time"])
print("---------")
print(df_clean.Delivery_Time.isna().sum())  # zero .
print("---------")

# for the other num features ideas i will fill it with thw mean
num_col= ["Distance_km","Preparation_Time_min","Courier_Experience_yrs"]
for col in num_col:
    df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])

# for object i will fill it with mode
cat_col = ["Weather","Traffic_Level","Time_of_Day","Vehicle_Type"]
for col in cat_col:
    df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])
print(df_clean.isnull().sum() )


```

```python
# Task 3: Write your code here:
def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df_clean) # first calling to remove duplicates
check_duplicates(df_clean) # just to check no duplicates left
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import LabelEncoder #import LabelEncoder

print('data before encoding:\n', df_clean) #show before encoding

for col in cat_col:
    le = LabelEncoder()
    df_clean[col] = le.fit_transform(df_clean[col].astype(str))

df_clean.head()
```

```python
# Task 2,3,4,5: Write your code here:
# i well use kfold , because there is no classes to care about thier balance
from sklearn.model_selection import KFold
import numpy as np
from sklearn.metrics import mean_absolute_error as MAE
from sklearn.ensemble import RandomForestRegressor


kf = KFold(n_splits=5, shuffle=True, random_state=42)
MAE_loss = []

for fold_idx, (train_index, test_index) in enumerate(kf.split(X)):
  print(f"\nFold {fold_idx + 1}/{5}")

  X_train, X_test = X.iloc[train_index], X.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]

  # Train
  RF = RandomForestRegressor(n_estimators= 200)
  RF.fit(X_train,y_train)

  # Validate
  y_pred = RF.predict(X_test)

  # Calculate evaluation metrics
  mae = MAE(y_test,y_pred)

  # Store results
  MAE_loss.append(mae)




```

```python
  print(f"  MAE:  {np.mean(MAE_loss):.4f}")
```
