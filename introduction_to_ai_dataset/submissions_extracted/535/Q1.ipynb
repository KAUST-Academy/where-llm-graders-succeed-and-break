```python
# Task 2: Handle missing values appropriately


#filling Weather+Time_of_Day+Traffic_Level columns with 'mode' value because its categoral
df_q1['Weather'] = df_q1['Weather'].fillna(df_q1['Weather'].mode()[0])
df_q1['Time_of_Day'] = df_q1['Time_of_Day'].fillna(df_q1['Time_of_Day'].mode()[0])
df_q1['Traffic_Level'] = df_q1['Traffic_Level'].fillna(df_q1['Traffic_Level'].mode()[0])
#filling Delivery_Time+Courier_Experience_yrs column with 'mean''median' value because its numerical
df_q1['Delivery_Time'] = df_q1['Delivery_Time'].fillna(df_q1['Delivery_Time'].mean())
df_q1['Courier_Experience_yrs'] = df_q1['Courier_Experience_yrs'].fillna(df_q1['Courier_Experience_yrs'].median())


print("Missing values After Hndling:")
print(df_q1.isnull().sum())
```
