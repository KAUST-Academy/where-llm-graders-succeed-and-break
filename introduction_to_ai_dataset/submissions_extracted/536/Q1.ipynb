```python
# Task 2: Write your code here:

stat_cols = ['Weather', 'Traffic_Level', 'Time_of_Day', 'Courier_Experience_yrs', 'Delivery_Time']

# Drop rows with missing stat values
df_Q1_data = df_Q1_data.dropna(subset=stat_cols).copy()
print(f"Shape after cleaning: {df_Q1_data.shape}")

print("Missing values:")
print(df_Q1_data.isnull().sum())
```

```python
# Task 3: Write your code here:

duplicates = df_Q1_data.duplicated().sum()
print(f"Number of Duplicate Samples: {duplicates}")
if duplicates > 0:
  print("Dropping Duplicates...")
  df_Q1_data.drop_duplicates(inplace=True)
  print("Duplicates Dropped.")
else:
  print("No Duplicate Samples Found.")
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import OneHotEncoder,LabelEncoder
categorical_cols = df_Q1_data.select_dtypes(include=["object"]).columns


for col in categorical_cols:
    le = LabelEncoder()
    df_Q1_data[col] = le.fit_transform(df_Q1_data[col].astype(str))
df_Q1_data
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler #import StandardScaler

numerical_cols = df_Q1_data.select_dtypes(include=["int64", "float64"]).columns.drop("Delivery_Time")  # DON'T SCALE THE TARGET

scaler = StandardScaler()
df_Q1_data[numerical_cols] = scaler.fit_transform(df_Q1_data[numerical_cols])
df_Q1_data.head()
```

```python
# Task 6: Write your code here:

df_Q1_data['Delivery_Time'].hist(bins=30, edgecolor='black')

plt.title(f"Target Distribution (Delivery_Time)")
plt.xlabel("Delivery_Time")
plt.ylabel("Frequency")
plt.grid(False)

plt.show()
```

```python
# Task 1: Write your code here:
X=df_Q1_data[numerical_cols]
y=df_Q1_data['Delivery_Time']
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error as sklearn_mse, mean_absolute_error, r2_score

n_splits = 5  # K=5 Folds

# 5-Fold Cross-Validation, shuffled
kf = KFold(n_splits=n_splits, shuffle=True, random_state=42)

for fold_idx, (train_index, test_index) in enumerate(kf.split(X)):
  print(f"\nFold {fold_idx + 1}/{n_splits}")

  X_train, X_test = X.iloc[train_index], X.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]
model = RandomForestRegressor(n_estimators=100, max_depth=20, random_state=42, n_jobs=-1)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
```

```python
# Task 1: Write your code here:
feature_importance = pd.DataFrame({
    'feature': numerical_cols,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

plt.figure(figsize=(10, 6))
plt.barh(feature_importance['feature'], feature_importance['importance'])
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show()
```
