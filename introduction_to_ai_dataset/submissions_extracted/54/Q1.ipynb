```python
# Task 2: Write your code here:
missing_data_number = df_food.isnull().sum()
print(missing_data_number)
# Select relevant columns, we do not select 'model' (too many unique values, too sparse and will hurt the model performance)
cols = ['Order_ID'	,'Distance_km'	,'Weather'	,'Traffic_Level'	,'Time_of_Day' ,	'Vehicle_Type'	,'Preparation_Time_min'	,'Courier_Experience_yrs'	,'Delivery_Time']
df_clean = df_food[cols].copy()

# Drop rows where target (price) or key features are missing - can't predict without them
print(f"Before: {df_clean.shape}")
df_clean = df_clean.dropna(subset=['Order_ID'	,'Distance_km' ,	'Weather',	'Traffic_Level',	'Time_of_Day'	,'Vehicle_Type'	,'Preparation_Time_min'	,'Courier_Experience_yrs'	,'Delivery_Time'])
print(f"After dropping missing : {df_clean.shape}")
```

```python
# Task 3: Write your code here:
def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df_clean)
```

```python
# Task 4: Write your code here:
categorical_cols = ['Weather',	'Traffic_Level',	'Time_of_Day',	'Vehicle_Type']
for col in categorical_cols:
    le = LabelEncoder()
    df_clean[col] = le.fit_transform(df_clean[col].astype(str))

df_clean.head()
```
