```python
# Task 1: Write your code here:
df_data_clean = df_data.copy().drop(columns=['Order_ID'])
```

```python
# Task 2: Write your code here:
missing_percentage = (df_data.isnull().sum() / len(df_data)) * 100
missing_data = pd.DataFrame({
    'Column': missing_percentage.index,
    'Missing_Percentage': missing_percentage.values
})
missing_data = missing_data[missing_data['Missing_Percentage'] > 0].sort_values('Missing_Percentage', ascending=False)

print("Missing Data Analysis:")
missing_data.head(10)

```

```python
df_data_clean['Delivery_Time'].fillna(df_data_clean['Delivery_Time'].mean())
df_data_clean['Weather'].fillna(df_data_clean['Weather'].mode()[0])
df_data_clean['Traffic_Level'].fillna(df_data_clean['Traffic_Level'].mode()[0])
df_data_clean['Time_of_Day'].fillna(df_data_clean['Time_of_Day'].mode()[0])
df_data_clean['Courier_Experience_yrs'].fillna(df_data_clean['Courier_Experience_yrs'].mean())
```

```python
# Task 3: Write your code here:
def check_duplicates(df):
    duplicates = df.duplicated().sum()
    print(f"Number of Duplicate Samples: {duplicates}")
    if duplicates > 0:

        df.drop_duplicates(inplace=True)
        print("Duplicates Dropped.")
    else:
        print("No Duplicate Samples Found.")

check_duplicates(df_data_clean)
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import  LabelEncoder

def encode_categorical_columns(df):
    categorical_cols = df.select_dtypes(include=["object"]).columns
    print("Categorical Columns:", list(categorical_cols))

label_encoders = encode_categorical_columns(df_data_clean)




categorical_cols = df_data_clean.select_dtypes(include=["object"]).columns
for col in categorical_cols:
    print(f"Encoding column: {col}")
    le = LabelEncoder()
    df_data_clean[col] = le.fit_transform(df_data_clean[col])
df_data_clean
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler #import StandardScaler

print('data before scaling:\n', df_data_clean) #show before scaling
standard_scaler = StandardScaler() # Instantiate StandardScaler
data_standard_scaled = standard_scaler.fit_transform(df_data_clean) # Apply fit_transform

print('\nData after scaling:\n', data_standard_scaled) #show after scaling
```

```python
# Task 1: Write your code here:
X = df_data_clean.drop("Delivery_Time",axis=1)
y = df_data_clean['Delivery_Time']
```

```python
# Task 2,3,4,5: Write your code here:


from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import KFold

scores = []
model = RandomForestRegressor(n_estimators=100, max_depth=20, random_state=42, n_jobs=-1)
skf = KFold(n_splits=5)
for train_index, test_index in skf.split(X, y):
# Split data into training and testing sets
  X_Train, X_Test = X.loc[train_index, :], X.loc[test_index, :]
  y_Train, y_Test = y.iloc[train_index], y.iloc[test_index]
  # Train the model
  model.fit(X_Train, y_Train)
  print("Model trained!")

  y_pred = model.predict(X_Test)
  MAE = np.abs(y_Test - y_pred)

    # Calculate metrics
  scores.append(MAE)
```
