```python
# Task 1: Write your code here:
missing_percentage = (df_data.isnull().sum() / len(df_data)) * 100
missing_data = pd.DataFrame({
    'Column': missing_percentage.index,
    'Missing_Percentage': missing_percentage.values
})
missing_data = missing_data[missing_data['Missing_Percentage'] > 0].sort_values('Missing_Percentage', ascending=True)
print(df_data.shape)
colu = []
for i in missing_data['Column']:
  colu.append(i)
print(colu)
df_data_clean = df_data.drop(columns = colu)

df_data_clean.shape
```

```python
# Task 2: Write your code here:
def check_duplicates(df):
    duplicates = df.duplicated().sum()
    print(f"Number of Duplicate Samples: {duplicates}")
    if duplicates > 0:
        print("Dropping Duplicates...")
        df.drop_duplicates(inplace=True)
        print("Duplicates Dropped.")
    else:
        print("No Duplicate Samples Found.")

check_duplicates(df_data_clean)
```

```python
# Task 3: Write your code here:
# Task 4: Write your code here:
from sklearn.preprocessing import  LabelEncoder

def encode_categorical_columns(df):
    categorical_cols = df.select_dtypes(include=["object"]).columns
    print("Categorical Columns:", list(categorical_cols))

label_encoders = encode_categorical_columns(df_data_clean)




categorical_cols = df_data_clean.select_dtypes(include=["object"]).columns
for col in categorical_cols:
    print(f"Encoding column: {col}")
    le = LabelEncoder()
    df_data_clean[col] = le.fit_transform(df_data_clean[col])
df_data_clean
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import StandardScaler #import StandardScaler

print('data before scaling:\n', df_data_clean) #show before scaling
standard_scaler = StandardScaler() # Instantiate StandardScaler
data_standard_scaled = standard_scaler.fit_transform(df_data_clean) # Apply fit_transform

print('\nData after scaling:\n', data_standard_scaled) #show after scaling
```

```python
# Task 1: Write your code here:
from sklearn.model_selection import KFold
X = df_data_clean.drop("Target",axis=1)
y = df_data_clean['Target']

model = CatBoostClassifier(
      verbose=0,
      n_estimators=320,
      max_depth=4
  )
```

```python
# Task 2,3,4,5: Write your code here:



scores_accuracy = []
scores_precision = []
scores_recall = []
scores_f1 = []

# Stratified 5-Fold Cross-Validation
skf = KFold(n_splits=5)
for train_index, test_index in skf.split(X, y):
    # Split data into training and testing sets
    X_Train, X_Test = X.loc[train_index, :], X.loc[test_index, :]
    y_Train, y_Test = y.iloc[train_index], y.iloc[test_index]
    # Train the model
    model.fit(X_Train, y_Train)
    # Predict on the test set
    y_pred = model.predict(X_Test)

    # Calculate metrics
    scores_f1.append(f1_score(y_Test, y_pred, average='weighted'))

    # Print the results
    print(scores_f1)
    print("\n")
```

```python
# Task 1: Write your code here:
feature_importance = pd.DataFrame({
    'feature': df_data_clean.iloc[0],
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

plt.figure(figsize=(10, 6))
plt.barh(feature_importance['feature'], feature_importance['importance'])
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show()
```
