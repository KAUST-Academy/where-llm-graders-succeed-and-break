```python
## dropping missing values from target column
df.dropna(subset=['Delivery_Time'], inplace=True)

## Explicit "Missing" for other columns
df['Weather'] = df['Weather'].fillna('Missing')
df['Traffic_Level'] = df['Traffic_Level'].fillna('Missing')
df['Time_of_Day'] = df['Time_of_Day'].fillna('Missing')
df['Courier_Experience_yrs'] = df['Courier_Experience_yrs'].fillna(mean_exp)
```

```python
# Task 3: Write your code here:
df = df.drop_duplicates()
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
for col in df.select_dtypes('object'):
  df[col] = encoder.fit_transform(df[col])
```

```python
df.head()
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
## splitting target from features
X = df.drop('Delivery_Time', axis=1)
y = df['Delivery_Time']

X = scaler.fit_transform(X)
```

```python
# Task 1: Write your code here:
## Done previously in Task 5 of part 3
print(X)
print(y)
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

mae_scores = []
# kfold
kfold = KFold(n_splits=5, shuffle=True, random_state=42)
# instantiate tree model
model = RandomForestRegressor(n_estimators=100, max_depth=20, random_state=42, n_jobs=-1)

for train_idx, val_idx in kfold.split(X):
    X_fold_train, X_fold_val = X[train_idx], X[val_idx]
    y_fold_train, y_fold_val = y.iloc[train_idx], y.iloc[val_idx]

    # Train and predict
    model.fit(X_fold_train, y_fold_train)
    y_fold_pred = model.predict(X_fold_val)

    # Calculate metrics
    mae_scores.append(mean_absolute_error(y_fold_val, y_fold_pred))


mae_scores = np.array(mae_scores)
print(f"MAE:  ${mae_scores.mean():,.2f}")
```

```python
# Task 1: Write your code here:
feature_importance = pd.DataFrame({
    'feature': df.drop('Delivery_Time', axis=1).columns,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

plt.figure(figsize=(10, 6))
plt.barh(feature_importance['feature'], feature_importance['importance'])
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show()
```
