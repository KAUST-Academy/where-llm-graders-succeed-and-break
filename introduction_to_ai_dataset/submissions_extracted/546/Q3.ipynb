```python
# Task 1: Write your code here:
## seeing missings
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_data = pd.DataFrame({
    'Column': missing_percentage.index,
    'Missing_Percentage': missing_percentage.values
})
missing_data = missing_data[missing_data['Missing_Percentage'] > 0].sort_values('Missing_Percentage', ascending=False)

print("Missing Data Analysis:")
missing_data
```

```python
## I will go ahead and drop all columns with high missing percentage

high_percentage = missing_data[(missing_data['Missing_Percentage'] > 15)]
for col in high_percentage['Column']:
  df.drop(col, axis=1, inplace=True)
```

```python
## for other columns, drop the rows with missing values
## seeing missings
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_data = pd.DataFrame({
    'Column': missing_percentage.index,
    'Missing_Percentage': missing_percentage.values
})
missing_data = missing_data[missing_data['Missing_Percentage'] > 0].sort_values('Missing_Percentage', ascending=False)

df.dropna(subset= list(missing_data['Column'].unique()), inplace=True)
```
