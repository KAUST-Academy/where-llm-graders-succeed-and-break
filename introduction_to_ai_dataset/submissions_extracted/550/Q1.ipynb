```python
# Task 1: Write your code here:
df_clean = df_food.drop('Order_ID',axis=1).copy()
```

```python
# Task 2: Write your code here:
categorical_cols = df_food.select_dtypes(include=["object"]).columns
df_clean[categorical_cols]=df_clean[categorical_cols].fillna('unkown')
numrical_cols = df_clean.select_dtypes(include=['float']).columns
numrical_cols = [i for i in numrical_cols if i != 'Delivery_Time']
df_clean[numrical_cols] = df_clean[numrical_cols].fillna(df_clean[numrical_cols].mean())
df_clean.info()
```

```python
# Task 3: Write your code here:
print("Checking for duplicate rows...")
duplicate_rows = df_clean.duplicated().sum()
if duplicate_rows > 0:
    print(f"Found {duplicate_rows} duplicate rows. Removing them...")
    df_clean.drop_duplicates(inplace=True,)
    print("Duplicate rows removed.")
else:
    print("No duplicate rows found.")

df_clean.info()
```
