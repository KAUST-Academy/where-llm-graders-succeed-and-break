```python
# Task 1: Write your code here:
df = df.drop(['Order_ID'] , axis=1)
```

```python
# Task 2: Write your code here:
# needs to fill Weather Traffic_Level Time_of_Day ourier_Experience_yrs Delivery_Time

df['Weather'].fillna(df['Weather'].mode()[0], inplace=True)
df['Traffic_Level'].fillna(df['Traffic_Level'].mode()[0], inplace=True)
df['Time_of_Day'].fillna(df['Time_of_Day'].mode()[0], inplace=True)



df['Courier_Experience_yrs'].fillna(df['Courier_Experience_yrs'].mean(), inplace=True)
df['Delivery_Time'].fillna(df['Delivery_Time'].mean(), inplace=True)
df.info()

```

```python
# Task 3: Write your code here:
duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicates}")

if duplicates > 0:
    df.drop_duplicates(inplace=True)
    print(f"Removed {duplicates} duplicate rows")
else:
    print("No duplicates found!")
```

```python
# Task 4: Write your code here:    Encode categorical variables if needed (Bonus if used One Hot Encoding)
categorical_cols = ['Weather','Traffic_Level','Time_of_Day','Vehicle_Type']


for col in categorical_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col].astype(str))

df.head()
```

```python
# Task 5: Write your code here: Apply feature scaling for all features (Use StandardScaler)
# Task 5: Write your code here: Apply feature scaling for all features (Use StandardScaler)



X = df.drop('Delivery_Time', axis=1)
y = df['Delivery_Time']


scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)




```

```python
# Task 2,3,4,5: Write your code here:
# Define Stratified K-Fold >>> could not do i will do normal splits because there isnt much time left
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y,
    test_size=0.2,        # 20% for test, remaining 80% for train
    random_state=42,      # reproducible output
    shuffle=True,         # representative splits

)
model = RandomForestRegressor(n_estimators=100, max_depth=20, random_state=42, n_jobs=-1)
model.fit(X_train, y_train)
print("Model trained!")
# Predict and evaluate
y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)

print("mean abs error = " , mae)
```
