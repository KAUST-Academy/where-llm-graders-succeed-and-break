```python
# Task 1: Write your code here:
# add the required imports
import os
import pandas as pd

anonymized_dataset_path = os.path.join(path, 'Q3_data.csv') # add the the path to the csv file using os.path.join
df_anonymized = pd.read_csv(anonymized_dataset_path) # read the file into a dataframe

print(f"Dataset shape: {df_anonymized.shape}") # print the shape of the dataframe
```

```python
# Task 1: Write your code here:
missing_values = df_anonymized.isnull().sum()
print("Missing Values per Column (before cleaning):")
print(missing_values[missing_values > 0])

df_clean = df_anonymized.copy()

for col in df_clean.isnull().columns:
    df_clean[col] = df_clean[col].fillna(df_clean[col].mean())


missing_values = df_clean.isnull().sum()
print("Missing Values per Column (after cleaning):")
print(missing_values[missing_values > 0])
```

```python
# Task 2: Write your code here:
def check_duplicates(df):
  duplicates = df_clean.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df_clean.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df_clean)
```

```python
# Task 3: Write your code here:
# The dataset doesn't contain categorical columns
categorical_cols = df_clean.select_dtypes(include=["object"]).columns

print("Categorical Columns:", list(categorical_cols))
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import StandardScaler #import StandardScaler

features = df_clean.columns.drop("Target") # select only the features for scaling

standard_scaler = StandardScaler() # Instantiate StandardScaler
df_clean[features] = standard_scaler.fit_transform(df_clean[features]) # Apply fit_transform
df_clean.head()
```

```python
# Task 5: Write your code here:
import seaborn as sns, matplotlib.pyplot as plt

def check_target_imbalance(df, target_column):
  print("Target Distribution:")
  print(df[target_column].value_counts(normalize=True))
  sns.countplot(x=df[target_column])
  plt.title("Target Distribution")
  plt.show()

check_target_imbalance(df_clean, "Target")
# the target is imbalanced
```

```python
# Task 1: Write your code here:
X = df_clean.drop("Target", axis=1).astype(float)
y = df_clean["Target"].astype(float)
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import f1_score

n_splits = 5 # K=5 Folds

# Stratified 5-Fold Cross-Validation, shuffled
skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)

from catboost import CatBoostRegressor

catboost = CatBoostRegressor(verbose=0)
f1 = []

for fold_idx, (train_index, test_index) in enumerate(skf.split(X)):
  print(f"\nFold {fold_idx + 1}/{n_splits}")

  X_train, X_test = X.iloc[train_index], X.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]


  # Train
  catboost.fit(X_train, y_train)

  # Predict
  y_pred = catboost.predict(X_test)

  # Calculate metrics
  f1_score = f1_score(y_test, y_pred)

  f1.append(f1_score)
```
