```python
# Task 1: Write your code here:
# 2. Do we have missing values?
def check_missing_values(df_Q3):

  # Get missing values using pandas
  missing_values = df_Q3.isnull().sum()

  print("Missing Values per Column:")
  print(missing_values[missing_values > 0])

  if missing_values.any():
    print("\nHandle Missing Values as needed.")
  else:
    print("\nNo Missing Values Found.")

check_missing_values(df_Q3)
```

```python
# Task 2: Write your code here:
# 4. Check for duplicates
def check_duplicates(df_Q3):

  #TODO: get duplicated data using pandas
  duplicates = df_Q3.duplicated().sum()

  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df_Q3.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df_Q3)
```

```python
# Task 3: Write your code here:
# 3. Do we have categorical columns?
categorical_cols = df_Q3.select_dtypes(include=["object"]).columns

print("Categorical Columns:", list(categorical_cols))
# no need for labelencoder
```

```python
# Task 4: Write your code here:
df_clean = df_Q3.copy()
df_clean.head()
from sklearn.preprocessing import StandardScaler

numerical_cols = df_clean.select_dtypes(include=["int64", "float64"]).columns.drop("Target")  # no need to scale target
scaler = StandardScaler()
df_clean[numerical_cols] = scaler.fit_transform(df_clean[numerical_cols])
df_clean.head()
```

```python
# Task 5: Write your code here: 5. **Check for target imbalance and state if it is imbalanced or not**

def check_target_imbalance(df_clean, target_column):
  print("Target Distribution:")

  df_clean[target_column].hist()
  plt.show()

check_target_imbalance(df_clean, "Target")
```
