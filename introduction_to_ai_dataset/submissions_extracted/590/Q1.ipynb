```python
# Task 2: Write your code here:
# Do we have missing values?
def check_missing_values(df_rest):
  missing_values = df_rest.isnull().sum()
  print("Missing Values per Column:")
  print(missing_values[missing_values > 0])
  if missing_values.any():
    print("\nHandle Missing Values as needed.")
    print("\nHandling Missing Values...")
  else:
    print("\nNo Missing Values Found.")

check_missing_values(df_rest)

df_rest_cl = df_rest.dropna(subset=['Weather', 'Time_of_Day', 'Traffic_Level', 'Courier_Experience_yrs', 'Delivery_Time'])

def check_missing_values(df_rest_cl):
  missing_values = df_rest_cl.isnull().sum()
  print("Missing Values per Column:")
  print(missing_values[missing_values > 0])
  if missing_values.any():
    print("\nHandle Missing Values as needed.")
    print("\nHandling Missing Values...")
  else:
    print("\nNo Missing Values Found.")
check_missing_values(df_rest_cl)
```

```python
# Task 3: Write your code here:
def check_duplicates(df_rest_cl):
  duplicates = df_rest_cl.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df_rest_cl.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df_rest_cl)
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import OneHotEncoder #import OneHotEncoder

categorical_cols = ['Weather', 'Traffic_Level', 'Time_of_Day', 'Vehicle_Type']
for col in categorical_cols:
    le = LabelEncoder()
    df_rest_cl[col] = le.fit_transform(df_rest_cl[col].astype(str))

df_rest_cl.head()
```

```python
# Task 6: Write your code here:
target_column = df_rest_cl.Delivery_Time
def check_target_imbalance(df_rest_cl, target_column):
  print("Target Distribution:")
  print(df_rest_cl[target_column].value_counts(normalize=True))
  sns.countplot(x=df_rest_cl[target_column])
  plt.title("Target Distribution")
  plt.show()

check_target_imbalance(df_rest_cl, "Delivery_Time")
```

```python
# Task 1: Write your code here:
x_split = df_rest.drop("Delivery_Time", axis=1)
y_split = df_rest['Delivery_Time']
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import KFold

# Use previously generated random data (example: regression data)
X, y = x_split.copy(), y_split.copy()

# Define K-Fold Cross Validation
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# Iterate through folds
for fold, (train_idx, test_idx) in enumerate(kf.split(X), start=1):
    # indexing for each fold
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
    # print shapes
    print(f"Fold {fold}")
    print("  X_train shape:", X_train.shape)
    print("  X_test shape :", X_test.shape)
    print("  y_train shape:", y_train.shape)
    print("  y_test shape :", y_test.shape)
    print("-" * 30)


model = RandomForestRegressor(n_estimators=100, max_depth=20, random_state=42, n_jobs=-1)
model.fit(X, y)
```
