```python
# Task 1: Write your code here:
X = df.drop('Target', axis=1)
y = df['Target']
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import KFold, StratifiedKFold
from sklearn.metrics import accuracy_score, f1_score
from catboost import CatBoostClassifier

kf = StratifiedKFold(n_splits=5, shuffle=True)
scores_accuracy = []
scores_f1 = []

model = CatBoostClassifier(iterations=100)

for train_index, test_index in kf.split(X, y):
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    scores_accuracy.append(accuracy)
    scores_f1.append(f1)

print("Accuracy scores:", scores_accuracy)
print("F1 scores:", scores_f1)

average_accuracy = sum(scores_accuracy) / len(scores_accuracy)
average_f1 = sum(scores_f1) / len(scores_f1)

print("Average Accuracy:", average_accuracy)
print("Average F1 Score:", average_f1)
```

```python
# Task 1: Write your code here:
feature_importance = model.get_feature_importance(prettified=True)
feature_importance
```

```python
plt.barh(feature_importance['Feature Id'], feature_importance['Importances'])
plt.xlabel('Importance')
plt.ylabel('Feature')
plt.show()
```

```python
# Task 2: Write your code here:
golden_feature = feature_importance.iloc[0]['Feature Id']
golden_feature_importance = feature_importance.iloc[0]['Importances']

print("Golden Feature:", golden_feature)
print("importance:", golden_feature_importance)
```

```python
# Task Bonus: Write your code here:
X_golden = df[golden_feature].values.reshape(-1, 1)

kf = StratifiedKFold(n_splits=5, shuffle=True)
scores_accuracy_golden = []
scores_f1_golden = []


for train_index, test_index in kf.split(X_golden, y):
    X_train, X_test = X_golden[train_index], X_golden[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    scores_accuracy_golden.append(accuracy)
    scores_f1_golden.append(f1)

print("Accuracy scores :")
print(scores_accuracy_golden)
print("F1 scores:")
print(scores_f1_golden)

average_accuracy_golden = sum(scores_accuracy_golden) / len(scores_accuracy_golden)
average_f1_golden = sum(scores_f1_golden) / len(scores_f1_golden)

print("\nAverage Accuracy with Golden Feature:", average_accuracy_golden)
print("Average F1 Score with Golden Feature:", average_f1_golden)
print("\nAverage Accuracy with Full Model:", average_accuracy)
print("Average F1 Score with Full Model:", average_f1)
```
