```python
# Task 1: Write your code here:
# Drop 'Order_ID' column
df = df.drop('Order_ID', axis=1)
df.head()
```

```python
# Task 2: Write your code here:
# Courier_Experience_yrs fill nan with zero
# Check missing values
missing_values = df.isnull().sum()
print("Columns with missing values:")
print(missing_values[missing_values > 0])

# Fill missing values with mean for each column
df['Courier_Experience_yrs'].fillna(df['Courier_Experience_yrs'].mean(),inplace=True)
df['Weather'].fillna(df["Weather"].mode()[0], inplace=True)
df['Delivery_Time'].dropna()
```

```python
# Task 3: Write your code here:
def check_duplicates(df):
  duplicates = df.duplicated().sum() # num of duplicated rows
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    # removes repeated rows, inplace=True → modify the df, default is to remove first occurance
    # because we dont want the model to learn redundent data
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df)
```

```python
# Task 4: Write your code here:
#Encode categorical variables
categorical_cols = df.select_dtypes(include=["object"]).columns # takes column that include label

print("Categorical Columns:", list(categorical_cols))
for col in categorical_cols:
    label_encoder = LabelEncoder() # define object
    df[col] = label_encoder.fit_transform(df[col]) # fit and transform
df.head()
```

```python
# Task 5: Write your code here:
# Apply feature scaling for all features (Use StandardScaler)
features_to_scale = ["Distance_km",'Weather',	'Traffic_Level',	'Time_of_Day',	'Vehicle_Type',	'Preparation_Time_min',	'Courier_Experience_yrs']

print("Before scaling:")
print(df[features_to_scale].describe())

scaler = StandardScaler()
df[features_to_scale] = scaler.fit_transform(df[features_to_scale])
```

```python
# Task 6: Write your code here:
import matplotlib.pyplot as plt
# Task 5: Write your code here: target disturbtion
plt.figure(figsize=(10, 6))
plt.hist(df['Delivery_Time'], bins=60, edgecolor='black')
plt.title('Delivery_Time')
plt.ylabel('time')
plt.show()
# its balanced
```

```python
# Task 1: Write your code here:
X=df.drop(['Delivery_Time'], axis=1)
y=df["Delivery_Time"].dropna().astype(float)
#X.head(10)
```

```python
# Task 2,3,4,5: Write your code here:
kf = KFold(n_splits=5, shuffle=True, random_state=42)

for fold, (train_index, test_index) in enumerate(kf.split(X)):
    print(f"\nFold {fold+ 1}/{5}")
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

    model = RandomForestRegressor( n_estimators=200, max_depth=None, random_state=42, n_jobs=-1 )

    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    MAE=mean_absolute_error(y_test,y_pred)
```
