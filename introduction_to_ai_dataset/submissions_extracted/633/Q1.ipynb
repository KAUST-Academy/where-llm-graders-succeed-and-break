```python
# all missing values are object type so i will fill it with unknown

for col in ['Weather', 'Traffic_Level', 'Time_of_Day', 'Courier_Experience_yrs']:
    df[col] = df[col].fillna('unknown')

# Drop rows where Delivery_Time are missing
df = df.dropna(subset=['Delivery_Time']) #drop rows with NaN
```

```python
print("Missing values remaining:", df.isnull().sum().sum())
```

```python
# Task 3: Write your code here:
def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df)
```

```python
# Task 4: Write your code here:
categorical_cols = ['Weather', 'Traffic_Level', 'Time_of_Day', 'Vehicle_Type','Courier_Experience_yrs']

for col in categorical_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col].astype(str))

df.head()


```

```python
df.info()
```

```python
# Task 5: Write your code here:

features = df.columns.drop("Delivery_Time")  # DON'T SCALE THE TARGET
standard_scaler = StandardScaler()

df[features] = standard_scaler.fit_transform(df[features])
df.head()

```

```python
# Task 6: Write your code here:
# we trying to predict the Delivery_Time and it not class so we dont need to know if it balance or not
```

```python
# Task 1: Write your code here:

X = df.drop("Delivery_Time", axis=1).astype(float)
y = df['Delivery_Time'].astype(float)
```

```python
# Task 2,3,4,5: Write your code here:
```

```python
n_splits = 5  # K=5 Folds we dont care about imbalnce

# 5-Fold Cross-Validation, shuffled
kf = KFold(n_splits=n_splits, shuffle=True, random_state=42)


for fold_idx, (train_index, test_index) in enumerate(kf.split(X)):
  print(f"\nFold {fold_idx + 1}/{n_splits}")

  X_train, X_test = X.iloc[train_index], X.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]

  # Train RandomForest
  model = RandomForestRegressor()
  model.fit(X_train, y_train)


  # Predict and evaluate
  y_pred = model.predict(X_test)

  mae = mean_absolute_error(y_test, y_pred)

  print(f"MSE:  ${mae:,.2f}")
```

```python
# Task 1: Write your code here:
# Feature importance
feature_cols = X_train.columns
feature_importance = pd.DataFrame({
    'feature': feature_cols,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)
plt.figure(figsize=(10, 6))
plt.barh(feature_importance['feature'], feature_importance['importance'])
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show()
```

```python
# Task 2: Write your code here:
# Plot Predictions vs Ground Truth
plt.figure(figsize=(6, 4))
plt.scatter(y_test, y_pred, alpha=0.6)
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    "r--",
    linewidth=2
)

plt.xlabel("(Ground Truth)")
plt.ylabel("predicted delivery time ")
plt.grid(True)

plt.tight_layout()
plt.show()
```
