```python
# Task 1: Write your code here:
df_clean = df.drop(columns="Order_ID").copy()
df_clean.info()
```

```python
# Task 2: Write your code here:
stat_cols = df_clean.columns.to_list()
df_clean = df_clean.dropna(subset=stat_cols).copy()
df_clean.info()
```

```python
# Task 3: Write your code here:
def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df_clean)
```

```python
# Task 4: Write your code here:
cat_data = df_clean.select_dtypes(include = ['object']).columns

df_clean_encode = pd.get_dummies(df_clean, columns=cat_data, drop_first=True)

df_clean_encode = df_clean_encode.astype(float)

df_clean_encode.info()
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler

df_scaled = df_clean_encode.drop('Delivery_Time', axis = 1)
cols = ['Distance_km','Preparation_Time_min','Courier_Experience_yrs']
for i in cols:
  scaler = StandardScaler()
  df_scaled[i] = scaler.fit_transform(df_scaled[[i]])

df_scaled.head()
```

```python
# Task 6: Write your code here:
```

```python
# Task 1: Write your code here:

X = df_scaled
y = df_clean_encode['Delivery_Time']

```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import StratifiedKFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

errors = []

kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
for train_index, test_index in kf.split(X, y):
  model = RandomForestRegressor()

  X_train, X_test = X.iloc[train_index], X.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]

  model.fit(X_train, y_train)
  y_pred = model.predict(X_test)

  mae = mean_absolute_error(y_test, y_pred)
  print(mae)
  errors.append(mae)
```

```python
# Task 1: Write your code here:
importance = pd.DataFrame({
    'feature': X.columns,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

plt.figure(figsize=(10, 6))
plt.barh(importance['feature'], importance['importance'], color='purple')
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show()
```
