```python
# Task 1: Write your code here:
df = df.drop(columns = ['Order_ID'])
df.info()
```

```python
# Task 2: Write your code here:
df2 = df.dropna(subset=['Weather','Traffic_Level','Time_of_Day','Courier_Experience_yrs','Delivery_Time']).copy()
df2.info()
```

```python
# Task 3: Write your code here:
duplicates = df2.duplicated().sum()
print(f"Number of Duplicate Samples: {duplicates}")
if duplicates > 0:
  print("Dropping Duplicates...")
  df2.drop_duplicates(inplace=True)
  print("Duplicates Dropped.")
else:
    print("No Duplicate.")
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import OneHotEncoder

onehot_encoder = OneHotEncoder(sparse_output=False)
dfcols= df.select_dtypes(include=["object"]).columns
df3 = onehot_encoder.fit_transform(df2[dfcols]).copy()
df3
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler
ss = StandardScaler()
df3witouttarget = df3.drop(columns =['Delivery_Time']).copy()
data_standard_scaled = ss.fit_transform(df3witouttarget).copy()
data_standard_scaled
```
