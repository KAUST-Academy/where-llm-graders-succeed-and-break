```python
# Task 4: Write your code here:
categorical_cols = df_clean.select_dtypes(include=["object"]).columns

print("Categorical Columns:", list(categorical_cols))
```

```python
# I used get dummies for one hot encoder
df_clean = pd.get_dummies(df_clean, columns=categorical_cols, drop_first=True)
```

```python
# Task 5: Write your code here:
# Apply feature scaling for all features (Use StandardScaler)
scaler = StandardScaler()
df_clean_scaled = pd.DataFrame(scaler.fit_transform(df_clean), columns=df_clean.columns)
```

```python
# Task 6: Write your code here:
# The target is numbers as log oredictions
```

```python
# Task 1: Write your code here:
# Split the dataset into features (X) and target (y)
X = df_clean_scaled.drop('Delivery_Time', axis=1)
y = df_clean_scaled['Delivery_Time']
```

```python
# Task 2,3,4,5: Write your code here:
model = RandomForestRegressor()
kf = KFold(n_splits=5, shuffle=True, random_state=42)
mae_scores = []
for train_index, test_index in kf.split(X):
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    mae = mean_absolute_error(y_test, y_pred)
    mae_scores.append(mae)

average_mae = np.mean(mae_scores)
print("MAE scores for each fold:", mae_scores)
print("Average MAE:", average_mae)
```
