```python
# Task 1: Write your code here:
X = df.drop('Target',axis=1)
y = df['Target']
```

```python
from IPython.display import clear_output

%pip install kagglehub catboost xgboost tqdm imbalanced-learn -q

clear_output()
```

```python
model =CatBoostClassifier(
      verbose=0,
      n_estimators=200,
      max_depth=4
  )
```

```python
# Task 2,3,4,5: Write your code here:
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scores_full_model = []
f1_scores_full_model = []
for train_index, test_index in skf.split(X, y):
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    score = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    f1_scores_full_model.append(f1)
    scores_full_model.append(score)

print("Accuracy scores:", scores_full_model)
print("Average accuracy score:", np.mean(scores_full_model))
print("F1 scores:", f1_scores_full_model)
print("Average F1 score:", np.mean(f1_scores_full_model))



```

```python
# We should use f1 above because our class is imbalance so ad we did
print(f1_scores_full_model)
print(np.mean(f1_scores_full_model))
```
