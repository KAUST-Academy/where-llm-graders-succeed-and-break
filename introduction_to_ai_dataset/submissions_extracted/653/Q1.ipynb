```python
# Task 1: Write your code here:
data_cleaned = data.copy()
data_cleaned.drop(columns="Order_ID")
len(data_cleaned)
```

```python
# Task 2: Write your code here:

# i will check the sum of the null values for each column
print(data_cleaned.isnull().sum())

colsWithNull = ["Weather", "Traffic_Level", "Time_of_Day", "Courier_Experience_yrs", "Delivery_Time"]
# will have a look at the weather column
#print(data_cleaned[colsWithNull].head(50))
data_cleaned = data_cleaned.dropna(subset=colsWithNull)
```

```python
# Task 3: Write your code here:
duplicates = data_cleaned.duplicated().sum()
print("Number of duplicates that exist in the samples", duplicates)

if duplicates > 0:
    print("Dropping Duplicates...")
    data_cleaned.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
else:
    print("No Duplicate Samples Found.")
```

```python
from sklearn.preprocessing import LabelEncoder
# Task 4: Write your code here:

# checking how oneHotEncoder will deal with it

# we encode any categorical columns so that the model understand
categorical_cols = data_cleaned.select_dtypes(include=["object"]).columns
encode = LabelEncoder()
for col in categorical_cols:
  data_cleaned[col] = encode.fit_transform(data_cleaned[col])
  print(col)
# checking if we changed the categoris into numbers
print(data_cleaned.head(20))
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler
features = ["Weather", "Traffic_Level", "Time_of_Day", "Courier_Experience_yrs", "Distance_km", "Vehicle_Type", "Preparation_Time_min"]
# will scale all the features into a specified range so that the model does not treat a specific feature better than other
scale = StandardScaler()
data_scaled = scale.fit_transform(data_cleaned[features])

data_scaled = pd.DataFrame(data_scaled, columns=features)
print(data_scaled)
```

```python
# Task 6: Write your code here:
# there will be no imbalance since it is a linear regression problem. If it was a classification then an imbalance might appear
data_cleaned["Delivery_Time"].value_counts()
data_cleaned['Delivery_Time']
```

```python
# Task 1: Write your code here:



x = data_scaled[features]
y = data_cleaned["Delivery_Time"]

print(x.head())
y
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

kf = KFold(n_splits=5, shuffle=True, random_state=42)
model = RandomForestRegressor()
for fold_idx, (train_index, test_index) in enumerate(kf.split(x)):
  print(f"\nFold {(fold_idx + 1) }")

# Splitting the data
  x_train, x_test = x.iloc[train_index], x.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]

  # Train
  model.fit(x_train, y_train)
  y_pred = model.predict(x_test)

  # Calculate evaluation metrics
  mae = mean_absolute_error(y_test, y_pred)
print("Mean Absolute Value", mae)
```
