```python
# Task 1: Write your code here:
df_delv.drop(columns=['Order_ID'], inplace=True)
```

```python
cols = ['Courier_Experience_yrs', 'Delivery_Time', 'Weather', 'Traffic_Level', 'Time_of_Day','Distance_km','Vehicle_Type','Preparation_Time_min']

df_clean = df_delv[cols].copy()
df_clean
```

```python
 # Task 2: Write your code here:

 df_clean['Courier_Experience_yrs'].fillna(df_clean['Courier_Experience_yrs'].mean(), inplace=True)

 df_clean['Delivery_Time'].fillna(df_clean['Delivery_Time'].mean(), inplace=True)


 df_clean['Weather'].fillna(df_clean['Weather'].mode()[0], inplace=True)
 df_clean['Traffic_Level'].fillna(df_clean['Traffic_Level'].mode()[0], inplace=True)
 df_clean['Time_of_Day'].fillna(df_clean['Time_of_Day'].mode()[0], inplace=True)
 df_clean
```

```python
# Task 3: Write your code here:
#duplicates = df_clean.duplicated().sum()
#print(f"Duplicates: {duplicates}")

def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df_clean)

df_clean
```
