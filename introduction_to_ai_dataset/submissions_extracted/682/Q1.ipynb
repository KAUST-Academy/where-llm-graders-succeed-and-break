```python
# Task 1: Write your code here:
df = df.drop(columns = ["Order_ID"])

display(df)
```

```python
# Task 2: Write your code here:

# Source: Day 1 EDA
def checkMissingValues():
  missing_percentage = (df.isna().sum() / len(df)) * 100
  missing_data = pd.DataFrame({
      'Column': missing_percentage.index,
      'Missing_Percentage': missing_percentage.values
  })

  missing_data = missing_data[missing_data['Missing_Percentage'] > 0].sort_values('Missing_Percentage', ascending=False)

  display(missing_data)

checkMissingValues()
# Replacing NA Values
cols = ["Weather", "Traffic_Level", "Time_of_Day"]
for col in cols:
  df[col] = df[col].fillna("Unknown")
df['Delivery_Time'] = df['Delivery_Time'].fillna(0)
df['Delivery_Time'] = df['Delivery_Time'].dropna()
df['Courier_Experience_yrs'] = df['Courier_Experience_yrs'].fillna(df['Courier_Experience_yrs'].median())

checkMissingValues()
# Good, no missing values
```

```python
# Task 3: Write your code here:

# Source: Day 2
duplicate_rows = df.duplicated().sum()
if duplicate_rows > 0:
    print("Deleting Duplicate")
    df.drop_duplicates(inplace=True)
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import StandardScaler, LabelEncoder, OneHotEncoder

cat_cols = df.select_dtypes(include = "object")

clean_df = df.copy()
for col in cat_cols.columns:
  le = LabelEncoder()
  clean_df[col] = le.fit_transform(clean_df[col])

#ohe = OneHotEncoder(sparse_output=False)
#cat_cols_encoded = pd.DataFrame(ohe.fit_transform(cat_cols), columns= ohe.get_feature_names_out(cat_cols.columns))

#df.join()

display(clean_df)
```

```python
# Task 5: Write your code here:

sc = StandardScaler()
df[clean_df.columns.drop("Delivery_Time")] = sc.fit_transform(clean_df[clean_df.columns.drop("Delivery_Time")])
```
