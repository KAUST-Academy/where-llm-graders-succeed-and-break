```python
# Task 1: Write your code here:
cols = ['Order_ID', 'Distance_km', 'Weather', 'Traffic_Level',	'Time_of_Day',	'Vehicle_Type'
        ,'Preparation_Time_min'	,'Courier_Experience_yrs'	,'Delivery_Time']

df_clean = df_food[cols].copy()
print(f"Before: {df_clean.shape}")
df_clean = df_clean.drop(columns=['Order_ID'])

print(f"After dropping missing Order_ID: {df_clean.shape}")
```

```python
# Task 2: Write your code here:
def check_missing_values(df):
  missing_values = df.isnull().sum()
  print("Missing Values per Column:")
  print(missing_values[missing_values > 0])
  if missing_values.any():
    print("\nHandle Missing Values as needed.")
  else:
    print("\nNo Missing Values Found.")


for col in ['Weather', 'Traffic_Level', 'Time_of_Day', 'Courier_Experience_yrs' , 'Delivery_Time']:
    df_clean[col] = df_clean[col].fillna('unknown')

df_clean['Delivery_Time'] = df_clean['Delivery_Time'].fillna(df_clean['Delivery_Time'].mode()[0])


print("Missing values remaining:", df_clean.isnull().sum().sum())
check_missing_values(df_clean)

df_clean.info()
```

```python
# Task 3: Write your code here:
def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df_clean)
```

```python
# Task 4: Write your code here:
categorical_cols = ['Weather', 'Traffic_Level', 'Time_of_Day', 'Vehicle_Type']
for col in categorical_cols:
    le = LabelEncoder()
    df_clean[col] = le.fit_transform(df_clean[col].astype(str))
df_clean["Delivery_Time"]= df_clean["Delivery_Time"].astype(str)
```
