```python
# Task 1: Write your code here:

df=df.drop(columns='Order_ID')
```

```python
# Task 2: Write your code here:

stat_cols=['Distance_km','Preparation_Time_min','Courier_Experience_yrs','Delivery_Time','Time_of_Day','Traffic_Level','Weather']


df_c = df.dropna(subset=stat_cols)
print(f"Shape after cleaning: {df_c.shape}")

print("Missing values:")
print(df_c.isnull().sum())
```

```python
# Task 3: Write your code here:

def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df_c)
df_c.drop_duplicates(inplace=True)
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import LabelEncoder

categ_feats = df_c.select_dtypes("object").columns
for cat in categ_feats:
  le = LabelEncoder()
  df_c[cat] = le.fit_transform(df_c[cat])
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler

print('data before scaling:\n', df_c) #show before scaling
standard_scaler = StandardScaler() # Instantiate StandardScaler
df_c = standard_scaler.fit_transform(df_c) # Apply fit_transform

print('\nData after scaling:\n', df_c) #show after scaling
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import mean_absolute_error
import numpy as np

# Use previously generated random data (example: regression data)
X, y = X, y

# Define K-Fold Cross Validation
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# Iterate through folds
for fold, (train_idx, test_idx) in enumerate(kf.split(X), start=1):
    # indexing for each fold
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
    # print shapes
    print(f"Fold {fold}")
    print("  X_train shape:", X_train.shape)
    print("  X_test shape :", X_test.shape)
    print("  y_train shape:", y_train.shape)
    print("  y_test shape :", y_test.shape)
    print("-" * 30)

model = RandomForestClassifier(n_estimators=100, max_depth=15,
                               class_weight='balanced', random_state=42)

model.fit(X_train, y_train)
print("Model trained!")

y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
print('mae =',mae)

kfold = KFold(n_splits=5, shuffle=True, random_state=42)

mae_scores = []
rmse_scores = []

```
