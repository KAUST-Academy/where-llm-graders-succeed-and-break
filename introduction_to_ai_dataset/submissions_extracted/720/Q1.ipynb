```python
# Task 1: Write your code here:

#droping the order_id column

df.drop(columns=['Order_ID'])
```

```python
# Task 2: Write your code here:

#Handle missing values

#The weather column using the mode
df['Weather']= df['Weather'].fillna(df['Weather'].mode()[0])

#The traffic level column using the mode
df['Traffic_Level']= df['Traffic_Level'].fillna(df['Traffic_Level'].mode()[0])

#The time of day using mode
df['Time_of_Day']= df['Time_of_Day'].fillna(df['Time_of_Day'].mode()[0])

#The experince years using mean
df['Courier_Experience_yrs']= df['Courier_Experience_yrs'].fillna(df['Courier_Experience_yrs'].mean())

#The delivrey time using mean
df['Delivery_Time']= df['Delivery_Time'].fillna(df['Delivery_Time'].mean())

#data info after handliing missing values
df.info()
```

```python
# Task 3: Write your code here:

#Chech for duplicates

def check_duplicates(df):

  #TODO: get duplicated data using pandas
  duplicates = df.duplicated().sum()

  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df)
```

```python
# Task 4: Write your code here:

#Encoding the (Weather, Traffic_Level,	Time_of_Day,	Vehicle_Type) using OneHotEncoder
from sklearn.preprocessing import LabelEncoder

LE = LabelEncoder()

df['Weather'] = LE.fit_transform(df['Weather'])
df['Traffic_Level'] = LE.fit_transform(df['Traffic_Level'])
df['Time_of_Day'] = LE.fit_transform(df['Time_of_Day'])
df['Vehicle_Type'] = LE.fit_transform(df['Vehicle_Type'])

```

```python
# Task 5: Write your code here:

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

numerical_cols = df.select_dtypes(include=["number"]).columns.drop("Delivery_Time")


df[numerical_cols] = scaler.fit_transform(df[numerical_cols])

df.head()
```
