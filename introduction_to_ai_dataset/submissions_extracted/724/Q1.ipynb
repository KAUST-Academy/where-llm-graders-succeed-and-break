```python
data_clean= data.copy()
```

```python
# Task 1: Write your code here:
data_clean.drop('Order_ID', axis=1, inplace=True)
```

```python
# Task 2: Write your code here: Handle missing values appropriately
missing_values = data_clean.isnull().sum()
missing_percent = (missing_values / len(data_clean)) * 100

missing_df = pd.DataFrame({
    'Missing_Count': missing_values,
    'Percentage': missing_percent
})
missing_df = missing_df[missing_df['Missing_Count'] > 0].sort_values('Missing_Count', ascending=False)

print("Missing values summary:")
print(missing_df)

plt.figure(figsize=(10, 5))
plt.bar(missing_df.index, missing_df['Percentage'])
plt.xlabel('Columns')
plt.ylabel('Missing Percentage (%)')
plt.title('Missing Values in  Dataset')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3, axis='y')
plt.show()
```

```python
# I hindel miising valu by take the median for float colmn and fill string colmns by not defiend

data_clean['Delivery_Time'].fillna(data_clean['Delivery_Time'].median(), inplace=True)
data_clean['Weather'].dropna()
data_clean['Time_of_Day'].dropna()
data_clean['Courier_Experience_yrs'].dropna()
```

```python
# Task 3: Write your code here:Check and remove duplicates if any exist

duplicates = data_clean.duplicated().sum()
print(f"Number of duplicate rows: {duplicates}")

if duplicates > 0:
    data_clean.drop_duplicates(inplace=True)
    print(f"Removed {duplicates} duplicate rows")
else:
    print("No duplicates found!")
```

```python
# Task 4: Write your code here: Encode categorical variables if needed (Bonus if used One Hot Encoding)

categorical_cols = ['Weather', 'Traffic_Level', 'Time_of_Day', 'Vehicle_Type']
for col in categorical_cols:
    le = LabelEncoder()
    data_clean[col] = le.fit_transform(data_clean[col].astype(str))

data_clean.head()
```

```python
# Task 5: Write your code here:Apply feature scaling for all features (Use StandardScaler)
print('data before scaling:\n', data_clean) #show before scaling
standard_scaler = StandardScaler() # Instantiate StandardScaler
data_standard_scaled = standard_scaler.fit_transform(data_clean) # Apply fit_transform

print('\nData after scaling:\n', data_standard_scaled) #show after scaling
```

```python
# Modeling
X = data_clean.drop("Delivery_Time",axis=1)
y = data_clean['Delivery_Time']
```

```python
from sklearn.model_selection import StratifiedKFold
# Define Stratified K-Fold
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

print("Stratified K-Fold Cross Validation\n" + "-"*40)

for fold, (train_idx, test_idx) in enumerate(skf.split(X, y), start=1):
    # indexing for each fold
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    print(f"Fold {fold}")
    print("  X_train shape:", X_train.shape)
    print("  X_test shape :", X_test.shape)
    print("  y_train shape:", y_train.shape)
    print("  y_test shape :", y_test.shape)
```
