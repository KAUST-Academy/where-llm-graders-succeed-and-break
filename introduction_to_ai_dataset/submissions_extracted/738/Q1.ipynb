```python
# Task 1: Write your code here:
df = df.drop('Order_ID', axis=1)
df.head()
```

```python
# Task 2: Write your code here:
df_no_na = df.dropna()
df_no_na.info()
```

```python
# Task 3: Write your code here:
print(f"Duplicate values before drop: {df_no_na.duplicated().sum()}")
df_clean = df_no_na.drop_duplicates()
print(f"Duplicate values before drop: {df_clean.duplicated().sum()}")
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import LabelEncoder

cat_columns = df_clean.select_dtypes('object').columns

le = LabelEncoder()
for col in cat_columns:
  df_clean[col] = le.fit_transform(df_clean[col])
df_clean.head()
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df_scaled = pd.DataFrame(scaler.fit_transform(df_clean), columns=df_clean.columns)
df_scaled.head()
```

```python
# Task 6: Write your code here:
# there is no need to check for target imbalance as target is a value and not a class
```

```python
# Task 1: Write your code here:
X = df_scaled.drop('Delivery_Time', axis=1)
y = df_scaled
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
import numpy as np


kf = KFold(n_splits=5, shuffle=True, random_state=42)

mae_scores = []
rmse_scores = []

model = RandomForestRegressor(n_estimators=100, max_depth=20, random_state=42, n_jobs=-1)

for train_idx, test_idx in kf.split(X):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    # Train and predict
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    # Calculate metrics
    mae_scores.append(mean_absolute_error(y_test, y_pred))

mae_scores = np.array(mae_scores)

print(f"5-Fold CV Results:")
print(f"MAE:  {mae_scores.mean():,.2f}")
```

```python
# Task 1: Write your code here:
importance= model.feature_importances_

fig, axes = plt.subplots(figsize=(18, 6))
features = df.drop('Delivery_Time', axis=1).columns

# Sort features by importance for a cleaner plot
sorted_idx = np.argsort(importance)[len(importance)-5:]

ax = axes
ax.barh(features[sorted_idx], importance[sorted_idx])
ax.set_title(f"Catboost Feature Importance")
ax.set_xlabel("Importance Score")

plt.tight_layout()
plt.show()
```
