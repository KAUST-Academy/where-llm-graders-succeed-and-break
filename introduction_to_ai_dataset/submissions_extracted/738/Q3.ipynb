```python
# Task 1: Write your code here:
X = df_scaled
y = df['Target']
```

```python

# Task 2,3,4,5: Write your code here:
from sklearn.metrics import accuracy_score, f1_score

kf = KFold(n_splits=5, shuffle=True, random_state=42)

mae_scores = []
rmse_scores = []

model = CatBoostClassifier(
      verbose=0,
      n_estimators=320,
      max_depth=4
  )

# used a dict to store scores
scores ={
    'Accuracy': [],
    "F1-score": []
}

for train_idx, test_idx in kf.split(X):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    # Train and predict
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    # Calculate metrics
    accuracy = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred, zero_division=0)
    scores['Accuracy'].append(accuracy)
    scores['F1-score'].append(f1)

print(f"Accuracy:  {np.array(scores['Accuracy']).mean()}")
print(f"F1-score:  {np.array(scores['F1-score']).mean()}")
```

```python
# Task 1: Write your code here:
# Gather importances from the models (from the last fold)

importance= model.feature_importances_

fig, axes = plt.subplots(figsize=(18, 6))
features = df.drop('Target', axis=1).columns

# Sort features by importance for a cleaner plot
sorted_idx = np.argsort(importance)[len(importance)-5:]

ax = axes
ax.barh(features[sorted_idx], importance[sorted_idx])
ax.set_title(f"Catboost Feature Importance")
ax.set_xlabel("Importance Score")

plt.tight_layout()
plt.show()
```
