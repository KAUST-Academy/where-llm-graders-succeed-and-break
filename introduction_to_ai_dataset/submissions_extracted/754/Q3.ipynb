```python
# Task 1: Write your code here:

cols = ['P_2','D_39',	'B_1','B_2','R_1','S_3','D_41','B_3','D_42','D_43','D_137','D_138','D_139','D_140','D_141','D_142','D_143','D_144','D_145','Target']
df_clean = df[cols].copy()

missing_values = df.isnull().sum()
print(missing_values)

df_clean['P_2'] = df_clean['P_2'].fillna(df_clean['P_2'].mode()[0])
df_clean['B_2'] = df_clean['B_2'].fillna(df_clean['B_2'].mode()[0])
df_clean['D_142'] = df_clean['D_142'].fillna(df_clean['D_142'].mode()[0])
df_clean['D_143'] = df_clean['D_143'].fillna(df_clean['D_143'].mode()[0])
df_clean['D_144'] = df_clean['D_144'].fillna(df_clean['D_144'].mode()[0])
df_clean['D_145'] = df_clean['D_145'].fillna(df_clean['D_145'].mode()[0])

print("Missing values remaining:", df_clean.isnull().sum().sum())

```

```python
# Task 2: Write your code here:

print("Checking for duplicate rows...")
duplicate_rows = df.duplicated().sum()
if duplicate_rows > 0:
    print(f"Found {duplicate_rows} duplicate rows. Removing them...")
    df.drop_duplicates(inplace=True)
    print("Duplicate rows removed.")
else:
    print("No duplicate rows found.")
```
