```python
# Task 1: Write your code here:
def check_missing_values(df):

  # Get missing values using pandas
  missing_values = df.isnull().sum()

  print("Missing Values per Column:")
  print(missing_values[missing_values > 0])

  if missing_values.any():
    print("\nHandle Missing Values as needed.")
  else:
    print("\nNo Missing Values Found.")

check_missing_values(df)
df_cleaned = df.fillna(df.mean())
check_missing_values(df_cleaned)
```

```python
# Task 2: Write your code here:
def check_duplicates(df):


  duplicates = df.duplicated().sum()

  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df_cleaned)
df_cleaned.head()
```

```python
# Task 3: Write your code here: Data is int and float
categorical_cols = df_cleaned.select_dtypes(include=["object"]).columns

print("Categorical Columns:", list(categorical_cols))
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import StandardScaler

# Pick only the numerical columns, NOT the target
numerical_cols = df_cleaned.select_dtypes(include=["number"]).columns.drop("Target")

scaler = StandardScaler()

# scale the `numerical_cols`
df_cleaned[numerical_cols] = scaler.fit_transform(df[numerical_cols])

df_cleaned.head()
```

```python
# Task 5: Write your code here:
# 1. Is the target imbalanced?
def check_target_imbalance(df, target_column):
  print("Target Distribution:")

  df_cleaned[target_column].hist()  # Yeah you can just do this :)
  plt.show()

check_target_imbalance(df, "Target")
```

```python
# Task 1: Write your code here:
X = df.drop("Target", axis=1).astype(float)
y = df["Target"].astype(float)
```

```python
!pip install catboost

from sklearn.model_selection import StratifiedKFold
from catboost import CatBoostClassifier
from sklearn.metrics import accuracy_score, f1_score
import numpy as np

X = df_cleaned.drop("Target", axis=1).astype(float)
y = df_cleaned["Target"].astype(float)

# Initialize StratifiedKFold
skf = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
accuracy_scores = []
f1_scores = []

# Loop through each fold
for fold, (train_index, test_index) in enumerate(skf.split(X, y)):
    print(f"Fold {fold + 1}/{skf.n_splits}")

    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

    model = CatBoostClassifier(iterations=100,
                               learning_rate=0.1,
                               depth=6,
                               loss_function='Logloss', # For binary classification
                               eval_metric='F1',       # Evaluate with F1 score
                               random_seed=42,
                               verbose=0)

    # Train the model
    model.fit(X_train, y_train)

    # Make predictions
    y_pred = model.predict(X_test)

    # Evaluate the model
    current_accuracy = accuracy_score(y_test, y_pred)
    current_f1 = f1_score(y_test, y_pred)

    accuracy_scores.append(current_accuracy)
    f1_scores.append(current_f1)

    print(f"  Accuracy: {current_accuracy:.4f}")
    print(f"  F1 Score: {current_f1:.4f}")

# Print averaged scores
print("\n--- Averaged Scores Across All Folds ---")
print(f"Average Accuracy: {np.mean(accuracy_scores):.4f}")
print(f"Average F1 Score: {np.mean(f1_scores):.4f}")
```
