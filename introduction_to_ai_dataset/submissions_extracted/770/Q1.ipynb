```python
# Task 1: Write your code here:
df=data.drop("Order_ID",axis=1)
df
```

```python
# Task 2: Write your code here:
nan_columns=["Weather","Traffic_Level","Time_of_Day","Courier_Experience_yrs","Delivery_Time"]
df=df.dropna(subset=nan_columns,axis=0)
df
```

```python
# Task 3: Write your code here:
def check_duplicates(df):
    duplicates = df.duplicated().sum()
    print(f"Number of Duplicate Samples: {duplicates}")
    if duplicates > 0:
        print("Dropping Duplicates...")
        df.drop_duplicates(inplace=True)
        print("Duplicates Dropped.")
    else:
        print("No Duplicate Samples Found.")
check_duplicates(df)
df
```

```python
# Task 4: Write your code here:

categorical_cols = df.select_dtypes(include=["object"]).columns
for col in categorical_cols:
    print(f"Encoding column: {col}")
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
df
```

```python
# Task 5: Write your code here:

standard_scaler = StandardScaler() # Instantiate StandardScaler
data_standard_scaled = standard_scaler.fit_transform(df[df.columns.drop("Delivery_Time")]) # Apply fit_transform

pd.DataFrame(data_standard_scaled, columns=df.columns.drop("Delivery_Time"))
```

```python
# Task 6: Write your code here:
```

```python
# Task 1: Write your code here:
y=df["Delivery_Time"]#shape (-1,)
x=data_standard_scaled.copy()
x.shape
```

```python
# Task 2,3,4,5: Write your code here:
kfold = KFold(n_splits=5, shuffle=True, random_state=42)
model = RandomForestRegressor(n_estimators=100, max_depth=20, random_state=42, n_jobs=-1)

mae_scores = []
rmse_scores = []

for train_idx, val_idx in kfold.split(x):
    X_fold_train, X_fold_val = x[train_idx], x[val_idx]
    y_fold_train, y_fold_val = y.iloc[train_idx], y.iloc[val_idx]

    # Train and predict
    model.fit(X_fold_train, y_fold_train)
    y_fold_pred = model.predict(X_fold_val)

    # Calculate metrics
    mae_scores.append(mean_absolute_error(y_fold_val, y_fold_pred))
    rmse_scores.append(np.sqrt(mean_squared_error(y_fold_val, y_fold_pred)))

mae_scores = np.array(mae_scores)
rmse_scores = np.array(rmse_scores)

print(f"5-Fold CV Results:")
print(f"MAE:  ${mae_scores.mean():,.2f}")
print(f"RMSE: ${rmse_scores.mean():,.2f}")
model = RandomForestRegressor(n_estimators=100, max_depth=20, random_state=42, n_jobs=-1)
model.fit(data_standard_scaled, y)
print("Model trained!")
```

```python
# Task 1: Write your code here:
feature_importance = pd.DataFrame({
    'feature': df.columns.drop("Delivery_Time"),
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

plt.figure(figsize=(10, 6))
plt.barh(feature_importance['feature'], feature_importance['importance'])
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show()
```
