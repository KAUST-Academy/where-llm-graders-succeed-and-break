```python
# Task 1: Write your code here:
missing_values = df.isnull().sum()
missing_values_col = df.isnull().columns
print("Missing Values per Column:")
print(missing_values[missing_values > 0])
print(missing_values_col)

```

```python



for i in missing_values_col:
  if df[i].isnull().sum() > 10000:# if the missing values are more than half of the entries just drop

      df_clean = df.copy().drop(columns=i)
  else:
    df_clean = df.copy().fillna(df[i].mean())

```
