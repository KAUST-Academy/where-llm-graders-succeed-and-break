```python
# Task 2:Handle missing values appropriately (Hint: I guess you want to have a closer look at the columns with missing values :) )
def check_missing_values(df):

  # Get missing values using pandas
  missing_values = df.isnull().sum()

  print("Missing Values per Column:")
  print(missing_values[missing_values > 0])

  if missing_values.any():
    print("\nHandle Missing Values as needed.")
  else:
    print("\nNo Missing Values Found.")

check_missing_values(df)
```

```python
# handel Weather,Traffic_Level ,Time_of_Daywith the mode cuse it is catgrical valu

df['Weather'] = df['Weather'].fillna(df['Weather'].mode()[0])# catgraicl data
```

```python
df['Traffic_Level'] = df['Traffic_Level'].fillna(df['Traffic_Level'].mode()[0])# catgraicl data
```

```python
df['Time_of_Day'] = df['Time_of_Day'].fillna(df['Time_of_Day'].mode()[0])# catgraicl data
```

```python
# Courier_Experience_yrs is numric with median ro prevent effect of outlirs

df['Courier_Experience_yrs'] = df['Courier_Experience_yrs'].fillna(df['Courier_Experience_yrs'].median())# if theres outlier
```

```python
# Delivery_Time with mean

df['Delivery_Time'] = df['Delivery_Time'].fillna(df['Delivery_Time'].mean())#
```

```python
# Task 3: Check and remove duplicates if any exist
# redundet info useles ,and make compution complex
def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df)
```

```python

# Task 4:Encode categorical variables if needed (Bonus if used One Hot Encoding)
from sklearn.preprocessing import LabelEncoder

categorical_cols = df.select_dtypes(include=["object"]).columns


label_encoders = {}
for col in categorical_cols:
  le = LabelEncoder()
  df[col] = le.fit_transform(df[col])
  label_encoders[col] = le

df
```

```python
# Task 1: Split the dataset into features (X) and target (y)
X = df.drop("Delivery_Time", axis=1).astype(float)
y = df['Delivery_Time'].astype(float)
```

```python
# Task 2,3,4,5: Write your code here:
#2 use k fold becuse there is no inpalance
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error as sklearn_mse, mean_absolute_error, r2_score
from sklearn.ensemble import RandomForestRegressor

lr_mse = []

n_splits = 5  # K=5 Folds

# 5-Fold Cross-Validation, shuffled
kf = KFold(n_splits=n_splits, shuffle=True, random_state=42)


for fold_idx, (train_index, test_index) in enumerate(kf.split(X)):
  print(f"\nFold {fold_idx + 1}/{n_splits}")

  X_train, X_test = X.iloc[train_index], X.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]

  # Train
  model=RandomForestRegressor(n_estimators=200)
  model.fit(X_train, y_train)

  # Validate
  y_pred=model.predict(X_test)

  # Calculate evaluation metrics
  mse = sklearn_mse(y_test, y_pred)
  print(mse)


  # Store results
  lr_mse.append(mse)


print(f"  Average MSE: {np.mean(lr_mse):.4f}")
```
