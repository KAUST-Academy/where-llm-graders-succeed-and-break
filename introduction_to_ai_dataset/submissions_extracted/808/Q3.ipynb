```python
# Task 1: Write your code here:
missing_percentage = (df_data.isnull().sum() / len(df_data)) * 100
missing_data = pd.DataFrame({
    'Column': missing_percentage.index,
    'Missing_Percentage': missing_percentage.values
})
missing_data = missing_data[missing_data['Missing_Percentage'] > 0].sort_values('Missing_Percentage', ascending=False)

print("Missing Data Analysis:")
print(df_data.isnull().sum())

```

```python
# Task 2: Write your code here:
dublicate=df_data.duplicated().sum()
print('befor drop' ,dublicate)
if dublicate>0:
  print('drop dublicate')
  df_data.drop_duplicates(inplace=True)
  print('dublicate dropped')
else:
  print("no dublicate")
print('after drop',dublicate)
```

```python
# Task 3: Write your code here:
from sklearn.preprocessing import OneHotEncoder

print('data before encoding:\n', df_data)

onehot_encoder = OneHotEncoder(sparse_output=False)
data_onehot_encoded = onehot_encoder.fit_transform(df_data)

print('\nData after encoding:\n', data_onehot_encoded)
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import StandardScaler
features = df_data.columns.drop("Target")

scaler = StandardScaler()
df_data[features] = scaler.fit_transform(df_data[features])
df_data.head()
```

```python
# Task 5: Write your code here:
df_data["Target"].hist()
```

```python
# Task 1: Write your code here:
x=df_data.drop("Target",axis=1)
y=df_data["Target"]
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import KFold
from catboost import CatBoostClassifier
from sklearn.metrics import accuracy_score, f1_score
x, y = x.copy(), y.copy()


kf = KFold(n_splits=5, shuffle=True, random_state=42)


for fold, (train_idx, test_idx) in enumerate(kf.split(x), start=1):

    x_train, x_test = x.iloc[train_idx], x.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    print(f"Fold {fold}")
    print("  X_train shape:", x_train.shape)
    print("  X_test shape :", x_test.shape)
    print("  y_train shape:", y_train.shape)
    print("  y_test shape :", y_test.shape)
    print("-" * 30)
model=CatBoostClassifier( verbose=0,
      n_estimators=200,
      max_depth=4)
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
accuracy = accuracy_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred, zero_division=0)
```
