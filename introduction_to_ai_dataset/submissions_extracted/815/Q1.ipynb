```python
# Split the dataset into features (X) and target (y)
target = "Delivery_Time"
y = df[target].values
X = df.drop(columns=[target])
```

```python
#Use the correct split: KFold OR StratifiedKFold
#Train a RandomForest model
#Evaluate using MAE (Mean Absolute Error) ONLY
#Print the averaged score across all folds

from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
import numpy as np

n_splits = 5
kf = KFold (n_splits=n_splits, shuffle=True, random_state=42)

mae_scores = []

for train_index, test_index in kf.split(X):


    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y[train_index], y[test_index]


    model = RandomForestRegressor(random_state=42)
    model.fit(X_train, y_train)


    y_pred = model.predict(X_test)
    mae = mean_absolute_error(y_test, y_pred)
    mae_scores.append(mae)


print( np.round(mae_scores, 3))
print(round(np.mean(mae_scores), 3))
```

```python

import numpy as np
from sklearn.model_selection import KFold
from sklearn.metrics import mean_absolute_error
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
```

```python

target = "Delivery_Time"
X = df.drop(columns=[target])
y = df[target]

kfold = KFold(n_splits=5, shuffle=True, random_state=42)
```

```python

rf_mae_scores = []
lr_mae_scores = []
avg_mae_scores = []

for train_idx, test_idx in kfold.split(X):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]


    rf = RandomForestRegressor(random_state=42)
    rf.fit(X_train, y_train)
    rf_pred = rf.predict(X_test)


    lr = LinearRegression()
    lr.fit(X_train, y_train)
    lr_pred = lr.predict(X_test)


    avg_pred = (rf_pred + lr_pred) / 2


    rf_mae_scores.append(mean_absolute_error(y_test, rf_pred))
    lr_mae_scores.append(mean_absolute_error(y_test, lr_pred))
    avg_mae_scores.append(mean_absolute_error(y_test, avg_pred))
```
