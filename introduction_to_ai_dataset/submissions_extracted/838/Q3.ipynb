```python
# Task 2: Write your code here:
# 4. Do we have duplicate samples?
def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df)
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
ft = df.select_dtypes(include=["int64", "float64"]).columns.drop("Target")
df[ft] = scaler.fit_transform(df[ft])
```

```python
# Task 1: Write your code here:
X = df.drop("Target", axis=1)
y = df["Target"]
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import StratifiedKFold
from catboost import CatBoostClassifier
from sklearn.metrics import accuracy_score, f1_score
import numpy as np

model = CatBoostClassifier(verbose=0,n_estimators=320,
      max_depth=5)
folds = 5
skf = StratifiedKFold(n_splits=folds, shuffle=True, random_state=42)
accuracy = []
f1 = []
for fold_idx, (train_index, test_index) in enumerate(skf.split(X, y)):
    print(f"\nFold {fold_idx + 1}/{folds}")

    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    f = f1_score(y_test, y_pred, zero_division=0)
    f1.append(f)
    accuracy.append(acc)

print(f'accuracy:{np.mean(accuracy):.4f}')
print(f'f1:{np.mean(f1):.4f}')

```

```python
# Task 1: Write your code here:
import matplotlib.pyplot as plt

importances = model.feature_importances_

feature_importance = pd.DataFrame({
    'feature': ft,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

plt.figure(figsize=(10, 6))
plt.barh(feature_importance['feature'], feature_importance['importance'])
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show()
```
