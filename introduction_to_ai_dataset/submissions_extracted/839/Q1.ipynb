```python
# Task 1: Write your code here:
df_food_delievery = df_food_delievery.drop(columns="Order_ID", axis=1)
```

```python
# Task 2: Write your code here:
# categorical columns
categorical_cols = df_food_delievery.select_dtypes(include=["object"]).columns

# print("Categorical Columns:", list(categorical_cols))

# categorical columns
numerical_cols = df_food_delievery.select_dtypes(include=["float","int"]).columns

# print("Numerical Columns:", list(numerical_cols))

df_food_delievery= df_food_delievery.dropna(subset=categorical_cols)
df_food_delievery= df_food_delievery.dropna(subset=numerical_cols)

df_food_delievery.isnull().sum()
```

```python
# Task 3: Write your code here:
def check_duplicates(df):
  duplicates = df_food_delievery.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df_food_delievery.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df_food_delievery)
```

```python
# Task 4: Write your code here:
for col in categorical_cols:
    le = LabelEncoder()
    df_food_delievery[col] = le.fit_transform(df_food_delievery[col].astype(object))
```

```python
# Task 5: Write your code here:

from sklearn.preprocessing import MinMaxScaler

features = df_food_delievery.columns.drop("Delivery_Time")  # WE DON'T SCALE THE TARGET

scaler = StandardScaler()
df_food_delievery[features] = scaler.fit_transform(df_food_delievery[features])
df_food_delievery.head()
```

```python
# Task 1: Write your code here:
X = df_food_delievery.drop("Delivery_Time", axis=1).astype(float)
y= df_food_delievery["Delivery_Time"].astype(float)
```

```python
# Task 2,3,4,5: Write your code here:

from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error as sklearn_mse, mean_absolute_error, r2_score
from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor(n_estimators=200)
kf = KFold(n_splits=5, shuffle=True, random_state=42)
MAE =[]
n_splits = 5

for fold_idx, (train_index, test_index) in enumerate(kf.split(X)):
  print(f"\nFold {fold_idx + 1}/{n_splits}")

  X_train, X_test = X.iloc[train_index], X.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]

  # Train
  model.fit(X_train, y_train)

  # Predict
  y_pred = model.predict(X_test)


  MAE.append(mean_absolute_error(y_test, y_pred))


mae_score = np.array(MAE)
print(f"MAE:  ${mae_score.mean():,.4f}")
```

```python
# Task 1: Write your code here:
# Feature importance
feature_importance = pd.DataFrame({
    'feature': features,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

plt.figure(figsize=(10, 6))
plt.barh(feature_importance['feature'], feature_importance['importance'])
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show()
```
