```python
# Task 5:

def check_target_distribution(df, target_column):
  df[target_column].hist(bins=30, edgecolor='black')

  plt.title(f"Target Distribution ({target_column})")
  plt.xlabel(target_column)
  plt.ylabel("Frequency")
  plt.grid(False)

  plt.show()

check_target_distribution(df, "Delivery_Time")
```

```python
# Task 1:
df = df.drop(columns=['Order_ID'])
```

```python
# Task 2:
def check_missing_values(df):
  missing_values = df.isnull().sum()
  print("Missing Values per Column:")
  print(missing_values[missing_values > 0])
  if missing_values.any():
    print("\nHandle Missing Values as needed.")
  else:
    print("\nNo Missing Values Found.")

check_missing_values(df)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_data = pd.DataFrame({
    'Column': missing_percentage.index,
    'Missing_Percentage': missing_percentage.values
})
missing_data = missing_data[missing_data['Missing_Percentage'] > 0].sort_values('Missing_Percentage', ascending=False)

print("Missing Data Analysis:")
missing_data
```

```python
cols = [	'Distance_km'	, 'Weather' ,	'Traffic_Level'	, 'Time_of_Day' ,	'Vehicle_Type',	'Preparation_Time_min',	'Courier_Experience_yrs',	'Delivery_Time']
df_clean = df[cols].copy()
print(f"Before: {df_clean.shape}")

# dropping the rows where the target is missing
df_clean = df_clean.dropna(subset=['Delivery_Time'])
print(f"After dropping missing Delivery_Time : {df_clean.shape}")
# catogrical values fill with mode
cat = ['Weather',
'Traffic_Level',
  'Time_of_Day',
       'Vehicle_Type']
for col in cat:
    df_clean[col] = df_clean[col].fillna( df_clean[col].mode()[0])
#numerical values fill it with mean
df_clean['Courier_Experience_yrs'] = df_clean['Courier_Experience_yrs'].fillna(df_clean['Courier_Experience_yrs'].mean())
```

```python
print("Missing values remaining:", df_clean.isnull().sum().sum())
```

```python
# Task 3:
def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df)
```

```python
# Task 1:
X = df_clean.drop("Delivery_Time", axis=1).astype(float)
y = df_clean['Delivery_Time'].astype(float)
```

```python
# Task 2,3,4,5: Write your code here:
#2
from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestRegressor

results = []
# Define K-Fold Cross Validation
kf = KFold(n_splits=5, shuffle=True, random_state=42)
#RandomForest
model = RandomForestRegressor(n_estimators=200)
# Iterate through folds
for fold, (train_idx, test_idx) in enumerate(kf.split(X), start=1):
    # indexing for each fold
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    mae = np.mean(np.abs((y_test - y_pred)))
    results.append(mae)
print(np.mean(results))

```

```python
#RandomForest
from sklearn.metrics import  mean_absolute_error
baseline_pred = np.full_like(y, y.mean())
baseline_mae = mean_absolute_error(y, baseline_pred)
baseline_mae
```
