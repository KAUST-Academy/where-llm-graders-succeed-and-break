```python
# Task Bonus: Write your code here:

from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestRegressor
from catboost import CatBoostRegressor
from sklearn.metrics import mean_absolute_error
import numpy as np

kf = KFold(n_splits=5, shuffle=True, random_state=42)
mae_scores = []

for train_index, val_index in kf.split(X_train):
    X_t, X_v = X_train[train_index], X_train[val_index]
    y_t, y_v = y_train[train_index], y_train[val_index]

    rf_model = RandomForestRegressor(random_state=42)
    rf_model.fit(X_t, y_t)

    cb_model = CatBoostRegressor(verbose=0, random_state=42)
    cb_model.fit(X_t, y_t)

    preds_rf = rf_model.predict(X_v)
    preds_cb = cb_model.predict(X_v)

    avg_preds = (preds_rf + preds_cb) / 2

    mae = mean_absolute_error(y_v, avg_preds)
    mae_scores.append(mae)

print(f"Average Ensemble MAE: {np.mean(mae_scores):.4f}")
```
