```python
# Task 3: Write your code here:
cat_cols = df_Q3.select_dtypes(include=["object"]).columns.tolist()
print("\nCategorical columns:", cat_cols)

for col in cat_cols:
    le = LabelEncoder()
    df_clean[col] = le.fit_transform(df_clean[col].astype(str))

df_clean.head()
```
