```python
# Task 1: Write your code here:

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sklearn
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.preprocessing import StandardScaler, LabelEncoder, OneHotEncoder
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error

data = pd.read_csv(path + "/Q1_data.csv")
```

```python
# Task 1: Write your code here:

data.drop("Order_ID", axis=1, inplace=True)
data.head()
```

```python
# Task 2: Write your code here:

missing_values = data.isnull().sum()
missing_values[missing_values > 0]

data.dropna(subset=["Delivery_Time", "Weather", "Traffic_Level", "Time_of_Day", "Courier_Experience_yrs"], inplace=True)
data["Delivery_Time"].isnull().sum()

data.isnull().sum()
```

```python
# Task 3: Write your code here:

if data.duplicated().sum():
  print("Removing duplicates")
  data.drop_duplicates(inplace=True)
else:
  print("No duplicates")

data.duplicated().sum()
```

```python
# Task 4: Write your code here:

catagorical_cols = data.select_dtypes(include="object").columns

for col in catagorical_cols:
  le = LabelEncoder()
  data[col] = le.fit_transform(data[col])

data.head()
```

```python
# Task 5: Write your code here:

for col in data.columns[:-1]:
  ss = StandardScaler()
  data[col] = ss.fit_transform(pd.DataFrame(data[col]))
```

```python
# Task 6: Write your code here:

data["Delivery_Time"].hist(bins=30, edgecolor='black')
plt.xlabel("Delivery Time")
plt.ylabel("Frequency")
plt.title("Delivery Time Distribution")
plt.show()
```

```python
# Task 1: Write your code here:

X = data.drop("Delivery_Time", axis=1)
y = data["Delivery_Time"]


```

```python
# Task 2,3,4,5: Write your code here:

skf = StratifiedKFold(n_splits=5, shuffle=True)

mae_list = []
preds_list = []

for fold_idx, (train_index, test_index) in enumerate(skf.split(X,y)):
  print(f"\nFold {fold_idx + 1}/{5}")

  X_train, X_test = X.iloc[train_index], X.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]

  model = RandomForestRegressor(n_estimators=200)

  # Train
  model.fit(X_train, y_train)

  # Predict
  y_pred = model.predict(X_test)

  # Calculate metrics
  mae = mean_absolute_error(y_test, y_pred)

  # Store results
  mae_list.append(mae)
  preds_list.append(y_pred)

  print(f"Mean absolute error: {mae:.4f}")

print(f"\n\n  Average MAE:  {np.mean(mae_list):.4f}")
```

```python
# Calculate the baseline predictions (mean of the target)
baseline_pred = np.full_like(y, y.median())

# Evaluate the baseline
baseline_mae = mean_absolute_error(y, baseline_pred)

print(f"Baseline MAE (using median target): {baseline_mae:.4f}")
```

```python
# Task 1: Write your code here:

feature_cols = X.columns

feature_importance = pd.DataFrame({
    'feature': feature_cols,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

plt.figure(figsize=(10, 6))
plt.barh(feature_importance['feature'], feature_importance['importance'])
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show()
```
