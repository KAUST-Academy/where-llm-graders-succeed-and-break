```python
# Task 1: Write your code here:

import pandas as pd

df = pd.read_csv(f"{path}/Q3_data.csv")
```

```python
# Task 1: Write your code here:

# Check how many values are missing in each column
missing = df.isnull().sum()
print("Missing values found in the dataset:\n")
print(missing[missing > 0])

# Use median for numerical features as a robust strategy
df = df.fillna(df.median())

print("\nMissing values have been handled successfully.")
```

```python
# Task 2: Write your code here:


duplicate_count = df.duplicated().sum()
print(f"Duplicate rows: {duplicate_count}")

df = df.drop_duplicates()

print("Duplicates removed if any existed.")
```

```python
# Task 3: Write your code here:

cat_cols = df.select_dtypes(include=['object']).columns

if len(cat_cols) > 0:
    df = pd.get_dummies(df, columns=cat_cols, drop_first=True)
    print("Categorical features encoded.")
else:
    print("No categorical features detected.")
```

```python
# Task 4: Write your code here: # كان في cell قبلها بس انحذفت

from sklearn.preprocessing import StandardScaler

X = df.drop("target", axis=1)
y = df["target"]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Scaling completed.")
```

```python
# Task 5: Write your code here:

import matplotlib.pyplot as plt

class_counts = y.value_counts()
print("Target distribution:\n")
print(class_counts)

imbalance_ratio = class_counts[1] / class_counts[0]
print(f"\nDefault-to-NonDefault ratio: {imbalance_ratio:.2f}")

# Visual check
plt.figure(figsize=(4,3))
class_counts.plot(kind='bar', color=['#4C72B0','#DD8452'])
plt.title("Target Class Distribution")
plt.xticks(rotation=0)
plt.show()

# Comment on imbalance
if imbalance_ratio < 0.5:
    print("\nThe target variable shows clear imbalance.")
else:
    print("\nThe target variable distribution is reasonably balanced.")
```

```python
# Task 1: Write your code here:


from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
import numpy as np

# Split data
X_train, X_test, y_train, y_test = train_test_split( X_scaled, y, test_size=0.2, random_state=42)
```

```python
# Task 2,3,4,5: Write your code here:

# Train model
rf = RandomForestClassifier(n_estimators=200, random_state=42)
rf.fit(X_train, y_train)

# Feature importances
importances = rf.feature_importances_
feature_names = X.columns

# Combine feature names with their importance
feat_importance = pd.DataFrame({
    "feature": feature_names,
    "importance": importances
})

# Sort descending
feat_importance = feat_importance.sort_values(by="importance", ascending=False)

feat_importance.head(10)
```

```python
# Task 1: Write your code here:
plt.figure(figsize=(8,5))
plt.barh(feat_importance.head(10)['feature'], feat_importance.head(10)['importance'])
plt.gca().invert_yaxis()
plt.title("Top 10 Most Important Features")
plt.show()
```
