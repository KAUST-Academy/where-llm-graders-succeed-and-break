```python
# Task 1: Write your code here:
missing_percentage = df.isnull().sum() / len(df) * 100
print("Percentage of missing values per column:\n", missing_percentage.sort_values(ascending=False).head(10))

drop_columns = missing_percentage[missing_percentage > 70].index
df = df.drop(columns=drop_columns)
print(f"\nDropped {len(drop_columns)} columns with more than 70% missing values.")

numerical_cols = [col for col in numerical_cols if col in df.columns and col != 'Target']
categorical_cols = [col for col in categorical_cols if col in df.columns]

for col in numerical_cols:
    if df[col].isnull().any():
        median_val = df[col].median()
        df[col].fillna(median_val, inplace=True)
print("\nMissing values in numerical columns filled with median.")

for col in categorical_cols:
    if df[col].isnull().any():
        mode_val = df[col].mode()[0]
        df[col].fillna(mode_val, inplace=True)
print("Missing values in categorical columns filled with mode.")

remaining_missing = df.isnull().sum().sum()
if remaining_missing == 0:
    print("\nAll missing values have been handled.")
else:
    print(f"\nWarning: {remaining_missing} missing values still remain.")

```

```python
# Task 2: Write your code here:
# Task 3: Write your code here:
initial_rows = len(df)
print(f"Number of rows before removing duplicates: {initial_rows}")

duplicate_rows = df.duplicated().sum()
print(f"Number of duplicate rows found: {duplicate_rows}")

if duplicate_rows > 0:
    df.drop_duplicates(inplace=True)
    print("Duplicate rows removed.")
else:
    print("No duplicate rows found.")

final_rows = len(df)
print(f"Number of rows after removing duplicates: {final_rows}")
```

```python
# Task 3: Write your code here:
categorical_features_indices = []

for i, col in enumerate(df.columns):
    if df[col].dtype == 'object':
        df[col] = df[col].astype('category')
        print(f"Converted column '{col}' to 'category' dtype.")
    if df[col].dtype == 'category':
        categorical_features_indices.append(i)

print(f"\nIdentified {len(categorical_features_indices)} categorical features.")
print(f"Categorical feature indices: {categorical_features_indices}")
```

```python
X = df.drop('Target', axis=1)
y = df['Target']

print(f"Shape of features (X): {X.shape}")
print(f"Shape of target (y): {y.shape}")
```

```python
from sklearn.model_selection import StratifiedKFold
from catboost import CatBoostClassifier
from sklearn.metrics import f1_score

# Task 2,3,4,5: Write your code here:

# 1. Initialize StratifiedKFold
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# 2. Create an empty list to store F1 scores from each fold
f1_scores = []

# Ensure categorical_features_indices only contains indices within the range of X's columns
# cat_features_indices from previous step might include indices for 'Target' column which is now removed from X
valid_cat_features_indices = [idx for idx in categorical_features_indices if idx < X.shape[1]]

# 3. Iterate through the splits generated by StratifiedKFold
for fold, (train_index, test_index) in enumerate(skf.split(X, y)):
    print(f"\n--- Fold {fold+1}/{skf.n_splits} ---")
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

    # Initialize CatBoostClassifier
    model = CatBoostClassifier(random_state=42, verbose=0, cat_features=valid_cat_features_indices)

    # Train the CatBoostClassifier model
    model.fit(X_train, y_train)

    # Make predictions on the testing data
    y_pred = model.predict(X_test)

    # Calculate the F1 score and append it to the list
    fold_f1_score = f1_score(y_test, y_pred)
    f1_scores.append(fold_f1_score)
    print(f"F1 Score for Fold {fold+1}: {fold_f1_score:.4f}")

# Calculate the mean F1 score across all folds
average_f1_score = sum(f1_scores) / len(f1_scores)

# Print the averaged F1 score
print(f"\nAverage F1 Score across all folds: {average_f1_score:.4f}")
```
