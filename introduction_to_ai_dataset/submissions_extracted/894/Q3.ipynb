```python
import kagglehub



# Download latest version

path = kagglehub.dataset_download("mohammad2012191/q3-ka-ai-2026")



print("Path to dataset files:", path)
```

```python
import os
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, f1_score
from sklearn.ensemble import RandomForestClassifier
import matplotlib.pyplot as plt
```

```python
csv_path = os.path.join(path, "Q3_data.csv")
df = pd.read_csv(csv_path)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
for col in df.select_dtypes(include=["float64", "int64"]).columns:
    df[col] = df[col].fillna(df[col].median())

for col in df.select_dtypes(include=["object"]).columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df = df.drop_duplicates()
```

```python
df = pd.get_dummies(df, drop_first=True)
```

```python
scaler = StandardScaler()
X = df.drop(columns=["Target"])
y = df["Target"]
X = scaler.fit_transform(X)
```

```python
df["Target"].value_counts()
```

```python
X = df.drop(columns=["Target"])
y = df["Target"]
```

```python
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

scores = []

for train_idx, val_idx in skf.split(X, y):
    X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
    y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

    model = RandomForestClassifier(
        n_estimators=200,
        random_state=42
    )

    model.fit(X_train, y_train)
    y_pred = model.predict(X_val)

    score = accuracy_score(y_val, y_pred)
    scores.append(score)

print("Average Score:", np.mean(scores))
```

```python
importances = model.feature_importances_
feature_names = X.columns

plt.barh(feature_names, importances)
plt.xlabel("Feature Importance")
plt.ylabel("Features")
plt.show()
```

```python
golden_feature = feature_names[np.argmax(importances)]
print("Golden Feature:", golden_feature)
```

```python
X_golden = X[[golden_feature]]

from sklearn.model_selection import StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
import numpy as np

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

scores_golden = []

for train_idx, val_idx in skf.split(X_golden, y):
    X_train, X_val = X_golden.iloc[train_idx], X_golden.iloc[val_idx]
    y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

    model_golden = RandomForestClassifier(
        n_estimators=200,
        random_state=42
    )

    model_golden.fit(X_train, y_train)
    y_pred = model_golden.predict(X_val)

    score = accuracy_score(y_val, y_pred)
    scores_golden.append(score)

print("Average Score (Golden Feature Only):", np.mean(scores_golden))

#   من الاوت بت نعرف ان السكور جدا كويسه
```
