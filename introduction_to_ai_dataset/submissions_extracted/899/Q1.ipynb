```python
import kagglehub
import pandas as pd
import matplotlib.pyplot as plt
import os
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

# Download latest version
path = kagglehub.dataset_download("mohammad2012191/q1-ka-ai-2026")

print("Path to dataset files:", path)
```

```python
# Task 1: Write your code here:
df = df.drop('Order_ID', axis=1)
```

```python
# Task 2: Write your code here:
# Defining all columns of type object
object_cols = list(df.select_dtypes(include='object').columns)

# Filling them with mode
for col in object_cols:
  df[col] = df[col].fillna(df[col].mode()[0])

# Defining all columns of numerical type
numerical_cols = list(df.select_dtypes(include='number').columns)

# Filling them with mean
for col in numerical_cols:
  df[col] = df[col].fillna(df[col].mean())

df.info()
```

```python
# Task 3: Write your code here:
df = df.drop_duplicates()
df.shape
```

```python
# Task 4: Write your code here:
onehot_encoder = OneHotEncoder(sparse_output=False)
X = df.drop('Delivery_Time', axis=1)
y = df['Delivery_Time']

X = onehot_encoder.fit_transform(X)

X
```

```python
# Task 5: Write your code here:
standard_scaler = StandardScaler()
X = standard_scaler.fit_transform(X)
X
```

```python
# Task 2,3,4,5: Write your code here:

feature_cols = df.drop('Delivery_Time', axis=1)
importance = []
mae_scores = []

# 2. K-Folds
kf = KFold(n_splits=5, shuffle=True, random_state=42)

for fold, (train_idx, test_idx) in enumerate(kf.split(X), start=1):
  X = pd.DataFrame(X)
  y = pd.DataFrame(y)
  X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
  y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

  print(f"Fold {fold}")
  print("  X_train shape:", X_train.shape)
  print("  X_test shape :", X_test.shape)
  print("  y_train shape:", y_train.shape)
  print("  y_test shape :", y_test.shape)
  print("-" * 30)

  # 3. Random Forest
  forest = RandomForestRegressor(n_estimators=100, max_depth=20, random_state=42, n_jobs=-1)
  forest.fit(X_train, y_train)

# 4. Evaluation using MAE
  y_pred = forest.predict(X_test)
  mae = mean_absolute_error(y_test, y_pred)
  print("MAE: ", mae)
  mae_scores.append(mae)

# Part 4 Task 1
  importance.append(forest.feature_importances_)


```

```python
# Task 1: Write your code here:
feature_importance = {
    'feature': feature_cols,
    'importance': importance
}

plt.figure(figsize=(10, 6))
plt.barh(feature_importance['feature'], feature_importance['importance'])
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show()
```
