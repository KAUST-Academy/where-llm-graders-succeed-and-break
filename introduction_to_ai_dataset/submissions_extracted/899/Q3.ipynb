```python
import kagglehub
import os
import pandas as pd
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
from sklearn.model_selection import StratifiedKFold
%pip install kagglehub catboost xgboost tqdm -q
from catboost import CatBoostClassifier
from sklearn.metrics import accuracy_score, f1_score


# Download latest version

path = kagglehub.dataset_download("mohammad2012191/q3-ka-ai-2026")



print("Path to dataset files:", path)
```

```python
# Task 1: Write your code here:
data_path = os.path.join(path, "Q3_data.csv")
df = pd.read_csv(data_path)
```

```python
# Task 1: Write your code here:
# Since all values are numerical, we will fill them with mean
# Filling them with mean
for col in df.columns:
  df[col] = df[col].fillna(df[col].mean())
df.shape
```

```python
# Task 2: Write your code here:
df = df.drop_duplicates()
df.shape  # (20001, 189), so no dupliactes
```

```python
# Task 3: Write your code here:
# No categorical variables
```

```python
# Task 4: Write your code here:
X = df.drop('Target', axis=1)
y = df['Target']
std_scaler = StandardScaler()
X = std_scaler.fit_transform(X)
```

```python
# Task 5: Write your code here:
plt.hist(y)

# It's imbalanced
```

```python
# Task 2,3,4,5: Write your code here:
n_splits = 5
acc_scores = []
f1_scores = []

feature_cols = df.drop('Target', axis=1)
importance = []

skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)

for fold_idx, (train_index, test_index) in enumerate(skf.split(X, y)):
    X = pd.DataFrame(X)
    y = pd.DataFrame(y)
    print(f"\nFold {fold_idx + 1}/{n_splits}")

    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

    catboost = CatBoostClassifier(verbose=0, n_estimators=320, max_depth=4)

    catboost.fit(X_train, y_train)
    y_pred = catboost.predict(X_test)

    acc = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred, zero_division=0)
    acc_scores.append(acc)
    f1_scores.append(f1)
    importance.append(catboost.feature_importances_)


print("Accuracy Score Averagee: ", sum(acc_scores)/len(acc_scores))
print("F1 Score Average: ", sum(f1_scores)/len(acc_scores))
```

```python
# Task 1: Write your code here:
feature_importance = pd.DataFrame({
    'feature': feature_cols,
    'importance': importance
}).sort_values('importance', ascending=False)

plt.figure(figsize=(10, 6))
plt.barh(feature_importance['feature'], feature_importance['importance'])
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show()
```
