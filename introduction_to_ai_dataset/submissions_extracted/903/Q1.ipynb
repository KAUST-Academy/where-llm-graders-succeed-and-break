```python
# Task 3: Write your code here:
def check_duplicates(df_fooddataset2):
  duplicates = df_fooddataset2.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df_fooddataset2 = df_fooddataset2.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df_fooddataset2)
```
