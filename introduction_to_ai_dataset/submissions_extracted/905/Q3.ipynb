```python
# Task 2: Write your code here:
def check_missing_values(df_clean):
  missing_values = df_clean.isnull().sum()
  print("Missing Values per Column:")
  print(missing_values[missing_values > 0])
  if missing_values.any():
    print("\nHandle Missing Values as needed.")
  else:
    print("\nNo Missing Values Found.")

print(f"Before: {df_clean.shape}")
df_clean = df_clean.dropna(subset=check_missing_values(df_clean))
print(f"After dropping missing price/year/odometer: {df_clean.shape}")
```

```python
# Task 3: Write your code here:
categorical_cols = df.select_dtypes(include=["object"]).columns

print("Categorical Columns:", list(categorical_cols))
for col in categorical_cols:
    le = LabelEncoder()
    df_clean[col] = le.fit_transform(df_clean[col].astype(str))

df_clean.head()
```

```python
# Task 4: Write your code here:

# Scale features - fit on train, transform both
from sklearn.preprocessing import StandardScaler

# Pick only the numerical columns, NOT the target
numerical_cols = df.select_dtypes(include=["number"]).columns.drop("Target")

scaler = StandardScaler()

# scale the `numerical_cols`
df_clean[numerical_cols] = scaler.fit_transform(df[numerical_cols])

df_clean.head()
```
