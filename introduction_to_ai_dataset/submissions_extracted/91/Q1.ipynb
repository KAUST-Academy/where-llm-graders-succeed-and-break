```python
# Task 1: Write your code here:
df_clean=df_q1.copy()
df_clean=df_clean.drop("Order_ID", axis=1).copy()
df_clean.info()
```

```python
# Analyze missing values
missing_percentage = (df_clean.isnull().sum() / len(df_clean)) * 100
missing_data = pd.DataFrame({
    'Column': missing_percentage.index,
    'Missing_Percentage': missing_percentage.values
})
missing_data = missing_data[missing_data['Missing_Percentage'] > 0].sort_values('Missing_Percentage', ascending=False)

print("Missing Data Analysis:")
missing_data.head(10)
```

```python
# Task 2: Write your code here:


print(f"\nBefore: {df_clean.shape}")
df_clean = df_clean.dropna(subset=['Courier_Experience_yrs','Delivery_Time'])
print(f"\nAfter dropping missing 'Weather', 'Traffic_Level', 'Time_of_Day','Courier_Experience_yrs': {df_clean.shape}")

# Fill categorical columns with 'unknown' - missing likely means "not specified"
for col in ['Weather', 'Traffic_Level', 'Time_of_Day']:
    df_clean[col] = df_clean[col].fillna('unknown')


print("\nMissing values remaining:", df_clean.isnull().sum().sum())
```

```python
# Task 3: Write your code here:
dup_sum=df_clean.duplicated().sum()
print("The sum of duplicates Before :",dup_sum)
df_clean=df_clean.drop_duplicates()
dup_sum=df_clean.duplicated().sum()
print("The sum of duplicates After :",dup_sum)
```

```python
# Task 4: Write your code here:
from sklearn.model_selection import train_test_split, KFold
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Encode categorical columns - converts text to integers
                  #	'Weather',	'Traffic_Level',	'Time_of_Day',	'Vehicle_Type'
categorical_cols = ['Weather', 'Traffic_Level', 'Time_of_Day','Vehicle_Type']
for col in categorical_cols:
    le = LabelEncoder()
    df_clean[col] = le.fit_transform(df_clean[col].astype(str))

df_clean.head()
```

```python
# Task 5: Write your code here:
num_cols =['Distance_km',	'Preparation_Time_min',	'Courier_Experience_yrs']
scaler = StandardScaler()
df_clean[num_cols] = scaler.fit_transform(df_clean[num_cols])
```

```python
# Task 1: Write your code here:
feature_cols=['Distance_km','Weather',	'Traffic_Level',	'Time_of_Day',	'Vehicle_Type','Preparation_Time_min',	'Courier_Experience_yrs']
X = df_clean[feature_cols]
y = df_clean['Delivery_Time']
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error as sklearn_mse, mean_absolute_error, r2_score
from sklearn.ensemble import RandomForestRegressor
from tqdm import tqdm

n_splits = 5  # K=5 Folds

# 5-Fold Cross-Validation, shuffled
kf = KFold(n_splits=n_splits, shuffle=True, random_state=42)

models ={"Random Forest Regressor": RandomForestRegressor(n_estimators=200)}

all_results = {}

for name in models:
  all_results[name] = {'mse': [], 'rmse': [], 'r2': []}



kf = KFold(n_splits=5, shuffle=True, random_state=42)

for fold_idx, (train_index, test_index) in enumerate(kf.split(X)):
  print(f"\nFold {fold_idx + 1}/{n_splits}")

  X_train, X_test = X.iloc[train_index], X.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]

  for model_name, model in models.items():
    print(f"Training {model_name}...")

    # Train
    model.fit(X_train, y_train)

    # Predict
    y_pred = model.predict(X_test)

    # Calculate metrics
    mse = sklearn_mse(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)

    # Store results
    all_results[model_name]["mse"].append(mse)
    all_results[model_name]["rmse"].append(rmse)
    all_results[model_name]["r2"].append(r2)
```
