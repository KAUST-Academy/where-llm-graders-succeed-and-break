```python
# deleting rows and cols that has nan values (i dont think its a good idea to delete data)
df.dropna().shape #rows na deletion
df.dropna(axis=1) ## Column wise deletion

```

```python
# Instead of deleting data, imputation techniques can be used to fill in missing values, like filling the na values with the mean of the col
# that would be better instead of losing data
df['Delivery_Time_mean']=df['Delivery_Time'].fillna(df['Delivery_Time'].mean())
print(df['Delivery_Time_mean'].isnull().sum())
# now its a 0 missing values
```

```python
df['Vehicle_Type'].dropna
df['Time_of_Day'].dropna
df['Traffic_Level'].dropna
df['Weather'].dropna
```

```python
# Task 3: Write your code here:
def check_duplicates(df):
  duplicates = df.duplicated().sum()
  print(f"Number of Duplicate Samples: {duplicates}")
  if duplicates > 0:
    print("Dropping Duplicates...")
    df.drop_duplicates(inplace=True)
    print("Duplicates Dropped.")
  else:
    print("No Duplicate Samples Found.")

check_duplicates(df)
```

```python
from sklearn.preprocessing import OneHotEncoder

##create an instance of Onehotencoder
encoder=OneHotEncoder()

encoded=encoder.fit_transform(df[['Vehicle_Type']]).toarray()
encoded1=encoder.fit_transform(df[['Time_of_Day']]).toarray()

encoded2=encoder.fit_transform(df[['Traffic_Level']]).toarray()
encoded3=encoder.fit_transform(df[['Weather']]).toarray()
encoder.categories_
encoder_df=pd.DataFrame(encoded,columns=['Clear', 'Foggy', 'Rainy', 'Snowy', 'Windy'])
encoder_df
```

```python
# Task 4 :using label encodeing
from sklearn.preprocessing import LabelEncoder
categorical_cols = df.select_dtypes(include=["object"]).columns
for col in categorical_cols:
          print(f"Encoding column: {col}")
          le = LabelEncoder()
          df[col] = le.fit_transform(df[col])
df
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler

numerical_cols = df.select_dtypes(include=["int64", "float64"]).columns.drop("Delivery_Time")  # without scaling the target ofc

scaler = StandardScaler() #initialize the scaler object

df[numerical_cols] = scaler.fit_transform(df[numerical_cols]) #applying scailing

df.head()
```

```python
# Task 6: Write your code here:
def check_target_imbalance(df, target_column):
  print("Target Distribution:")
  print(df[target_column].value_counts(normalize=True))
  sns.countplot(x=df[target_column])
  plt.title("Target Distribution")
  plt.show()

check_target_imbalance(df, "Delivery_Time")
```

```python
# the target clearly imbalanced
```

```python
#handling imbalanced dataset ima use upsampling
## upsampling
df_minority=df[df['Delivery_Time']<10] #defining which target data is cosidered as minority or majoraity
df_majority=df[df['Delivery_Time']>10]


#Upsampling involves increasing the number of minority class data points to match the majority class.


from sklearn.utils import resample
df_minority_upsampled=resample(df_minority,replace=True, #Sample With replacement
         n_samples=len(df_majority),
         random_state=42
        )
```
