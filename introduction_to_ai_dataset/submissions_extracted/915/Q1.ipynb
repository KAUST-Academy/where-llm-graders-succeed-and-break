```python
# Task 1: Write your code here:
df_cleaning = df.drop('Order_ID', axis=1)
df_cleaning
```

```python
# Task 2: Write your code here:
df_cleaning.isnull().sum()
df_cleaning = df_cleaning.dropna(subset=['Weather', 'Traffic_Level', 'Time_of_Day', 'Courier_Experience_yrs', 'Delivery_Time'])
df_cleaning
```

```python
# Task 3: Write your code here:
df_cleaning.duplicated().count()

```

```python
# Remove duplicates
df_cleaning = df_cleaning.drop_duplicates()
df_cleaning
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import OneHotEncoder

categ_cols = ['Weather', 'Traffic_Level', 'Time_of_Day', 'Vehicle_Type']

for col in categ_cols:
    le = LabelEncoder()
    df_cleaning[col] = le.fit_transform(df_cleaning[col])
df_cleaning.head()
```

```python
# Task 5: Write your code here:
feature_cols = ['Distance_km', 'Weather', 'Traffic_Level', 'Time_of_Day', 'Vehicle_Type', 'Preparation_Time_min', 'Courier_Experience_yrs']
X = df_cleaning[feature_cols]
y = df_cleaning['Delivery_Time']
# Scale features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X)
X_test_scaled = scaler.fit_transform(X)
X_train_scaled
X_test_scaled
```

```python
# Task 6: Write your code here:
```

```python
# Task 1: Write your code here:
# Train-test split (80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(X_train_scaled, y, test_size=0.2, random_state=42)
```

```python
# Task 2,3,4,5: Write your code here:


from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error as sklearn_mse, mean_absolute_error, r2_score
n_splits = 5  # K=5 Folds

# 5-Fold Cross-Validation, shuffled
kf = KFold(n_splits=n_splits, shuffle=True, random_state=42)

for fold_idx, (train_index, test_index) in enumerate(kf.split(X)):
  print(f"\nFold {fold_idx + 1}/{n_splits}")

  X_train, X_test = X.iloc[train_index], X.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]

  # Train

  model = RandomForestRegressor(n_estimators=100, max_depth=20, random_state=42, n_jobs=-1)
  model.fit(X_train, y_train)
  # Validate
  y_pred = model.predict(X_test)




print("Model trained!")


mae = mean_absolute_error(y_test, y_pred)

print(f"MAE:  ${mae:,.2f}")
```

```python
# Task 1: Write your code here:# Feature importance
feature_importance = pd.DataFrame({
    'feature': feature_cols,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

plt.figure(figsize=(10, 6))
plt.barh(feature_importance['feature'], feature_importance['importance'])
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show()
```
