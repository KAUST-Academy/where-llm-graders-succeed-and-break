```python
# Task 1: Write your code here:
def check_missing_values(df):
  missing_values = df.isnull().sum()
  print("Missing Values per Column:")
  print(missing_values[missing_values > 0])
  if missing_values.any():
    print("\nHandle Missing Values as needed.")
  else:
    print("\nNo Missing Values Found.")

check_missing_values(df)
```

```python
# Analyze missing values
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_data = pd.DataFrame({
    'Column': missing_percentage.index,
    'Missing_Percentage': missing_percentage.values
})
missing_data = missing_data[missing_data['Missing_Percentage'] > 0].sort_values('Missing_Percentage', ascending=False)

print("Missing Data Analysis:")
missing_data.head(10)
```

```python
df_clean = df.dropna(axis=1)
```

```python
# Task 2:
duplicates = df_clean.duplicated().sum()
print(f" Number of duplicated Records: {duplicates}")
```

```python
# Task 3
categorical_cols = df_clean.select_dtypes(include=["object"]).columns
categorical_cols
```

```python
# Task 4
from sklearn.preprocessing import StandardScaler

features = df_clean.columns.drop("Target")  # DON'T SCALE THE TARGET
scaler = StandardScaler()
df_clean[features] = scaler.fit_transform(df_clean[features])
df_clean.head()
```

```python
# Task 5: Check for target imbalance and state if it is imbalanced or not
df_clean['Target'].hist(edgecolor='black')
```

```python
# Task 1:
X = df_clean.drop("Target",axis=1)
y = df_clean['Target']
```

```python
# Task 2,3,4,5:
# Stratified split
# StratifiedKFold
from sklearn.model_selection import KFold, StratifiedKFold, cross_val_score
from catboost import CatBoostClassifier

skfold = StratifiedKFold(n_splits =5 , shuffle=True, random_state=42)
model = CatBoostClassifier(,verbose=0,
      n_estimators=320,
      max_depth=4)
model.fit(X, y)
```

```python
scores_strat = cross_val_score(model, X, y, cv=skfold, scoring='accuracy')
print("\nStratifiedKFold scores:", scores_strat)
print(f"Mean accuracy: {scores_strat.mean():.3f}")
```

```python
# Task 1: Write your code here:
# Feature importance
import matplotlib.pyplot as plt

importance = pd.DataFrame({
    'feature': features ,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

plt.figure(figsize=(10, 6))
plt.barh(importance['feature'], importance['importance'], color='purple')
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show()
```
