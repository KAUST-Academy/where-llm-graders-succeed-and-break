```python
# 1. Drop Order_ID column
df = df.drop(columns=["Order_ID"])
# Note: Order_ID is an identifier and does not add predictive value.
```

```python
 2. Handle missing values
 num_cols=
 df.select_dtypes(include=["int64", "float64"]).columns
cat_cols=
df.select_dtypes(include=["object"]).columns
# Note: Separate numerical and categorical columns to handle missing values properly.
for col in num_cols
    df[col] = df[col].fillna(df[col].median())
# Note: Median is used for numerical features to reduce the effect of outliers.
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df=df.drop_duplicates()
#removing duplicates prevent
```

```python
# 4. Encode categorical variables (One-Hot Encoding)
df = pd.get_dummies(df, drop_first=True)
# Note: One-hot encoding converts categorical variables into numeric format
```

```python
# Task 5: Write your code here:
# 5. Feature scaling using StandardScaler
X = df.drop(columns=["Delivery_Time"])
y = df["Delivery_Time"]
# Note: Feature scaling ensures all features have the same scale.
```

```python
# Task 6: Write your code here:
# Target imbalance check (not required for regression)
# Note: Target imbalance is typically checked only for classification problems.
```

```python
import numpy as nup
from sklearn.model_selection import KFold, StratifiedKFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
x = df.drop(columns=["delivery_time"])
y = df["Dilivery_Time"]
```

```python
# Task 2,3,4,5: Write your code here:
# Task 2,3,4,5: CV split + Train RandomForest + Evaluate using MAE + Print average MAE
import numpy as np
from sklearn.model_selection import KFold, StratifiedKFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
if y.nunique() <= 10:
    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    splits = cv.split(X, y)
else:
    cv = KFold(n_splits=5, shuffle=True, random_state=42)
    splits = cv.split(X)
mae_scores = []
for train_idx, test_idx in splits:
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
    model = RandomForestRegressor(random_state=42)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    mae = mean_absolute_error(y_test, y_pred)
    mae_scores.append(mae)
print("MAE scores per fold:", np.round(mae_scores, 3))
(mae_scores), 3
```

```python
# Task 1: Write your code here:
import matplotlib.pyplot as plt
import pandas as pd
feature_importance = pd.Series(
model.feature_importances_, index=X.columns).sort_values(ascending=False)
plt.figure(figsize=(8,4))
feature_importance.plot(kind="bar")
plt.ylabel("Importance")
plt.title("Feature Importance")
plt.show()
```

```python
# Task 2: Write your code here:
y_pred = model.predict(X)

plt.figure(figsize=(6,4))
plt.hist(y_pred, bins=20)
plt.xlabel("Predicted Delivery Time")
plt.ylabel("Frequency")
plt.title("Predicted Delivery Time Distribution")
plt.show()
```
