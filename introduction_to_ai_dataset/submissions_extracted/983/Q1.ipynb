```python
# Task 1: Write your code here:
df=df.drop('Order_ID',axis=1)
```

```python
# Task 2: Write your code here:
#df.isnull().sum()
columns_with_miss=df.columns[df.isnull().any()] #here the missing val in columns
for col in columns_with_miss:
  df[col].fillna(df[col][0],inplace=True) # here i replace each missing in each columns  with the first value only
```

```python
# Task 3: Write your code here:
df.duplicated().sum()
df.drop_duplicates(inplace=True)
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import LabelEncoder

cat_cols=df.select_dtypes(include=['object']).columns

for col in cat_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col].astype(str))

```

```python
features_cols=list(df.columns[:-1]) # i need it leater for features importance without the target column
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()

df=scaler.fit_transform(df)
```

```python
# Task 1: Write your code here:
x=df[:,:-1]
y=df[:,-1]
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.model_selection import  KFold
from sklearn.metrics import mean_absolute_error
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(n_estimators=100, max_depth=20, random_state=42, n_jobs=-1)
model.fit(x, y)
pred=model.predict(x)
print("Model trained!")

mae=mean_absolute_error(pred,y)

# there is no need for kFold because its predicting continuse val there is no classes
```

```python
# Task 1: Write your code here:

feature_importance = pd.DataFrame({
    'feature': features_cols,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

plt.figure(figsize=(10, 6))
plt.barh(feature_importance['feature'], feature_importance['importance'])
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.gca().invert_yaxis()
plt.show()

# the most importange feature is distance_km
```
