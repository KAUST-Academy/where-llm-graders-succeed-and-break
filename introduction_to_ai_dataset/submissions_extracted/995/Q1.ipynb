```python
# Task 1: Write your code here:
Q1_data = Q1_data.drop("Order_ID", axis = 1)
```

```python
# Task 2: Write your code here:

print(Q1_data.info())

#Labels should be correct so drop nans
Q1_data = Q1_data[~Q1_data["Delivery_Time"].isna()]
print(len(Q1_data))

num_cols = Q1_data.select_dtypes(include = "number").columns
cat_cols = Q1_data.select_dtypes(exclude = "number").columns
for col in num_cols:
  Q1_data[col] = Q1_data[col].fillna(Q1_data[col].median())

for col in cat_cols:
  Q1_data[col] = Q1_data[col].fillna(Q1_data[col].mode()[0])
Q1_data.info()
```

```python
# Task 3: Write your code here:
Q1_data = Q1_data.drop_duplicates()
```

```python
# Task 4: Write your code here:
from sklearn.preprocessing import OneHotEncoder
import numpy as np

Q1_data = pd.get_dummies(Q1_data)
```

```python
# Task 5: Write your code here:
from sklearn.preprocessing import StandardScaler


features = Q1_data.columns.drop("Delivery_Time")
scaler = StandardScaler()
Q1_data[features] = scaler.fit_transform(Q1_data[features])
Q1_data.shape
```

```python
# Task 1: Write your code here:
X = Q1_data.drop("Delivery_Time", axis = 1)
y = Q1_data["Delivery_Time"]
```

```python
# Task 2,3,4,5: Write your code here:
from sklearn.ensemble import RandomForestRegressor

import numpy as np
from sklearn.model_selection import KFold
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error



lr_mae = []
n_splits = 5
kf = KFold(n_splits=5, shuffle=True, random_state=42)
model = RandomForestRegressor(n_estimators=300, max_depth=None, min_samples_leaf=1, random_state=42)

for fold_idx, (train_index, test_index) in enumerate(kf.split(X)):
  print(f"\nFold {fold_idx + 1}/{n_splits}")

  X_train, X_test = X.iloc[train_index], X.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]

  # Train
  model.fit(X_train, y_train)
  y_pred = model.predict(X_test)

  # Calculate evaluation metrics

  mae = mean_absolute_error(y_test, y_pred)


  # Store results
  lr_mae.append(mae)

print(f"Average MAE = {sum(lr_mae) / len(lr_mae)}")
```
